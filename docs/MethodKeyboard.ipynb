# MethodKeyboard

The working surface. An on-screen keyboard for NSC Mathematics
learners who type instead of write.

Not a calculator. Not a CAS. Not handwriting.

A South African keyboard for SA exam method - built so the memo
notation comes out the way the DBE prints it.

## Why a keyboard, not a calculator

The soul of MethodMarker is *grade the working, not the answer*.
A calculator computes. The learner needs to **type method**.

Two separate surfaces:

| Surface | What it does | When |
|---|---|---|
| **MethodKeyboard** | Types working line-by-line | Always (this notebook) |
| Scientific calculator | Evaluates numbers | Optional side panel (later, if at all) |

The keyboard is the entry point to step_grader. No keyboard, no
typed method, no way for a phone/tablet learner to use MethodMarker.
This is the pilot's hard prerequisite.

## The learner contract

**Three freedoms**

1. **Expand / collapse at will.** Some learners want the working
   area big; some want the keyboard big. They choose. Setting persists
   across questions in a sitting.
2. **No line cap.** Write as many method lines as the question
   demands. A quadratic takes 3. A Euclidean proof takes 10. Both
   are fine. The `3 steps` counter is informative, not a ceiling.
3. **Optional tab auto-switch.** When a learner enters a calculus
   drill, should the keyboard jump to the Calculus tab? Only if they
   opt in. Default off - manual tab selection.

**Four locks (non-negotiable)**

1. **One line = one method step.** Press Enter -> new line. This is
   what makes learner_lines map cleanly to step_grader.
2. **Keyboard never computes.** No `=` key that evaluates. The `=`
   key inserts ` = `.
3. **`or` is one key.** NSC memos write `x = -4 or x = 3`, never
   `x = -4, x = 3`. This single decision prevents an entire class
   of grading failures.
4. **Done submits.** Not solve. Not check. The learner types,
   the learner submits, the grader grades.

## The notation the keyboard must produce

Extracted from the memo CSVs already in the corpus. These are the
symbols the DBE actually prints - not what a textbook preference
would dictate.

**Universal (all tabs)**
`=` `or` `=>` `<=>` `therefore` `because` `;` `( )`

**Semicolon, not comma** - SA writes `A(1; 9)`, not `A(1, 9)`.
This is non-negotiable for coordinates and lists.

**Algebra-specific**
`x y a b c n k` `x^2` `x^n` `x^-1` `sqrt` `cbrt` `nth root` `+-` `< > <= >= !=`
`a/b` `|x|` `Sigma` `in` `R`

**Calculus-specific**
`dy/dx` `d/dx` `d2y/dx2` `D_x[ ]` `f'(x)` `f''(x)`
`lim` `->` `x->inf` `h->0` `inf` `e` `ln` `log` `Delta`

**No integral sign** - CAPS P1 is differentiation only. The soul spec forbids it.

**Trig-specific**
`sin cos tan` `arcsin arccos arctan`
`theta alpha beta gamma` `pi` `deg` `rad` `2theta` `half theta` `sin^2 cos^2 tan^2`

**Stats-specific**
`x-bar` `sigma` `sigma^2` `Sigma Sigma x Sigma x2` `n N` `x_i f_i`
`Q1 Q2 Q3` `min max` `P( )` `P(A)` `A|B` `union` `intersect`

**Geometry-specific**
`angle` `triangle` `parallel` `perp` `cong` `similar`
`A-hat B-hat C-hat` (angle with hat)
`arc AB` `chord` `tan` `rad`
Reason keywords insert as plain text: `vert opp angles`, `alt angles`,
`co-int angles`, `tan chord`, `angle at centre` etc.

## The six tabs

| Tab | Visible in | Purpose |
|---|---|---|
| **Algebra** | P1 default | Solve, factorise, inequality, simultaneous |
| **Calculus** | P1 | First principles, derivatives, tangents, cubic |
| **Trig** | P1 + P2 | Identities, equations, general solution |
| **Stats** | P1 + P2 | Mean, SD, regression, ogive, probability |
| **Geometry** | P2 | Euclidean, analytical circle, proofs |
| **Greek** | Fallback | alpha beta gamma theta lambda pi sigma phi omega |

**Tab auto-switch mapping** (when learner opts in):

| Skill prefix | Tab |
|---|---|
| `calculus.*` | Calculus |
| `trig.*` | Trig |
| `stats.*` / `probability.*` | Stats |
| `geometry.*` / `analytical.*` | Geometry |
| `algebra.*` / `sequences.*` / `functions.*` / `finance.*` | Algebra |

## Behaviour rules

**Line model**
- Enter -> new line, cursor at line start of the next one
- Backspace on empty line -> deletes the line, cursor back to previous
- Multi-line working area, one memo step per line
- learner_lines = ["(x+4)(x-3)=0", "x+4=0 or x-3=0", "x=-4 or x=3"]

**Insert semantics**
- Most keys insert with a leading and trailing space if surrounded
  by alphanumeric characters (prevents `x+4=0` becoming `x +4=0`)
- `or` always inserts as ` or ` (space both sides)
- `=>`, `<=>`, `therefore`, `because` insert as ` X ` (space both sides)
- `;` inserts as `; ` (trailing space only - coordinates use trailing)
- `sin(` `cos(` `tan(` - the `(` is part of the key
- `sqrt` inserts `sqrt(` - cursor inside

**Tab switching**
- Manual tap always works
- Auto-switch (when enabled): on question start, tab matches skill prefix
- Manual override during a question sticks until the next question

**Keyboard mode**
- `ABC` key toggles letter input for geometry reasons
- Letters on the ABC layer insert plain text
- Digits always visible on the number row - never hidden by ABC

**What the keyboard does NOT do**
- No `=` that evaluates
- No `solve` key
- No history / recall that recomputes
- No import from calculator

```
+-----------------------------------------+
| Algebra|Calculus|Trig|Stats|Geom|Greek  |
+-----------------------------------------+
|  x    y    a    b    c    n             |
|  =    or   =>   <=>  .:   :.            |
|  ( )  x^2  x^n  x^-1 sqrt cbrt          |
|  +/-  <    >    <=   >=   !=            |
|  a/b  |x|  Sigma ;   in   R             |
+-----------------------------------------+
|  1 2 3 4 5 6 7 8 9 0 . (-)  backspace   |
+-----------------------------------------+
|  Enter New Line   Done   ABC   Collapse |
+-----------------------------------------+
```

The `Collapse` control on the bottom-right toggles the symbol grid
off, leaving the working area, digits, and Done - for learners who
want more room to review their working before submitting.

## Output contract

The keyboard produces, on Done:

```python
learner_lines: list[str] = [
    "(x+4)(x-3)=0",
    "x+4=0 or x-3=0",
    "x=-4 or x=3",
]
```

Same shape as the CLI's `_collect_working`. Same shape `step_grader`
and `math_grader` already consume. **No interface change** - the
keyboard is just a different producer.

## Integration with the drill

The drill's existing call site:

```python
submission = _collect_working(i, SITTING_SIZE)
```

On web/mobile, `_collect_working` renders the keyboard, waits for
Done, and returns `learner_lines`-wrapped working. Same downstream
pipeline:

```
keyboard -> learner_lines -> math_grader -> step_grader -> feedback
```

Nothing else in the drill changes.

## MVP vs full build

**MVP (ship first)**
- Digits, `.`, `(-)`, `+ - x /`, `( )`, `=`
- Backspace, New Line, Done, ABC, Collapse
- **Algebra tab** (all 30 keys)
- **Trig tab** (all 24 keys)
- Line counter, no cap
- Manual tab switching only

Ships the pilot. Unblocks all quadratic / surd / inequality / trig drills.

**Full (after pilot data)**
- Calculus, Stats, Geometry, Greek tabs
- Tab auto-switch (opt-in)
- Cursor position persistence across questions
- Settings panel (persisted to learner row)

## Open questions

1. **Cursor tracking.** Does the keyboard need to support mid-line
   insertion (edit line 2 after line 3 is written)? Probably yes for
   proofs, but adds complexity. Defer to full build.
2. **Bilingual.** SA has English + Afrikaans papers. Learners may
   switch. Do we ship EN-only first, or both?
3. **Reason shortlist.** Geometry proofs have a fixed vocabulary of
   reasons. Do we offer a shortlist picker (vert opp angles, alt
   angles, co-int angles, tan chord, ...), or leave the learner to
   type? Shortlist risks over-helping; freeform risks misspellings
   the grader cannot match.
4. **Line cap on the drill side.** Even if the keyboard has no cap,
   the step grader has a fixed memo length. What happens when a
   learner writes 15 lines for a 3-line memo? Log as a discovery
   for later review.
