```python
# v1 label → v2 topic_id
# Used to translate existing 2021/2022 provisional maps during re-mapping.
v1_to_v2 = {
    "Algebra & Equations":          "ALG",
    "Number Patterns & Sequences":  "SEQ",
    "Finance":                      "FIN",
    "Functions & Graphs":           "FUNC",
    "Calculus":                     "CALC",
    "Probability":                  "PROB",
    "Statistics":                   "STAT",
    "Analytical Geometry":          "AGEO",
    "Trigonometry":                 "TRIG",
    "Euclidean Geometry":           "EUCL",
    "Mixed / Multi-topic":          "MIXED",
    "Unmapped":                     "UNMAPPED",
}

map_df = pd.DataFrame(
    [{"v1_label": k, "v2_topic_id": v} for k, v in v1_to_v2.items()]
)
map_df["taxonomy_version"] = "v2_caps"

out = TAX_DIR / "v1_to_v2_topic_map.csv"
map_df.to_csv(out, index=False)
print(map_df.to_string(index=False))
print(f"\nSaved: {out}")
```
