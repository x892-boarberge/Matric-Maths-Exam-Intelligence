````python
# Create docs/MethodKeyboard.ipynb (safe version — no nested triple quotes)

import json
from pathlib import Path

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
NB_PATH = ROOT / "docs" / "MethodKeyboard.ipynb"


def md(text):
    return {
        "cell_type": "markdown",
        "metadata": {},
        "source": text.splitlines(keepends=True),
    }


def code(text):
    return {
        "cell_type": "code",
        "metadata": {},
        "execution_count": None,
        "outputs": [],
        "source": text.splitlines(keepends=True),
    }


# ------------------------------------------------------------------
# Markdown sections — each as a plain string with \n line joins
# ------------------------------------------------------------------

SEC1 = "\n".join([
    "# MethodKeyboard",
    "",
    "The working surface. An on-screen keyboard for NSC Mathematics",
    "learners who type instead of write.",
    "",
    "Not a calculator. Not a CAS. Not handwriting.",
    "",
    "A South African keyboard for SA exam method - built so the memo",
    "notation comes out the way the DBE prints it.",
])

SEC2 = "\n".join([
    "## Why a keyboard, not a calculator",
    "",
    "The soul of MethodMarker is *grade the working, not the answer*.",
    "A calculator computes. The learner needs to **type method**.",
    "",
    "Two separate surfaces:",
    "",
    "| Surface | What it does | When |",
    "|---|---|---|",
    "| **MethodKeyboard** | Types working line-by-line | Always (this notebook) |",
    "| Scientific calculator | Evaluates numbers | Optional side panel (later, if at all) |",
    "",
    "The keyboard is the entry point to step_grader. No keyboard, no",
    "typed method, no way for a phone/tablet learner to use MethodMarker.",
    "This is the pilot's hard prerequisite.",
])

SEC3 = "\n".join([
    "## The learner contract",
    "",
    "**Three freedoms**",
    "",
    "1. **Expand / collapse at will.** Some learners want the working",
    "   area big; some want the keyboard big. They choose. Setting persists",
    "   across questions in a sitting.",
    "2. **No line cap.** Write as many method lines as the question",
    "   demands. A quadratic takes 3. A Euclidean proof takes 10. Both",
    "   are fine. The `3 steps` counter is informative, not a ceiling.",
    "3. **Optional tab auto-switch.** When a learner enters a calculus",
    "   drill, should the keyboard jump to the Calculus tab? Only if they",
    "   opt in. Default off - manual tab selection.",
    "",
    "**Four locks (non-negotiable)**",
    "",
    "1. **One line = one method step.** Press Enter -> new line. This is",
    "   what makes learner_lines map cleanly to step_grader.",
    "2. **Keyboard never computes.** No `=` key that evaluates. The `=`",
    "   key inserts ` = `.",
    "3. **`or` is one key.** NSC memos write `x = -4 or x = 3`, never",
    "   `x = -4, x = 3`. This single decision prevents an entire class",
    "   of grading failures.",
    "4. **Done submits.** Not solve. Not check. The learner types,",
    "   the learner submits, the grader grades.",
])

SEC4 = "\n".join([
    "## The notation the keyboard must produce",
    "",
    "Extracted from the memo CSVs already in the corpus. These are the",
    "symbols the DBE actually prints - not what a textbook preference",
    "would dictate.",
    "",
    "**Universal (all tabs)**",
    "`=` `or` `=>` `<=>` `therefore` `because` `;` `( )`",
    "",
    "**Semicolon, not comma** - SA writes `A(1; 9)`, not `A(1, 9)`.",
    "This is non-negotiable for coordinates and lists.",
    "",
    "**Algebra-specific**",
    "`x y a b c n k` `x^2` `x^n` `x^-1` `sqrt` `cbrt` `nth root` `+-` `< > <= >= !=`",
    "`a/b` `|x|` `Sigma` `in` `R`",
    "",
    "**Calculus-specific**",
    "`dy/dx` `d/dx` `d2y/dx2` `D_x[ ]` `f'(x)` `f''(x)`",
    "`lim` `->` `x->inf` `h->0` `inf` `e` `ln` `log` `Delta`",
    "",
    "**No integral sign** - CAPS P1 is differentiation only. The soul spec forbids it.",
    "",
    "**Trig-specific**",
    "`sin cos tan` `arcsin arccos arctan`",
    "`theta alpha beta gamma` `pi` `deg` `rad` `2theta` `half theta` `sin^2 cos^2 tan^2`",
    "",
    "**Stats-specific**",
    "`x-bar` `sigma` `sigma^2` `Sigma Sigma x Sigma x2` `n N` `x_i f_i`",
    "`Q1 Q2 Q3` `min max` `P( )` `P(A)` `A|B` `union` `intersect`",
    "",
    "**Geometry-specific**",
    "`angle` `triangle` `parallel` `perp` `cong` `similar`",
    "`A-hat B-hat C-hat` (angle with hat)",
    "`arc AB` `chord` `tan` `rad`",
    "Reason keywords insert as plain text: `vert opp angles`, `alt angles`,",
    "`co-int angles`, `tan chord`, `angle at centre` etc.",
])

SEC5 = "\n".join([
    "## The six tabs",
    "",
    "| Tab | Visible in | Purpose |",
    "|---|---|---|",
    "| **Algebra** | P1 default | Solve, factorise, inequality, simultaneous |",
    "| **Calculus** | P1 | First principles, derivatives, tangents, cubic |",
    "| **Trig** | P1 + P2 | Identities, equations, general solution |",
    "| **Stats** | P1 + P2 | Mean, SD, regression, ogive, probability |",
    "| **Geometry** | P2 | Euclidean, analytical circle, proofs |",
    "| **Greek** | Fallback | alpha beta gamma theta lambda pi sigma phi omega |",
    "",
    "**Tab auto-switch mapping** (when learner opts in):",
    "",
    "| Skill prefix | Tab |",
    "|---|---|",
    "| `calculus.*` | Calculus |",
    "| `trig.*` | Trig |",
    "| `stats.*` / `probability.*` | Stats |",
    "| `geometry.*` / `analytical.*` | Geometry |",
    "| `algebra.*` / `sequences.*` / `functions.*` / `finance.*` | Algebra |",
])

SEC6 = "\n".join([
    "## Behaviour rules",
    "",
    "**Line model**",
    "- Enter -> new line, cursor at line start of the next one",
    "- Backspace on empty line -> deletes the line, cursor back to previous",
    "- Multi-line working area, one memo step per line",
    '- learner_lines = ["(x+4)(x-3)=0", "x+4=0 or x-3=0", "x=-4 or x=3"]',
    "",
    "**Insert semantics**",
    "- Most keys insert with a leading and trailing space if surrounded",
    "  by alphanumeric characters (prevents `x+4=0` becoming `x +4=0`)",
    "- `or` always inserts as ` or ` (space both sides)",
    "- `=>`, `<=>`, `therefore`, `because` insert as ` X ` (space both sides)",
    "- `;` inserts as `; ` (trailing space only - coordinates use trailing)",
    "- `sin(` `cos(` `tan(` - the `(` is part of the key",
    "- `sqrt` inserts `sqrt(` - cursor inside",
    "",
    "**Tab switching**",
    "- Manual tap always works",
    "- Auto-switch (when enabled): on question start, tab matches skill prefix",
    "- Manual override during a question sticks until the next question",
    "",
    "**Keyboard mode**",
    "- `ABC` key toggles letter input for geometry reasons",
    "- Letters on the ABC layer insert plain text",
    "- Digits always visible on the number row - never hidden by ABC",
    "",
    "**What the keyboard does NOT do**",
    "- No `=` that evaluates",
    "- No `solve` key",
    "- No history / recall that recomputes",
    "- No import from calculator",
])

SEC7 = "\n".join([
    "## Wireframe (phone portrait)",
    "",
    "```",
    "+-----------------------------------------+",
    "|  <- MethodMarker              (gear)    |",
    "+-----------------------------------------+",
    "|  Working                    3 steps     |",
    "|                                         |",
    "|  1  (x+4)(x-3)=0                        |",
    "|  2  x+4=0 or x-3=0                      |",
    "|  3  x=-4 or x=3|                        |",
    "|                                         |",
    "+-----------------------------------------+",
    "| Algebra|Calculus|Trig|Stats|Geom|Greek  |",
    "+-----------------------------------------+",
    "|  x    y    a    b    c    n             |",
    "|  =    or   =>   <=>  .:   :.            |",
    "|  ( )  x^2  x^n  x^-1 sqrt cbrt          |",
    "|  +/-  <    >    <=   >=   !=            |",
    "|  a/b  |x|  Sigma ;   in   R             |",
    "+-----------------------------------------+",
    "|  1 2 3 4 5 6 7 8 9 0 . (-)  backspace   |",
    "+-----------------------------------------+",
    "|  Enter New Line   Done   ABC   Collapse |",
    "+-----------------------------------------+",
    "```",
    "",
    "The `Collapse` control on the bottom-right toggles the symbol grid",
    "off, leaving the working area, digits, and Done - for learners who",
    "want more room to review their working before submitting.",
])

SEC8 = "\n".join([
    "## Output contract",
    "",
    "The keyboard produces, on Done:",
    "",
    "```python",
    "learner_lines: list[str] = [",
    '    "(x+4)(x-3)=0",',
    '    "x+4=0 or x-3=0",',
    '    "x=-4 or x=3",',
    "]",
    "```",
    "",
    "Same shape as the CLI's `_collect_working`. Same shape `step_grader`",
    "and `math_grader` already consume. **No interface change** - the",
    "keyboard is just a different producer.",
    "",
    "## Integration with the drill",
    "",
    "The drill's existing call site:",
    "",
    "```python",
    "submission = _collect_working(i, SITTING_SIZE)",
    "```",
    "",
    "On web/mobile, `_collect_working` renders the keyboard, waits for",
    "Done, and returns `learner_lines`-wrapped working. Same downstream",
    "pipeline:",
    "",
    "```",
    "keyboard -> learner_lines -> math_grader -> step_grader -> feedback",
    "```",
    "",
    "Nothing else in the drill changes.",
])

SEC9 = "\n".join([
    "## MVP vs full build",
    "",
    "**MVP (ship first)**",
    "- Digits, `.`, `(-)`, `+ - x /`, `( )`, `=`",
    "- Backspace, New Line, Done, ABC, Collapse",
    "- **Algebra tab** (all 30 keys)",
    "- **Trig tab** (all 24 keys)",
    "- Line counter, no cap",
    "- Manual tab switching only",
    "",
    "Ships the pilot. Unblocks all quadratic / surd / inequality / trig drills.",
    "",
    "**Full (after pilot data)**",
    "- Calculus, Stats, Geometry, Greek tabs",
    "- Tab auto-switch (opt-in)",
    "- Cursor position persistence across questions",
    "- Settings panel (persisted to learner row)",
])

SEC10 = "\n".join([
    "## Open questions",
    "",
    "1. **Cursor tracking.** Does the keyboard need to support mid-line",
    "   insertion (edit line 2 after line 3 is written)? Probably yes for",
    "   proofs, but adds complexity. Defer to full build.",
    "2. **Bilingual.** SA has English + Afrikaans papers. Learners may",
    "   switch. Do we ship EN-only first, or both?",
    "3. **Reason shortlist.** Geometry proofs have a fixed vocabulary of",
    "   reasons. Do we offer a shortlist picker (vert opp angles, alt",
    "   angles, co-int angles, tan chord, ...), or leave the learner to",
    "   type? Shortlist risks over-helping; freeform risks misspellings",
    "   the grader cannot match.",
    "4. **Line cap on the drill side.** Even if the keyboard has no cap,",
    "   the step grader has a fixed memo length. What happens when a",
    "   learner writes 15 lines for a 3-line memo? Log as a discovery",
    "   for later review.",
])


# ------------------------------------------------------------------
# The one code cell: emit keyboard_layout.json
# ------------------------------------------------------------------

CODE_CELL = "\n".join([
    "# Emit the keyboard_layout.json the drill will consume.",
    "",
    "import json",
    "from pathlib import Path",
    "",
    "KEYBOARD = {",
    '    "schema_version": 1,',
    '    "tabs": {',
    '        "Algebra": [',
    '            {"label": "x",   "insert": "x",    "engine_token": "x"},',
    '            {"label": "y",   "insert": "y",    "engine_token": "y"},',
    '            {"label": "a",   "insert": "a",    "engine_token": "a"},',
    '            {"label": "b",   "insert": "b",    "engine_token": "b"},',
    '            {"label": "c",   "insert": "c",    "engine_token": "c"},',
    '            {"label": "n",   "insert": "n",    "engine_token": "n"},',
    '            {"label": "=",   "insert": " = ",  "engine_token": "="},',
    '            {"label": "or",  "insert": " or ", "engine_token": " or "},',
    '            {"label": "=>",  "insert": " => ", "engine_token": "=>"},',
    '            {"label": "<=>", "insert": " <=> ","engine_token": "<=>"},',
    '            {"label": ".:",  "insert": " .: ", "engine_token": "therefore"},',
    '            {"label": ":.",  "insert": " :. ", "engine_token": "because"},',
    '            {"label": "( )", "insert": "()",   "engine_token": "()"},',
    '            {"label": "x^2", "insert": "x^2",  "engine_token": "x^2"},',
    '            {"label": "x^n", "insert": "x^",   "engine_token": "x^"},',
    '            {"label": "x^-1","insert": "x^-1", "engine_token": "x^-1"},',
    '            {"label": "sqrt","insert": "sqrt(", "engine_token": "sqrt("},',
    '            {"label": "cbrt","insert": "cbrt(", "engine_token": "cbrt("},',
    '            {"label": "+-",  "insert": "+-",    "engine_token": "+/-"},',
    '            {"label": "<",   "insert": "<",    "engine_token": "<"},',
    '            {"label": ">",   "insert": ">",    "engine_token": ">"},',
    '            {"label": "<=",  "insert": "<=",   "engine_token": "<="},',
    '            {"label": ">=",  "insert": ">=",   "engine_token": ">="},',
    '            {"label": "!=",  "insert": "!=",   "engine_token": "!="},',
    '            {"label": "a/b", "insert": "a/b",  "engine_token": "a/b"},',
    '            {"label": "|x|", "insert": "|x|",  "engine_token": "abs(x)"},',
    '            {"label": "Sum", "insert": "Sum",  "engine_token": "sum"},',
    '            {"label": ";",   "insert": "; ",   "engine_token": ";"},',
    '            {"label": "in",  "insert": "in",   "engine_token": "in"},',
    '            {"label": "R",   "insert": "R",    "engine_token": "R"},',
    "        ],",
    '        "Trig": [',
    '            {"label": "sin",   "insert": "sin(", "engine_token": "sin("},',
    '            {"label": "cos",   "insert": "cos(", "engine_token": "cos("},',
    '            {"label": "tan",   "insert": "tan(", "engine_token": "tan("},',
    '            {"label": "sin^-1","insert": "asin(","engine_token": "asin("},',
    '            {"label": "cos^-1","insert": "acos(","engine_token": "acos("},',
    '            {"label": "tan^-1","insert": "atan(","engine_token": "atan("},',
    '            {"label": "theta", "insert": "theta","engine_token": "theta"},',
    '            {"label": "alpha", "insert": "alpha","engine_token": "alpha"},',
    '            {"label": "beta",  "insert": "beta", "engine_token": "beta"},',
    '            {"label": "gamma", "insert": "gamma","engine_token": "gamma"},',
    '            {"label": "pi",    "insert": "pi",   "engine_token": "pi"},',
    '            {"label": "deg",   "insert": "deg",  "engine_token": "deg"},',
    '            {"label": "2theta","insert": "2theta","engine_token": "2theta"},',
    '            {"label": "sin^2", "insert": "sin^2(","engine_token": "sin^2("},',
    '            {"label": "cos^2", "insert": "cos^2(","engine_token": "cos^2("},',
    '            {"label": "tan^2", "insert": "tan^2(","engine_token": "tan^2("},',
    '            {"label": "rad",   "insert": "rad",  "engine_token": "rad"},',
    '            {"label": "=",     "insert": " = ",  "engine_token": "="},',
    '            {"label": "or",    "insert": " or ", "engine_token": " or "},',
    '            {"label": "+-",    "insert": "+-",    "engine_token": "+/-"},',
    '            {"label": "=>",    "insert": " => ", "engine_token": "=>"},',
    '            {"label": ".:",    "insert": " .: ", "engine_token": "therefore"},',
    '            {"label": "( )",   "insert": "()",   "engine_token": "()"},',
    "        ],",
    "    },",
    '    "always_visible": {',
    '        "digits": ["1","2","3","4","5","6","7","8","9","0",".","(-)"],',
    '        "operators": ["+","-","x","/"],',
    '        "actions": ["Backspace","New Line","Done","ABC","Collapse"],',
    "    },",
    '    "settings": {',
    '        "line_cap": None,',
    '        "tab_auto_switch": False,',
    '        "keyboard_collapsible": True,',
    '        "keyboard_default": "expanded",',
    "    },",
    '    "tab_auto_switch_map": {',
    '        "calculus": "Calculus",',
    '        "trig": "Trig",',
    '        "stats": "Stats",',
    '        "probability": "Stats",',
    '        "geometry": "Geometry",',
    '        "analytical": "Geometry",',
    '        "algebra": "Algebra",',
    '        "sequences": "Algebra",',
    '        "functions": "Algebra",',
    '        "finance": "Algebra",',
    "    },",
    "}",
    "",
    'out = Path("data/processed/tutor/keyboard_layout.json")',
    "out.parent.mkdir(parents=True, exist_ok=True)",
    'out.write_text(json.dumps(KEYBOARD, indent=2, ensure_ascii=False), encoding="utf-8")',
    "",
    'n_keys = sum(len(v) for v in KEYBOARD["tabs"].values())',
    'print(f"Wrote {out}")',
    'print(f"  Tabs  : {len(KEYBOARD[\'tabs\'])}")',
    'print(f"  Keys  : {n_keys}")',
    'print(f"  Settings: line_cap={KEYBOARD[\'settings\'][\'line_cap\']}, "',
    '      f"tab_auto_switch={KEYBOARD[\'settings\'][\'tab_auto_switch\']}")',
])


CELLS = [
    md(SEC1),
    md(SEC2),
    md(SEC3),
    md(SEC4),
    md(SEC5),
    md(SEC6),
    md(SEC7),
    md(SEC8),
    md(SEC9),
    md(SEC10),
    code(CODE_CELL),
]


NOTEBOOK = {
    "cells": CELLS,
    "metadata": {
        "kernelspec": {
            "display_name": "Python 3",
            "language": "python",
            "name": "python3",
        },
        "language_info": {
            "name": "python",
            "version": "3.11",
        },
    },
    "nbformat": 4,
    "nbformat_minor": 5,
}

NB_PATH.parent.mkdir(parents=True, exist_ok=True)
NB_PATH.write_text(json.dumps(NOTEBOOK, indent=1, ensure_ascii=False),
                   encoding="utf-8")
print(f"  wrote {NB_PATH}")
print(f"  size  {NB_PATH.stat().st_size:,} bytes")
print(f"  cells {len(CELLS)}")
````

````python
# Write slash_commands.py (safe — no nested triple quotes)

from pathlib import Path

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
SRC = ROOT / "src" / "tutor"
SC_PATH = SRC / "slash_commands.py"
DRILL = SRC / "drill.py"

lines = [
    '# src/tutor/slash_commands.py',
    '"""',
    'Slash-command notation expander for the CLI drill.',
    '',
    'A learner typing working on a physical keyboard cannot easily type',
    'unicode maths symbols. Instead they type slash commands that expand',
    'to the notation the NSC memo uses.',
    '',
    '    /sqrt      ->  sqrt(',
    '    /theta     ->  theta',
    '    /therefore ->  therefore',
    '    /or        ->  " or "',
    '    /deg       ->  deg',
    '',
    'Expansion happens at submission time, before grading. The engine',
    'stores ASCII-safe tokens internally so the math_grader and',
    'step_grader stay stable.',
    '',
    'Design rules:',
    '  * Only expands /name followed by a non-word char or end-of-string.',
    '  * Unknown /foo is left alone (learner can still write /paths).',
    '  * /help, /?, /commands prints the full list.',
    '"""',
    'from __future__ import annotations',
    'import re',
    '',
    '',
    'SLASH_COMMANDS = {',
    '    "sqrt":    "sqrt(",',
    '    "cbrt":    "cbrt(",',
    '    "pi":      "pi",',
    '    "pm":      "+/-",',
    '    "le":      "<=",',
    '    "ge":      ">=",',
    '    "ne":      "!=",',
    '    "in":      "in",',
    '    "R":       "R",',
    '    "sum":     "sum",',
    '    "inf":     "inf",',
    '    "or":      " or ",',
    '    "then":    " therefore ",',
    '    "because": " because ",',
    '    "implies": " => ",',
    '    "iff":     " <=> ",',
    '    "deg":     "deg",',
    '    "cdot":    " x ",',
    '    "times":   " x ",',
    '    "div":     " / ",',
    '    "theta":   "theta",',
    '    "alpha":   "alpha",',
    '    "beta":    "beta",',
    '    "gamma":   "gamma",',
    '    "sin":     "sin(",',
    '    "cos":     "cos(",',
    '    "tan":     "tan(",',
    '    "asin":    "asin(",',
    '    "acos":    "acos(",',
    '    "atan":    "atan(",',
    '    "dy":      "dy/dx",',
    '    "dx":      "d/dx",',
    '    "d2y":     "d2y/dx2",',
    '    "lim":     "lim",',
    '    "to":      "->",',
    '    "angle":   "angle",',
    '    "tri":     "triangle",',
    '    "para":    "parallel",',
    '    "perp":    "perp",',
    '    "cong":    "cong",',
    '    "sim":     "similar",',
    '    "arc":     "arc ",',
    '    "chord":   "chord ",',
    '    "tchord":  "tan chord",',
    '    "atcentre":"angle at centre",',
    '    "xbar":    "xbar",',
    '    "sigma":   "sigma",',
    '    "sigma2":  "sigma^2",',
    '}',
    '',
    '',
    'HELP_TEXT = (',
    '    "\\nMethodKeyboard - slash commands\\n"',
    '    "\\n  Algebra"',
    '    "\\n    /sqrt     sqrt(         /cbrt     cbrt("',
    '    "\\n    /pi       pi            /pm       +/-"',
    '    "\\n    /le       <=            /ge       >="',
    '    "\\n    /ne       !=            /in       in"',
    '    "\\n    /R        R             /sum      sum"',
    '    "\\n    /inf      inf           /deg      deg"',
    '    "\\n    /or       or            /then     therefore"',
    '    "\\n    /because  because       /implies  =>"',
    '    "\\n    /iff      <=>           /cdot     x"',
    '    "\\n"',
    '    "\\n  Trig"',
    '    "\\n    /theta    theta         /alpha    alpha"',
    '    "\\n    /beta     beta          /gamma    gamma"',
    '    "\\n    /sin      sin(          /cos      cos("',
    '    "\\n    /tan      tan(          /asin     asin("',
    '    "\\n    /acos     acos(         /atan     atan("',
    '    "\\n"',
    '    "\\n  Calculus"',
    '    "\\n    /dy       dy/dx         /dx       d/dx"',
    '    "\\n    /d2y      d2y/dx2       /lim      lim"',
    '    "\\n    /to       ->"',
    '    "\\n"',
    '    "\\n  Geometry"',
    '    "\\n    /angle    angle         /tri      triangle"',
    '    "\\n    /para     parallel      /perp     perp"',
    '    "\\n    /cong     cong          /sim      similar"',
    '    "\\n    /arc      arc           /chord    chord"',
    '    "\\n    /tchord   tan chord     /atcentre angle at centre"',
    '    "\\n"',
    '    "\\n  Stats"',
    '    "\\n    /xbar     xbar          /sigma    sigma"',
    '    "\\n    /sigma2   sigma^2"',
    '    "\\n"',
    '    "\\nType /help for this list."',
    '    "\\n"',
    ')',
    '',
    '',
    '_CMD_PATTERN = re.compile(r"/([A-Za-z][A-Za-z0-9_]*)(?![A-Za-z0-9_])")',
    '',
    '',
    'def expand(text: str) -> str:',
    '    """Expand every /command in text. Unknown commands are left alone."""',
    '    if not text or "/" not in text:',
    '        return text',
    '',
    '    def _replace(match):',
    '        name = match.group(1)',
    '        return SLASH_COMMANDS.get(name, match.group(0))',
    '',
    '    return _CMD_PATTERN.sub(_replace, text)',
    '',
    '',
    'def expand_lines(lines):',
    '    """Expand a list of learner lines. Returns a new list."""',
    '    return [expand(line) for line in lines]',
    '',
    '',
    'def is_help(text: str) -> bool:',
    '    """Is this submission just a help request?"""',
    '    return text.strip() in ("/help", "/?", "/commands")',
    '',
    '',
    '__all__ = [',
    '    "SLASH_COMMANDS", "HELP_TEXT",',
    '    "expand", "expand_lines", "is_help",',
    ']',
]

SC_PATH.write_text("\n".join(lines) + "\n", encoding="utf-8")
print(f"  ✓ wrote {SC_PATH.relative_to(ROOT)}")
print(f"    {SC_PATH.stat().st_size:,} bytes")


# ============================================================
# Wire into drill.py
# ============================================================

src = DRILL.read_text(encoding="utf-8")

ANCHOR = "\n".join([
    '            if submission == "HELP" or submission == "MORE":',
    '                # Show a worked example from the same template, then retry',
    '                _show_sibling(pool, shown_sibling_keys)',
    '                print()',
    '                print("  Try the question again when ready — or type \'next\' to move on.")',
    '                continue',
    '',
    '            attempts_on_this_q += 1',
])

PATCH = "\n".join([
    '            if submission == "HELP" or submission == "MORE":',
    '                # Show a worked example from the same template, then retry',
    '                _show_sibling(pool, shown_sibling_keys)',
    '                print()',
    '                print("  Try the question again when ready — or type \'next\' to move on.")',
    '                continue',
    '',
    '            # --- slash-command notation expansion ---',
    '            from . import slash_commands as _sc',
    '',
    '            _steps = getattr(submission, "steps", None) or []',
    '            _first_raw = ""',
    '            if _steps and hasattr(_steps[0], "raw_text"):',
    '                _first_raw = _steps[0].raw_text.strip()',
    '',
    '            if _sc.is_help(_first_raw) and len(_steps) == 1:',
    '                print(_sc.HELP_TEXT)',
    '                continue',
    '',
    '            for _s in _steps:',
    '                if hasattr(_s, "raw_text"):',
    '                    _s.raw_text = _sc.expand(_s.raw_text)',
    '',
    '            attempts_on_this_q += 1',
])

if "from . import slash_commands as _sc" in src:
    print("  · drill.py already wired")
elif ANCHOR in src:
    src = src.replace(ANCHOR, PATCH)
    DRILL.write_text(src, encoding="utf-8")
    print("  ✓ drill.py patched")
else:
    print("  !! anchor not matched in drill.py")
    # Show what we tried to match so we can adjust
    print()
    print("  Looking for this block:")
    for line in ANCHOR.splitlines()[:6]:
        print(f"    {line!r}")


# ============================================================
# Smoke test
# ============================================================

import sys
for mod in list(sys.modules):
    if mod == "tutor" or mod.startswith("tutor."):
        del sys.modules[mod]
sys.path.insert(0, str(ROOT / "src"))

from tutor import slash_commands as sc

print()
print("=" * 72)
print("SLASH EXPANDER SMOKE TEST")
print("=" * 72)

samples = [
    "x = 0 /or x = 3",
    "/sqrt(x+1) = x - 2",
    "/theta = 30/deg",
    "sin(/theta) + cos(/theta) = 1",
    "/dy = 6x + 3",
    "A(1; 9) and B(/pi; 0)",
    "/le and /ge are inequalities",
    "unknown /foo is left alone",
    "3 /cdot 4 = 12",
]

for line in samples:
    expanded = sc.expand(line)
    arrow = " -> " if expanded != line else "    "
    print(f"  {line}")
    if expanded != line:
        print(f"    {arrow}{expanded}")

print()
print(f"  /help recognized: {sc.is_help('/help')}")
print(f"  /? recognized   : {sc.is_help('/?')}")
print(f"  not help        : {sc.is_help('solve')}")
print(f"  commands loaded : {len(sc.SLASH_COMMANDS)}")
````

````python
# Corpus map — text summary (read first, before any HTML)

import csv
import json
from pathlib import Path
from collections import Counter, defaultdict

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
TUTOR_DIR = ROOT / "data" / "processed" / "tutor"
TEMPLATE_DIR = TUTOR_DIR / "question_templates"
DIAG_CSV = ROOT / "data" / "processed" / "diagnostics" / "diagnostic_errors_with_skill.csv"


# ============================================================
# 1. Templates
# ============================================================

templates = []
for p in sorted(TEMPLATE_DIR.glob("template_*.json")):
    try:
        t = json.loads(p.read_text(encoding="utf-8"))
    except Exception:
        continue
    src = t.get("source", {})
    subs = t.get("subquestions", [])
    skills = sorted({s.get("skill_id") for s in subs if s.get("skill_id")})
    templates.append({
        "file": p.name,
        "year": src.get("year"),
        "paper": src.get("paper"),
        "qnum": src.get("qnum"),
        "diagram": t.get("diagram"),
        "has_spec": bool(t.get("diagram_spec")),
        "has_image": bool(t.get("diagram_image")),
        "skills": skills,
        "n_subqs": len(subs),
    })


# ============================================================
# 2. Variants
# ============================================================

variants = {}
vp = TUTOR_DIR / "variant_pool_spec_canonical.csv"
if vp.exists():
    with vp.open(encoding="utf-8-sig") as f:
        for r in csv.DictReader(f):
            sid = r.get("skill_id")
            if sid:
                variants[sid] = {
                    "pool_size": int(r.get("pool_size") or 0),
                    "quality": r.get("attribution_quality") or "-",
                }


# ============================================================
# 3. Diagrams index (skill → pages)
# ============================================================

diag_index = {}
dp = TUTOR_DIR / "skill_diagram_index.json"
if dp.exists():
    diag_index = json.loads(dp.read_text(encoding="utf-8"))


# ============================================================
# 4. Diagnostics
# ============================================================

diag_rows = []
if DIAG_CSV.exists():
    with DIAG_CSV.open(encoding="utf-8-sig", errors="ignore") as f:
        for r in csv.DictReader(f):
            diag_rows.append(r)


# ============================================================
# 5. Skill aggregation
# ============================================================

all_skills = set()
for t in templates:
    all_skills.update(t["skills"])

skill_stats = {}
for sid in sorted(all_skills):
    tpl_count = sum(1 for t in templates if sid in t["skills"])
    var = variants.get(sid, {})
    diag_count = len(diag_index.get(sid, []))
    err_count = sum(1 for r in diag_rows if r.get("skill_id") == sid)
    skill_stats[sid] = {
        "templates": tpl_count,
        "variants": var.get("pool_size", 0),
        "quality": var.get("quality", "-"),
        "diagrams": diag_count,
        "diagnostics": err_count,
    }


# ============================================================
# 6. Print the summary
# ============================================================

print("=" * 72)
print("METHODMARKER CORPUS MAP")
print("=" * 72)
print()
print(f"  Templates        : {len(templates)}")
print(f"  Subquestions     : {sum(t['n_subqs'] for t in templates)}")
print(f"  Distinct skills  : {len(all_skills)}")
print(f"  Variants indexed : {len(variants)} skills")
print(f"  Diagram pages    : {len(diag_index)} skills indexed")
print(f"  Diagnostics rows : {len(diag_rows)}")
print()

# By year × paper
print("=" * 72)
print("TEMPLATES BY YEAR × PAPER")
print("=" * 72)
matrix = defaultdict(lambda: Counter())
for t in templates:
    matrix[t["year"]][t["paper"]] += 1

years = sorted({t["year"] for t in templates}, reverse=True)
print(f"  {'Year':>6} | {'P1':>4} | {'P2':>4} | total")
print(f"  {'-'*6} | {'-'*4} | {'-'*4} | -----")
for y in years:
    p1 = matrix[y].get("P1", 0)
    p2 = matrix[y].get("P2", 0)
    print(f"  {y:>6} | {p1:>4} | {p2:>4} | {p1 + p2:>5}")
print()

# Diagram coverage
with_diagram = sum(1 for t in templates if t["diagram"])
with_spec = sum(1 for t in templates if t["has_spec"])
with_image = sum(1 for t in templates if t["has_image"])
print("=" * 72)
print("DIAGRAM COVERAGE")
print("=" * 72)
print(f"  Templates with a diagram      : {with_diagram}")
print(f"  Templates with diagram_spec   : {with_spec}")
print(f"  Templates with diagram_image  : {with_image}")
print(f"  → page-image coverage         : "
      f"{100 * with_image / max(with_diagram, 1):.0f}%")
print()

# Skill coverage
print("=" * 72)
print("SKILL COVERAGE (top 25 by template count)")
print("=" * 72)
print(f"  {'Skill':<32} {'Tpl':>4} {'Var':>4} {'Qual':>8} {'Diag':>5} {'Errs':>5}")
print(f"  {'-'*32} {'-'*4} {'-'*4} {'-'*8} {'-'*5} {'-'*5}")
for sid, st in sorted(skill_stats.items(),
                      key=lambda x: -x[1]["templates"])[:25]:
    print(f"  {sid:<32} {st['templates']:>4} {st['variants']:>4} "
          f"{st['quality']:>8} {st['diagrams']:>5} {st['diagnostics']:>5}")
print()

# Gaps
no_variants = [s for s, st in skill_stats.items() if st["variants"] == 0]
no_diagrams = [s for s, st in skill_stats.items() if st["diagrams"] == 0]
print("=" * 72)
print("GAPS")
print("=" * 72)
print(f"  Skills without variants      : {len(no_variants)}")
print(f"  Skills without diagrams      : {len(no_diagrams)}")
print()
print("  Skills with most templates but no variants (biggest gaps):")
no_var_sorted = sorted(
    [(s, skill_stats[s]["templates"]) for s in no_variants],
    key=lambda x: -x[1],
)
for sid, n in no_var_sorted[:15]:
    print(f"    {sid:<32}  {n} template(s)")
print()

# Top topic_v2 codes from diagnostics
if diag_rows:
    topics = Counter(r.get("topic_v2") or r.get("topic") or "?"
                     for r in diag_rows)
    print("=" * 72)
    print("DIAGNOSTICS BY TOPIC")
    print("=" * 72)
    for t, n in topics.most_common(15):
        print(f"  {t:<32}  {n:>4}")
````

````python
# Complete keyboard layout (all 6 tabs) + generate HTML mockup

import json
from pathlib import Path

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
TUTOR = ROOT / "data" / "processed" / "tutor"
LAYOUT_PATH = TUTOR / "keyboard_layout.json"
HTML_PATH = ROOT / "docs" / "keyboard_preview.html"

# ============================================================
# 1. Full six-tab layout
# ============================================================

KEYBOARD = {
    "schema_version": 1,
    "tabs_order": ["Algebra", "Calculus", "Trig", "Stats", "Geometry", "Greek"],
    "tabs": {
        "Algebra": [
            # row 1 — variables
            [{"l": "x", "i": "x"}, {"l": "y", "i": "y"},
             {"l": "a", "i": "a"}, {"l": "b", "i": "b"},
             {"l": "c", "i": "c"}, {"l": "n", "i": "n"}],
            # row 2 — relations + logic
            [{"l": "=", "i": " = ", "sp": True},
             {"l": "or", "i": " or ", "sp": True},
             {"l": "\u21d2", "i": " \u21d2 ", "sp": True},
             {"l": "\u21d4", "i": " \u21d4 ", "sp": True},
             {"l": "\u2234", "i": " \u2234 ", "sp": True},
             {"l": "\u2235", "i": " \u2235 ", "sp": True}],
            # row 3 — powers + roots
            [{"l": "( )", "i": "()", "skip": 1},
             {"l": "x\u00b2", "i": "x^2"},
             {"l": "x\u207f", "i": "x^"},
             {"l": "x\u207b\u00b9", "i": "x^-1"},
             {"l": "\u221a", "i": "sqrt"},
             {"l": "\u221b", "i": "cbrt"}],
            # row 4 — inequalities
            [{"l": "\u00b1", "i": "+/-"},
             {"l": "<", "i": "<"}, {"l": ">", "i": ">"},
             {"l": "\u2264", "i": "<="},
             {"l": "\u2265", "i": ">="},
             {"l": "\u2260", "i": "!="}],
            # row 5 — algebra extras
            [{"l": "a/b", "i": "a/b"},
             {"l": "|x|", "i": "abs(x)"},
             {"l": "\u03a3", "i": "sum"},
             {"l": ";", "i": "; ", "sp": False, "trail": True},
             {"l": "\u2208", "i": " in "},
             {"l": "\u211d", "i": "R"}],
        ],
        "Calculus": [
            [{"l": "dy/dx", "i": "dy/dx"}, {"l": "d/dx", "i": "d/dx"},
             {"l": "d\u00b2y/dx\u00b2", "i": "d2y/dx2"},
             {"l": "D\u2093[ ]", "i": "Dx[]"}],
            [{"l": "f'(x)", "i": "f'(x)"}, {"l": "f''(x)", "i": "f''(x)"},
             {"l": "f'''(x)", "i": "f'''(x)"},
             {"l": "\u0394", "i": "Delta"}],
            [{"l": "lim", "i": "lim("}, {"l": "\u2192", "i": "->"},
             {"l": "h\u21920", "i": "h->0"}, {"l": "x\u2192\u221e", "i": "x->inf"}],
            [{"l": "e", "i": "e"}, {"l": "ln", "i": "ln("},
             {"l": "log", "i": "log("}, {"l": "\u221e", "i": "inf"}],
            [{"l": "=", "i": " = ", "sp": True},
             {"l": "\u2234", "i": " \u2234 ", "sp": True},
             {"l": "( )", "i": "()", "skip": 1},
             {"l": "\u221a", "i": "sqrt"}],
        ],
        "Trig": [
            [{"l": "sin", "i": "sin"}, {"l": "cos", "i": "cos"},
             {"l": "tan", "i": "tan"}],
            [{"l": "sin\u207b\u00b9", "i": "asin"},
             {"l": "cos\u207b\u00b9", "i": "acos"},
             {"l": "tan\u207b\u00b9", "i": "atan"}],
            [{"l": "\u03b8", "i": "theta"}, {"l": "\u03b1", "i": "alpha"},
             {"l": "\u03b2", "i": "beta"}, {"l": "\u03b3", "i": "gamma"}],
            [{"l": "\u03c0", "i": "pi"},
             {"l": "\u00b0", "i": "deg"},
             {"l": "rad", "i": "rad"}],
            [{"l": "2\u03b8", "i": "2theta"},
             {"l": "\u00bd\u03b8", "i": "theta/2"},
             {"l": "sin\u00b2", "i": "sin^2"},
             {"l": "cos\u00b2", "i": "cos^2"},
             {"l": "tan\u00b2", "i": "tan^2"}],
            [{"l": "=", "i": " = ", "sp": True},
             {"l": "or", "i": " or ", "sp": True},
             {"l": "\u00b1", "i": "+/-"},
             {"l": "\u2234", "i": " \u2234 ", "sp": True},
             {"l": "( )", "i": "()", "skip": 1}],
        ],
        "Stats": [
            [{"l": "x\u0304", "i": "xbar"}, {"l": "\u03c3", "i": "sigma"},
             {"l": "\u03c3\u00b2", "i": "sigma^2"},
             {"l": "n", "i": "n"}, {"l": "N", "i": "N"}],
            [{"l": "\u03a3", "i": "sum"},
             {"l": "\u03a3x", "i": "sum(x)"},
             {"l": "\u03a3x\u00b2", "i": "sum(x^2)"},
             {"l": "x\u1d62", "i": "x_i"},
             {"l": "f\u1d62", "i": "f_i"}],
            [{"l": "Q1", "i": "Q1"}, {"l": "Q2", "i": "Q2"},
             {"l": "Q3", "i": "Q3"},
             {"l": "min", "i": "min"}, {"l": "max", "i": "max"}],
            [{"l": "P( )", "i": "P()", "skip": 1},
             {"l": "A|B", "i": "A|B"},
             {"l": "\u222a", "i": " union "},
             {"l": "\u2229", "i": " intersect "}],
            [{"l": "=", "i": " = ", "sp": True},
             {"l": ";", "i": "; ", "trail": True},
             {"l": "( )", "i": "()", "skip": 1},
             {"l": "%", "i": "%"}],
        ],
        "Geometry": [
            [{"l": "\u2220", "i": "angle "},
             {"l": "\u25b3", "i": "triangle "},
             {"l": "\u2225", "i": "parallel "},
             {"l": "\u22a5", "i": "perp "},
             {"l": "\u2245", "i": "cong "},
             {"l": "\u223c", "i": "similar "}],
            [{"l": "\u2234", "i": " \u2234 ", "sp": True},
             {"l": "\u2235", "i": " \u2235 ", "sp": True},
             {"l": "\u00c2", "i": "A^"}, {"l": "B\u0302", "i": "B^"},
             {"l": "\u0108", "i": "C^"}, {"l": "D\u0302", "i": "D^"}],
            [{"l": "arc", "i": "arc "},
             {"l": "chord", "i": "chord "},
             {"l": "tan", "i": "tan "},
             {"l": "rad", "i": "rad "},
             {"l": "=", "i": " = ", "sp": True},
             {"l": "( )", "i": "()", "skip": 1}],
            [{"l": "vert opp", "i": "vert opp \u2220s"},
             {"l": "alt \u2220s", "i": "alt \u2220s"},
             {"l": "co-int \u2220s", "i": "co-int \u2220s"}],
            [{"l": "tan chord", "i": "tan chord"},
             {"l": "\u2220 at centre", "i": "\u2220 at centre"},
             {"l": "cyclic quad", "i": "cyclic quad"}],
            [{"l": ";", "i": "; ", "trail": True},
             {"l": "|", "i": "|"},
             {"l": "\u2194", "i": "<->"},
             {"l": "\u21d2", "i": " \u21d2 ", "sp": True}],
        ],
        "Greek": [
            [{"l": "\u03b1", "i": "alpha"}, {"l": "\u03b2", "i": "beta"},
             {"l": "\u03b3", "i": "gamma"}, {"l": "\u03b4", "i": "delta"},
             {"l": "\u03b5", "i": "epsilon"}],
            [{"l": "\u03b8", "i": "theta"}, {"l": "\u03bb", "i": "lambda"},
             {"l": "\u03bc", "i": "mu"}, {"l": "\u03c0", "i": "pi"},
             {"l": "\u03c1", "i": "rho"}],
            [{"l": "\u03c3", "i": "sigma"}, {"l": "\u03c4", "i": "tau"},
             {"l": "\u03c6", "i": "phi"}, {"l": "\u03c9", "i": "omega"},
             {"l": "\u0394", "i": "Delta"}],
            [{"l": "\u03a3", "i": "Sigma"}, {"l": "\u03a9", "i": "Omega"},
             {"l": "\u03a6", "i": "Phi"}, {"l": "\u03a8", "i": "Psi"},
             {"l": "\u0398", "i": "Theta"}],
        ],
    },
    "always_visible": {
        "digits": ["1", "2", "3", "4", "5", "6", "7", "8", "9", "0", ".", "(-)"],
        "operators": ["+", "\u2212", "\u00d7", "\u00f7"],
    },
    "settings": {
        "line_cap": None,
        "tab_auto_switch": False,
        "keyboard_collapsible": True,
        "keyboard_default": "expanded",
    },
    "tab_auto_switch_map": {
        "calculus": "Calculus", "trig": "Trig",
        "stats": "Stats", "probability": "Stats",
        "geometry": "Geometry", "analytical": "Geometry",
        "algebra": "Algebra", "sequences": "Algebra",
        "functions": "Algebra", "finance": "Algebra",
    },
}

LAYOUT_PATH.write_text(
    json.dumps(KEYBOARD, indent=2, ensure_ascii=False), encoding="utf-8")
n_keys = sum(len(row) for tab in KEYBOARD["tabs"].values() for row in tab)
print(f"  ✓ keyboard_layout.json — {len(KEYBOARD['tabs'])} tabs, {n_keys} keys")


# ============================================================
# 2. Generate the HTML mockup
# ============================================================

HTML = []
A = HTML.append

A("<!doctype html>")
A("<html lang='en'>")
A("<head>")
A("<meta charset='utf-8'>")
A("<meta name='viewport' content='width=device-width,initial-scale=1'>")
A("<title>MethodKeyboard — preview</title>")
A("<style>")
A("""
  * { box-sizing: border-box; -webkit-tap-highlight-color: transparent; }
  body {
    margin: 0; background: #eef2f8; font-family: -apple-system,
    BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif;
    display: flex; justify-content: center; padding: 12px;
    min-height: 100vh;
  }
  .phone {
    width: 380px; max-width: 100%; background: #ffffff;
    border-radius: 24px; overflow: hidden;
    box-shadow: 0 8px 32px rgba(20, 40, 90, .12);
    display: flex; flex-direction: column;
  }
  .topbar {
    display: flex; align-items: center; gap: 8px;
    padding: 14px 16px; background: #f6f8fd;
    border-bottom: 1px solid #e4e9f2;
  }
  .topbar .logo {
    width: 30px; height: 30px; border-radius: 8px;
    background: linear-gradient(135deg, #3b5bf6, #5b8cf6);
    display: flex; align-items: center; justify-content: center;
    color: #fff; font-weight: 700; font-size: 16px;
  }
  .topbar .brand { font-weight: 700; font-size: 15px; color: #0b1f44; }
  .topbar .tag { font-size: 10px; color: #6b7a99; }

  .working {
    padding: 14px 16px; background: #ffffff;
    border-bottom: 1px solid #e4e9f2;
  }
  .working .head {
    display: flex; justify-content: space-between;
    align-items: center; font-size: 12px;
    color: #465a80; font-weight: 600; margin-bottom: 8px;
  }
  .working .steps { color: #6b7a99; font-weight: 500; }
  .lines {
    display: grid; grid-template-columns: 22px 1fr;
    column-gap: 8px; font-size: 15px; line-height: 26px;
    color: #0b1f44; max-height: 200px; overflow-y: auto;
  }
  .nums { color: #9aa8c2; font-size: 11px; text-align: right;
          user-select: none; padding-top: 2px; }
  .nums div { height: 26px; line-height: 26px; }
  .text { white-space: pre-wrap; word-break: break-word;
          min-height: 26px; outline: none; padding: 0;
          font-family: 'SF Mono', Menlo, Consolas, monospace; }
  .text:focus { outline: none; }
  .caret {
    display: inline-block; width: 1px; height: 18px;
    background: #3b5bf6; vertical-align: -3px;
    animation: blink 1s step-end infinite;
  }
  @keyframes blink { 50% { opacity: 0; } }

  .tabs {
    display: flex; overflow-x: auto; gap: 4px;
    padding: 8px 12px; background: #f6f8fd;
    border-bottom: 1px solid #e4e9f2; scrollbar-width: none;
  }
  .tabs::-webkit-scrollbar { display: none; }
  .tab {
    flex: none; padding: 8px 14px; border-radius: 18px;
    font-size: 12px; font-weight: 600; color: #465a80;
    background: transparent; border: none; cursor: pointer;
    white-space: nowrap;
  }
  .tab.active { background: #3b5bf6; color: #fff; }

  .grid { padding: 10px 12px; background: #ffffff;
          display: flex; flex-direction: column; gap: 6px;
          transition: max-height .25s ease; overflow: hidden;
          max-height: 1000px; }
  .grid.collapsed { max-height: 0; padding-top: 0;
                    padding-bottom: 0; }
  .row { display: grid; gap: 6px; }
  .key {
    padding: 12px 6px; border-radius: 10px;
    background: #ffffff; border: 1px solid #dfe5ef;
    font-size: 15px; color: #0b1f44; cursor: pointer;
    text-align: center; user-select: none; font-family: inherit;
    transition: transform .05s, background .1s;
    min-height: 44px; display: flex; align-items: center;
    justify-content: center;
  }
  .key:active { transform: scale(.96); background: #eef2fb; }
  .key.accent {
    background: #eef2fb; border-color: #c5d1e9;
    color: #223a6b; font-weight: 600;
  }

  .numrow {
    padding: 8px 12px; display: grid;
    grid-template-columns: repeat(12, 1fr); gap: 4px;
    background: #f6f8fd; border-top: 1px solid #e4e9f2;
  }
  .numrow .key { padding: 10px 2px; min-height: 40px;
                 font-size: 14px; }

  .actions {
    padding: 10px 12px 14px; display: grid;
    grid-template-columns: 1fr 1fr 1.4fr 1fr; gap: 6px;
    background: #f6f8fd;
  }
  .action {
    padding: 12px 6px; border-radius: 10px;
    background: #eef2fb; border: 1px solid #dfe5ef;
    color: #223a6b; font-size: 12px; font-weight: 600;
    cursor: pointer; font-family: inherit; text-align: center;
    min-height: 44px;
  }
  .action.primary { background: #3b5bf6; color: #fff;
                    border-color: #3b5bf6; }
  .action:active { transform: scale(.97); }

  .result {
    padding: 12px 16px; background: #0b1f44; color: #fff;
    font-family: 'SF Mono', Menlo, monospace;
    font-size: 11px; line-height: 1.5; max-height: 0;
    overflow: hidden; transition: max-height .25s ease;
  }
  .result.shown { max-height: 240px; overflow-y: auto; }
  .result .label { color: #9ab; margin-bottom: 4px; }
  .result pre { margin: 0; color: #cfe2ff; }
""")
A("</style>")
A("</head>")
A("<body>")

A("<div class='phone'>")
A("  <div class='topbar'>")
A("    <div class='logo'>M</div>")
A("    <div>")
A("      <div class='brand'>MethodMarker</div>")
A("      <div class='tag'>Better Methods. Higher Marks.</div>")
A("    </div>")
A("  </div>")

# Working area
A("  <div class='working'>")
A("    <div class='head'>")
A("      <span>&#9998; Working</span>")
A("      <span class='steps' id='stepCount'>0 steps</span>")
A("    </div>")
A("    <div class='lines'>")
A("      <div class='nums' id='nums'></div>")
A("      <div class='text' id='text' contenteditable='true' "
  "spellcheck='false'></div>")
A("    </div>")
A("  </div>")

# Tabs
A("  <div class='tabs' id='tabs'></div>")

# Symbol grid
A("  <div class='grid' id='grid'></div>")

# Number row
A("  <div class='numrow' id='numrow'></div>")

# Actions
A("  <div class='actions'>")
A("    <div class='action' data-action='backspace'>&#9003; Back</div>")
A("    <div class='action' data-action='newline'>&#8629; New Line</div>")
A("    <div class='action primary' data-action='done'>&#10003; Done</div>")
A("    <div class='action' data-action='collapse'>&#9660; Collapse</div>")
A("  </div>")

A("  <div class='result' id='result'></div>")
A("</div>")

# ---- Embedded layout ----
A("<script>")
A(f"const LAYOUT = {json.dumps(KEYBOARD, ensure_ascii=False)};")
A("""
const state = {
  activeTab: LAYOUT.tabs_order[0],
  collapsed: false,
};

// ---------- render working text ----------
const textEl = document.getElementById('text');
const numsEl = document.getElementById('nums');
const stepCountEl = document.getElementById('stepCount');

function refreshNumbers() {
  const lines = textEl.innerText.split('\\n');
  numsEl.innerHTML = lines.map((_, i) => `<div>${i + 1}</div>`).join('');
  const n = lines.length;
  stepCountEl.textContent = n === 1 && lines[0] === ''
    ? '0 steps' : `${n} step${n === 1 ? '' : 's'}`;
}
textEl.addEventListener('input', refreshNumbers);
refreshNumbers();

// ---------- insert helpers ----------
function getCaretContext() {
  const sel = window.getSelection();
  if (!sel.rangeCount) return { before: '', after: '' };
  const r = sel.getRangeAt(0);
  const pre = r.cloneRange();
  pre.selectNodeContents(textEl);
  pre.setEnd(r.startContainer, r.startOffset);
  const before = pre.toString();
  const post = r.cloneRange();
  post.selectNodeContents(textEl);
  post.setStart(r.endContainer, r.endOffset);
  const after = post.toString();
  return { before, after };
}

function insertAtCaret(str, opts) {
  opts = opts || {};
  textEl.focus();
  const sel = window.getSelection();
  if (!sel.rangeCount) return;
  const r = sel.getRangeAt(0);
  r.deleteContents();
  const node = document.createTextNode(str);
  r.insertNode(node);
  r.setStartAfter(node);
  r.setEndAfter(node);
  sel.removeAllRanges();
  sel.addRange(r);
  refreshNumbers();
}

function smartInsert(key) {
  const { before, after } = getCaretContext();
  let ins = key.i;

  // smart spacing: if 'sp' flag, ensure spaces on both sides
  if (key.sp) {
    const needLead = before.length > 0 && !/\\s$/.test(before);
    const needTrail = after.length > 0 && !/^\\s/.test(after);
    ins = (needLead ? ' ' : '') + ins.trim() + (needTrail ? ' ' : '');
  }

  // trail-only (semicolon)
  if (key.trail) {
    ins = ins; // already has trailing space
  }

  insertAtCaret(ins);

  // if 'skip' is 1, move cursor inside the parens
  if (key.skip === 1 && key.i.includes('()')) {
    // Move cursor back one char (past the ')')
    const sel = window.getSelection();
    if (sel.rangeCount) {
      const r = sel.getRangeAt(0);
      const newR = r.cloneRange();
      // Shift selection left one character
      if (newR.startContainer.nodeType === 3 && newR.startOffset > 0) {
        newR.setStart(newR.startContainer, newR.startOffset - 1);
        newR.collapse(true);
        sel.removeAllRanges();
        sel.addRange(newR);
      }
    }
  }
}

// ---------- render tabs ----------
const tabsEl = document.getElementById('tabs');
LAYOUT.tabs_order.forEach(name => {
  const b = document.createElement('button');
  b.className = 'tab' + (name === state.activeTab ? ' active' : '');
  b.textContent = name;
  b.onclick = () => {
    state.activeTab = name;
    document.querySelectorAll('.tab').forEach(t =>
      t.classList.toggle('active', t.textContent === name));
    renderGrid();
  };
  tabsEl.appendChild(b);
});

// ---------- render grid ----------
const gridEl = document.getElementById('grid');
function renderGrid() {
  const rows = LAYOUT.tabs[state.activeTab] || [];
  gridEl.innerHTML = '';
  const cols = Math.max(...rows.map(r => r.length));
  rows.forEach(row => {
    const r = document.createElement('div');
    r.className = 'row';
    r.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
    row.forEach(key => {
      const k = document.createElement('div');
      k.className = 'key' + (
        ['=', 'or', '\\u21d2', '\\u2234'].includes(key.l) ? ' accent' : '');
      k.textContent = key.l;
      k.onclick = () => smartInsert(key);
      r.appendChild(k);
    });
    // pad if fewer cols than max
    for (let i = row.length; i < cols; i++) {
      const ph = document.createElement('div');
      r.appendChild(ph);
    }
    gridEl.appendChild(r);
  });
}

// ---------- number row ----------
const numrowEl = document.getElementById('numrow');
LAYOUT.always_visible.digits.forEach(d => {
  const k = document.createElement('div');
  k.className = 'key';
  k.textContent = d;
  k.onclick = () => smartInsert({ l: d, i: d === '(-)' ? '(-)' : d });
  numrowEl.appendChild(k);
});

// ---------- actions ----------
document.querySelectorAll('.action').forEach(a => {
  a.onclick = () => {
    const act = a.dataset.action;
    if (act === 'backspace') {
      textEl.focus();
      document.execCommand('delete');
      refreshNumbers();
    } else if (act === 'newline') {
      insertAtCaret('\\n');
    } else if (act === 'done') {
      submit();
    } else if (act === 'collapse') {
      state.collapsed = !state.collapsed;
      gridEl.classList.toggle('collapsed', state.collapsed);
      a.innerHTML = state.collapsed ? '&#9650; Expand' : '&#9660; Collapse';
    }
  };
});

// ---------- submit ----------
const resultEl = document.getElementById('result');
function submit() {
  const lines = textEl.innerText
    .split('\\n')
    .map(l => l.trim())
    .filter(l => l.length > 0);

  // Expand /cmds into tokens (mirror of Python slash_commands.py)
  const CMD = {
    or: ' or ', then: ' \\u2234 ', because: ' \\u2235 ',
    implies: ' \\u21d2 ', iff: ' \\u21d4 ',
    sqrt: 'sqrt', cbrt: 'cbrt', pi: 'pi',
    theta: 'theta', alpha: 'alpha', beta: 'beta',
    deg: 'deg', pm: '+/-',
    le: '<=', ge: '>=', ne: '!=',
    in: ' in ', R: 'R', sum: 'sum', inf: 'inf',
    dy: 'dy/dx', dx: 'd/dx', lim: 'lim', to: '->',
    angle: 'angle ', tri: 'triangle ',
    para: 'parallel ', perp: 'perp ',
  };
  const expand = s => s.replace(/([A-Za-z][A-Za-z0-9_]*)(?![A-Za-z0-9_])/g,
    (m, name) => CMD[name] || m);

  const expanded = lines.map(l =>
    l.replace(/\\/([A-Za-z][A-Za-z0-9_]*)/g, (m, name) => CMD[name] || m));

  const payload = {
    learner_lines: expanded,
    tab_used: state.activeTab,
    steps: expanded.length,
  };

  resultEl.classList.add('shown');
  resultEl.innerHTML = `
    <div class='label'>submitted to step_grader</div>
    <pre>${JSON.stringify(payload, null, 2)}</pre>
  `;
}

// focus working on load
textEl.focus();
renderGrid();
""")
A("</script>")

A("</body>")
A("</html>")

HTML_PATH.parent.mkdir(parents=True, exist_ok=True)
HTML_PATH.write_text("\n".join(HTML), encoding="utf-8")
print(f"  ✓ {HTML_PATH.relative_to(ROOT)}")
print(f"    {HTML_PATH.stat().st_size:,} bytes")

print()
print("=" * 72)
print("OPEN THIS FILE IN A BROWSER")
print("=" * 72)
print(f"  {HTML_PATH}")
print()
print("  What you'll see:")
print("    - Phone-width preview with the MethodMarker top bar")
print("    - Working area with numbered lines (adds a number per Enter)")
print("    - Six tabs: Algebra · Calculus · Trig · Stats · Geometry · Greek")
print("    - Tap any symbol key to insert it at the cursor")
print("    - 'or' and '=' auto-space; () puts cursor inside")
print("    - Numbers row always visible")
print("    - Collapse hides the symbol grid")
print("    - Done expands /slash-commands and prints the learner_lines JSON")
print()
print("  Try typing: solve (x+4)(x-3)=0 then Done")
````

````python
# 1. Print the notebook content here (readable) 
# 2. Convert to standalone HTML for browser viewing

import json
from pathlib import Path
from html import escape

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
NB = ROOT / "docs" / "MethodKeyboard.ipynb"
HTML_OUT = ROOT / "docs" / "MethodKeyboard.html"

nb = json.loads(NB.read_text(encoding="utf-8"))
cells = nb.get("cells", [])

# ============================================================
# 1. Print the notebook as readable text in this cell
# ============================================================

print("=" * 72)
print("MethodKeyboard.ipynb — RENDERED VIEW")
print("=" * 72)
print()

for i, cell in enumerate(cells):
    src = "".join(cell.get("source", []))
    ctype = cell.get("cell_type")
    if ctype == "markdown":
        print(src)
        print()
        print("─" * 72)
        print()
    elif ctype == "code":
        print(f"┌─ [code cell {i}] ───────────────────────────────────")
        for line in src.splitlines():
            print(f"│ {line}")
        print(f"└──────────────────────────────────────────────────────")
        print()

# ============================================================
# 2. Generate standalone HTML viewable in browser
# ============================================================

# Tiny markdown → html (headings, tables, code fences, bold, inline code)
def md_to_html(md_text):
    lines = md_text.splitlines()
    out = []
    in_code = False
    in_table = False
    table_rows = []

    def close_table():
        nonlocal in_table, table_rows
        if not in_table:
            return
        html = ["<table>"]
        for i, row in enumerate(table_rows):
            tag = "th" if i == 0 else "td"
            html.append("<tr>")
            for cell in row:
                html.append(f"<{tag}>{cell}</{tag}>")
            html.append("</tr>")
        html.append("</table>")
        out.append("".join(html))
        in_table = False
        table_rows = []

    for line in lines:
        if line.startswith("```"):
            if in_code:
                out.append("</pre>")
                in_code = False
            else:
                out.append("<pre>")
                in_code = True
            continue
        if in_code:
            out.append(escape(line))
            continue
        # table rows
        if line.strip().startswith("|") and line.strip().endswith("|"):
            cells = [c.strip() for c in line.strip().strip("|").split("|")]
            if all(set(c) <= set("- :") for c in cells):
                continue  # separator row
            table_rows.append([escape(c) for c in cells])
            in_table = True
            continue
        else:
            close_table()
        if not line.strip():
            out.append("")
            continue
        # headings
        if line.startswith("# "):
            out.append(f"<h1>{escape(line[2:])}</h1>")
        elif line.startswith("## "):
            out.append(f"<h2>{escape(line[3:])}</h2>")
        elif line.startswith("### "):
            out.append(f"<h3>{escape(line[4:])}</h3>")
        else:
            # inline code + bold
            import re
            line = re.sub(r"`([^`]+)`", r"<code>\1</code>", escape(line))
            line = re.sub(r"\*\*([^*]+)\*\*", r"<strong>\1</strong>", line)
            out.append(f"<p>{line}</p>")
    close_table()
    if in_code:
        out.append("</pre>")
    return "\n".join(out)


body_parts = []
for cell in cells:
    src = "".join(cell.get("source", []))
    ctype = cell.get("cell_type")
    if ctype == "markdown":
        body_parts.append(md_to_html(src))
    elif ctype == "code":
        body_parts.append("<pre class='code'>" + escape(src) + "</pre>")

body = "\n".join(body_parts)

HTML = f"""<!doctype html>
<html lang="en">
<head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width,initial-scale=1">
<title>MethodKeyboard — design</title>
<style>
  body {{
    margin: 0; background: #f6f8fd; color: #0b1f44;
    font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif;
    line-height: 1.6; padding: 32px 16px;
  }}
  .container {{
    max-width: 780px; margin: 0 auto; background: #ffffff;
    padding: 40px 48px; border-radius: 12px;
    box-shadow: 0 2px 16px rgba(20, 40, 90, .06);
  }}
  h1 {{ font-size: 28px; margin: 0 0 8px; color: #0b1f44;
       border-bottom: 2px solid #e4e9f2; padding-bottom: 12px; }}
  h2 {{ font-size: 20px; margin: 40px 0 12px; color: #223a6b;
       border-bottom: 1px solid #e4e9f2; padding-bottom: 6px; }}
  h3 {{ font-size: 16px; margin: 24px 0 8px; color: #223a6b; }}
  p {{ margin: 8px 0; }}
  code {{
    background: #eef2fb; color: #223a6b; padding: 2px 6px;
    border-radius: 4px; font-family: 'SF Mono', Menlo, Consolas, monospace;
    font-size: 90%;
  }}
  pre {{
    background: #0b1f44; color: #cfe2ff; padding: 16px 20px;
    border-radius: 8px; overflow-x: auto; font-size: 12px;
    font-family: 'SF Mono', Menlo, Consolas, monospace;
    line-height: 1.5; margin: 12px 0;
  }}
  pre.code {{
    background: #f6f8fd; color: #223a6b;
    border: 1px solid #e4e9f2;
  }}
  table {{
    border-collapse: collapse; margin: 12px 0; width: 100%;
    font-size: 14px;
  }}
  th, td {{
    text-align: left; padding: 8px 12px;
    border-bottom: 1px solid #e4e9f2;
  }}
  th {{ background: #f6f8fd; color: #223a6b; font-weight: 600; }}
  tr:hover {{ background: #fafbff; }}
  strong {{ color: #223a6b; }}
</style>
</head>
<body>
<div class="container">
{body}
</div>
</body>
</html>
"""

HTML_OUT.write_text(HTML, encoding="utf-8")
print()
print("=" * 72)
print("HTML VIEW WRITTEN")
print("=" * 72)
print(f"  Path  : {HTML_OUT}")
print(f"  Size  : {HTML_OUT.stat().st_size:,} bytes")
print()
print("  Double-click to open in browser — same experience as the")
print("  keyboard mockup. Rendered headings, tables, code blocks.")
````

````python
# Regenerate keyboard_layout.json (v2 with base_layer) then audit

import json
from pathlib import Path

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
TUTOR = ROOT / "data" / "processed" / "tutor"
LAYOUT_PATH = TUTOR / "keyboard_layout.json"


def K(l, i=None, sp=False, trail=False, skip=None):
    d = {"l": l}
    d["i"] = i if i is not None else l
    if sp:    d["sp"] = True
    if trail: d["trail"] = True
    if skip:  d["skip"] = skip
    return d


BASE_VARIABLES = [K(c) for c in "xyzabcnkmpqt"]
BASE_DIGITS = [K("1"), K("2"), K("3"), K("4"), K("5"), K("6"),
               K("7"), K("8"), K("9"), K("0"), K("."), K("(−)", "(-)")]
BASE_OPS = [
    K("+", " + ", sp=True), K("−", " - ", sp=True),
    K("×", " x ", sp=True), K("÷", " / ", sp=True),
    K("=", " = ", sp=True), K("(", "("), K(")", ")"),
    K("^", "^"), K("√", "sqrt"), K(";", "; ", trail=True),
]
BASE_ACTIONS = [
    K("⌫", "__back__"), K("Space", " "),
    K("↵", "__newline__"), K("✓ Done", "__done__"),
    K("▼", "__collapse__"),
]

ALG = [
    [K("or", " or ", sp=True), K("⇒", " ⇒ ", sp=True),
     K("⇔", " ⇔ ", sp=True), K("∴", " ∴ ", sp=True),
     K("∵", " ∵ ", sp=True), K("≠", "!=")],
    [K("x²", "x^2"), K("xⁿ", "x^"), K("x⁻¹", "x^-1"),
     K("∛", "cbrt"), K("ⁿ√", "nroot"), K("a/b", "a/b")],
    [K("<"), K(">"), K("≤", "<="), K("≥", ">="),
     K("|x|", "abs(x)"), K("Σ", "sum")],
    [K("∈", " in "), K("ℝ", "R"), K("π", "pi"),
     K("∞", "inf"), K("%", "%"), K(",", ",")],
]

CALC = [
    [K("dy/dx", "dy/dx"), K("d/dx", "d/dx"),
     K("d²y/dx²", "d2y/dx2"), K("Dₓ[ ]", "Dx[]"),
     K("f'(x)", "f'(x)"), K("f''(x)", "f''(x)")],
    [K("lim", "lim("), K("→", "->"), K("h→0", "h->0"),
     K("x→∞", "x->inf"), K("Δ", "Delta"), K("∞", "inf")],
    [K("e", "e"), K("ln", "ln("), K("log", "log("),
     K("∴", " ∴ ", sp=True), K("⇒", " ⇒ ", sp=True),
     K("( )", "()", skip=1)],
]

TRIG = [
    [K("sin", "sin"), K("cos", "cos"), K("tan", "tan"),
     K("sin⁻¹", "asin"), K("cos⁻¹", "acos"), K("tan⁻¹", "atan")],
    [K("θ", "theta"), K("α", "alpha"), K("β", "beta"),
     K("γ", "gamma"), K("π", "pi"), K("°", "deg")],
    [K("2θ", "2theta"), K("½θ", "theta/2"),
     K("sin²", "sin^2"), K("cos²", "cos^2"),
     K("tan²", "tan^2"), K("rad", "rad")],
    [K("or", " or ", sp=True), K("⇒", " ⇒ ", sp=True),
     K("∴", " ∴ ", sp=True), K("±", "+/-"),
     K("( )", "()", skip=1), K("∈", " in ")],
]

STATS = [
    [K("x̄", "xbar"), K("σ", "sigma"), K("σ²", "sigma^2"),
     K("n", "n"), K("N", "N"), K("Σ", "sum")],
    [K("Σx", "sum(x)"), K("Σx²", "sum(x^2)"),
     K("xᵢ", "x_i"), K("fᵢ", "f_i"),
     K("Q1", "Q1"), K("Q3", "Q3")],
    [K("Q2", "Q2"), K("min", "min"), K("max", "max"),
     K("P( )", "P()", skip=1), K("A|B", "A|B"), K("%", "%")],
    [K("∪", " union "), K("∩", " intersect "),
     K("=", " = ", sp=True), K("∴", " ∴ ", sp=True),
     K("( )", "()", skip=1), K(",", ",")],
]

GEO = [
    [K("∠", "angle "), K("△", "triangle "),
     K("∥", "parallel "), K("⊥", "perp "),
     K("≅", "cong "), K("∼", "similar ")],
    [K("Â", "A^"), K("B̂", "B^"), K("Ĉ", "C^"),
     K("D̂", "D^"), K("Ê", "E^"), K("F̂", "F^")],
    [K("tan ⊥ rad", "tan ⊥ rad"),
     K("tan chord", "tan chord"),
     K("2 tans", "2 tans from same pt"),
     K("∠ centre", "∠ at centre = 2×∠ at circumf"),
     K("∠ semicircle", "∠ in semicircle = 90°"),
     K("same seg", "∠s in same segment")],
    [K("cyc quad opp", "opp ∠s of cyclic quad"),
     K("cyc quad ext", "ext ∠ of cyclic quad"),
     K("∠ = chord", "equal chords subtend equal ∠s"),
     K("midpt chord", "line from centre to midpt of chord"),
     K("⊥ chord", "line from centre ⊥ to chord"),
     K("perp bisect", "perp bisector of chord")],
    [K("midpt thm", "midpoint theorem"),
     K("prop thm", "prop theorem; line ∥ one side of Δ"),
     K("||| Δs", "similar Δs"),
     K("Pyth", "Pythagoras"),
     K("ext ∠ Δ", "ext ∠ of Δ"),
     K("sum ∠ Δ", "sum of ∠s in Δ")],
    [K("∴", " ∴ ", sp=True), K("∵", " ∵ ", sp=True),
     K("⇒", " ⇒ ", sp=True), K("given", "given"),
     K("common", "common"), K("vert opp", "vert opp ∠s")],
    [K("alt ∠s", "alt ∠s"), K("co-int ∠s", "co-int ∠s"),
     K("corresp ∠s", "corresp ∠s"),
     K("arc", "arc "), K("chord", "chord "),
     K("rad", "rad ")],
]

GREEK = [
    [K("α", "alpha"), K("β", "beta"), K("γ", "gamma"),
     K("δ", "delta"), K("ε", "epsilon"), K("θ", "theta")],
    [K("λ", "lambda"), K("μ", "mu"), K("π", "pi"),
     K("ρ", "rho"), K("σ", "sigma"), K("τ", "tau")],
    [K("φ", "phi"), K("ψ", "psi"), K("ω", "omega"),
     K("Δ", "Delta"), K("Σ", "Sigma"), K("Ω", "Omega")],
    [K("⇒", " ⇒ ", sp=True), K("∴", " ∴ ", sp=True),
     K("∈", " in "), K("∞", "inf"),
     K("≈", "approx"), K("≠", "!=")],
]

KEYBOARD = {
    "schema_version": 2,
    "tabs_order": ["Algebra", "Calculus", "Trig", "Stats", "Geometry", "Greek"],
    "base_layer": {
        "variables": BASE_VARIABLES,
        "digits": BASE_DIGITS,
        "operators": BASE_OPS,
        "actions": BASE_ACTIONS,
    },
    "tabs": {
        "Algebra": ALG, "Calculus": CALC, "Trig": TRIG,
        "Stats": STATS, "Geometry": GEO, "Greek": GREEK,
    },
    "settings": {
        "line_cap": None,
        "tab_auto_switch": False,
        "keyboard_collapsible": True,
        "keyboard_default": "expanded",
    },
    "tab_auto_switch_map": {
        "calculus": "Calculus", "trig": "Trig",
        "stats": "Stats", "probability": "Stats",
        "geometry": "Geometry", "analytical": "Geometry",
        "algebra": "Algebra", "sequences": "Algebra",
        "functions": "Algebra", "finance": "Algebra",
    },
}

LAYOUT_PATH.write_text(
    json.dumps(KEYBOARD, indent=2, ensure_ascii=False), encoding="utf-8")

n_tab = sum(len(r) for tab in KEYBOARD["tabs"].values() for r in tab)
n_base = (len(BASE_VARIABLES) + len(BASE_DIGITS)
          + len(BASE_OPS) + len(BASE_ACTIONS))
print(f"  ✓ keyboard_layout.json regenerated (v2)")
print(f"    schema_version  : 2")
print(f"    tab keys        : {n_tab}")
print(f"    base layer keys : {n_base}")
print(f"    total           : {n_tab + n_base}")
print()

# Sanity check it loads with the expected shape
kb = json.loads(LAYOUT_PATH.read_text(encoding="utf-8"))
print(f"  Keys in file    : {list(kb.keys())}")
print(f"  base_layer keys : {list(kb['base_layer'].keys())}")
print(f"  tabs            : {list(kb['tabs'].keys())}")
````

````python
# Rebuild HTML from the v2 layout + open in browser

import json
import webbrowser
from pathlib import Path

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
LAYOUT_PATH = ROOT / "data" / "processed" / "tutor" / "keyboard_layout.json"
HTML_PATH = ROOT / "docs" / "keyboard_preview.html"

KEYBOARD = json.loads(LAYOUT_PATH.read_text(encoding="utf-8"))


HTML = r"""<!doctype html>
<html lang="en">
<head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width,initial-scale=1">
<title>MethodKeyboard — preview v2</title>
<style>
  * { box-sizing: border-box; -webkit-tap-highlight-color: transparent; }
  body {
    margin: 0; background: #eef2f8; font-family: -apple-system,
    BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif;
    display: flex; justify-content: center; padding: 12px;
    min-height: 100vh;
  }
  .phone {
    width: 420px; max-width: 100%; background: #fff;
    border-radius: 20px; overflow: hidden;
    box-shadow: 0 8px 32px rgba(20, 40, 90, .12);
    display: flex; flex-direction: column;
  }
  .topbar {
    display: flex; align-items: center; gap: 10px;
    padding: 12px 16px; background: #f6f8fd;
    border-bottom: 1px solid #e4e9f2;
  }
  .topbar .logo {
    width: 28px; height: 28px; border-radius: 7px;
    background: linear-gradient(135deg, #3b5bf6, #5b8cf6);
    display: flex; align-items: center; justify-content: center;
    color: #fff; font-weight: 700; font-size: 15px;
  }
  .topbar .brand { font-weight: 700; font-size: 14px; color: #0b1f44; }
  .topbar .tag { font-size: 10px; color: #6b7a99; }

  .working { padding: 12px 16px; background: #fff; }
  .working .head {
    display: flex; justify-content: space-between;
    align-items: center; font-size: 12px;
    color: #465a80; font-weight: 600; margin-bottom: 6px;
  }
  .working .steps { color: #6b7a99; font-weight: 500; }
  .lines {
    display: grid; grid-template-columns: 22px 1fr;
    column-gap: 8px; font-size: 15px; line-height: 26px;
    color: #0b1f44; max-height: 160px; overflow-y: auto;
  }
  .nums { color: #9aa8c2; font-size: 11px; text-align: right;
          user-select: none; padding-top: 2px; }
  .nums div { height: 26px; line-height: 26px; }
  .text { white-space: pre-wrap; word-break: break-word;
          min-height: 26px; outline: none; padding: 0;
          font-family: 'SF Mono', Menlo, Consolas, monospace; }

  .tabs {
    display: flex; overflow-x: auto; gap: 4px;
    padding: 6px 12px; background: #f6f8fd;
    border-top: 1px solid #e4e9f2;
    border-bottom: 1px solid #e4e9f2;
    scrollbar-width: none;
  }
  .tabs::-webkit-scrollbar { display: none; }
  .tab {
    flex: none; padding: 6px 12px; border-radius: 16px;
    font-size: 11px; font-weight: 600; color: #465a80;
    background: transparent; border: none; cursor: pointer;
    white-space: nowrap;
  }
  .tab.active { background: #3b5bf6; color: #fff; }

  .grid { padding: 8px 10px; background: #fff;
          display: flex; flex-direction: column; gap: 4px;
          transition: max-height .25s ease; overflow: hidden;
          max-height: 900px; }
  .grid.collapsed { max-height: 0; padding: 0; }
  .row { display: grid; gap: 4px; }
  .key {
    padding: 9px 3px; border-radius: 8px;
    background: #fff; border: 1px solid #dfe5ef;
    font-size: 13px; color: #0b1f44; cursor: pointer;
    text-align: center; user-select: none; font-family: inherit;
    transition: transform .05s, background .1s;
    min-height: 38px; display: flex; align-items: center;
    justify-content: center;
  }
  .key:active { transform: scale(.95); background: #eef2fb; }
  .key.accent {
    background: #eef2fb; border-color: #c5d1e9;
    color: #223a6b; font-weight: 600;
  }

  .basegrid {
    padding: 6px 10px; background: #f6f8fd;
    border-top: 1px solid #e4e9f2;
    display: flex; flex-direction: column; gap: 4px;
  }
  .basegrid.collapsed { display: none; }

  .action {
    padding: 10px 3px; border-radius: 8px;
    background: #eef2fb; border: 1px solid #dfe5ef;
    color: #223a6b; font-size: 11px; font-weight: 600;
    cursor: pointer; font-family: inherit; text-align: center;
    min-height: 38px; display: flex; align-items: center;
    justify-content: center;
  }
  .action.primary { background: #3b5bf6; color: #fff;
                    border-color: #3b5bf6; }
  .action:active { transform: scale(.97); }

  .result {
    padding: 10px 16px; background: #0b1f44; color: #fff;
    font-family: 'SF Mono', Menlo, monospace;
    font-size: 10px; line-height: 1.5;
    max-height: 0; overflow: hidden;
    transition: max-height .25s ease;
  }
  .result.shown { max-height: 200px; overflow-y: auto; }
  .result .label { color: #9ab; margin-bottom: 4px; }
  .result pre { margin: 0; color: #cfe2ff; }
</style>
</head>
<body>

<div class="phone">
  <div class="topbar">
    <div class="logo">M</div>
    <div>
      <div class="brand">MethodMarker</div>
      <div class="tag">Better Methods. Higher Marks.</div>
    </div>
  </div>

  <div class="working">
    <div class="head">
      <span>&#9998; Working</span>
      <span class="steps" id="stepCount">0 steps</span>
    </div>
    <div class="lines">
      <div class="nums" id="nums"></div>
      <div class="text" id="text" contenteditable="true" spellcheck="false"></div>
    </div>
  </div>

  <div class="tabs" id="tabs"></div>

  <div class="grid" id="grid"></div>

  <div class="basegrid" id="basegrid"></div>

  <div class="result" id="result"></div>
</div>

<script>
const LAYOUT = __LAYOUT__;
const state = { activeTab: LAYOUT.tabs_order[0], collapsed: false };

const textEl = document.getElementById('text');
const numsEl = document.getElementById('nums');
const stepCountEl = document.getElementById('stepCount');

function refreshNumbers() {
  const lines = textEl.innerText.split('\n');
  numsEl.innerHTML = lines.map((_, i) => `<div>${i + 1}</div>`).join('');
  const n = lines.length;
  stepCountEl.textContent = (n === 1 && lines[0] === '')
    ? '0 steps' : `${n} step${n === 1 ? '' : 's'}`;
}
textEl.addEventListener('input', refreshNumbers);
refreshNumbers();

function getCaretContext() {
  const sel = window.getSelection();
  if (!sel.rangeCount) return { before: '', after: '' };
  const r = sel.getRangeAt(0);
  const pre = r.cloneRange(); pre.selectNodeContents(textEl);
  pre.setEnd(r.startContainer, r.startOffset);
  const before = pre.toString();
  const post = r.cloneRange(); post.selectNodeContents(textEl);
  post.setStart(r.endContainer, r.endOffset);
  const after = post.toString();
  return { before, after };
}

function insertAtCaret(str) {
  textEl.focus();
  const sel = window.getSelection();
  if (!sel.rangeCount) return;
  const r = sel.getRangeAt(0);
  r.deleteContents();
  const node = document.createTextNode(str);
  r.insertNode(node);
  r.setStartAfter(node);
  r.setEndAfter(node);
  sel.removeAllRanges();
  sel.addRange(r);
  refreshNumbers();
}

function smartInsert(key) {
  if (key.i === '__back__') {
    textEl.focus();
    document.execCommand('delete');
    refreshNumbers();
    return;
  }
  if (key.i === '__newline__') { insertAtCaret('\n'); return; }
  if (key.i === '__done__')    { submit(); return; }
  if (key.i === '__collapse__') {
    state.collapsed = !state.collapsed;
    document.getElementById('grid').classList.toggle('collapsed', state.collapsed);
    document.getElementById('basegrid').classList.toggle('collapsed', state.collapsed);
    return;
  }
  const { before, after } = getCaretContext();
  let ins = key.i;
  if (key.sp) {
    const needLead  = before.length > 0 && !/\s$/.test(before);
    const needTrail = after.length  > 0 && !/^\s/.test(after);
    ins = (needLead ? ' ' : '') + ins.trim() + (needTrail ? ' ' : '');
  }
  insertAtCaret(ins);
  if (key.skip === 1 && key.i === '()') {
    const sel = window.getSelection();
    if (sel.rangeCount) {
      const r = sel.getRangeAt(0);
      if (r.startContainer.nodeType === 3 && r.startOffset > 0) {
        const nr = r.cloneRange();
        nr.setStart(nr.startContainer, nr.startOffset - 1);
        nr.collapse(true);
        sel.removeAllRanges(); sel.addRange(nr);
      }
    }
  }
}

// ---------- tabs ----------
const tabsEl = document.getElementById('tabs');
LAYOUT.tabs_order.forEach(name => {
  const b = document.createElement('button');
  b.className = 'tab' + (name === state.activeTab ? ' active' : '');
  b.textContent = name;
  b.onclick = () => {
    state.activeTab = name;
    document.querySelectorAll('.tab').forEach(t =>
      t.classList.toggle('active', t.textContent === name));
    renderGrid();
  };
  tabsEl.appendChild(b);
});

// ---------- tab grid ----------
const gridEl = document.getElementById('grid');
function renderGrid() {
  const rows = LAYOUT.tabs[state.activeTab] || [];
  gridEl.innerHTML = '';
  const cols = Math.max(...rows.map(r => r.length));
  rows.forEach(row => {
    const r = document.createElement('div');
    r.className = 'row';
    r.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
    row.forEach(key => {
      const k = document.createElement('div');
      const accent = ['or','=','⇒','∴','∵'].includes(key.l) ? ' accent' : '';
      k.className = 'key' + accent;
      k.textContent = key.l;
      k.onclick = () => smartInsert(key);
      r.appendChild(k);
    });
    for (let i = row.length; i < cols; i++) r.appendChild(document.createElement('div'));
    gridEl.appendChild(r);
  });
}

// ---------- base layer ----------
function renderBase() {
  const b = LAYOUT.base_layer;
  const container = document.getElementById('basegrid');
  container.innerHTML = '';

  const layout = [
    { keys: b.variables, cols: 12 },
    { keys: b.digits,    cols: 12 },
    { keys: b.operators, cols: 10 },
    { keys: b.actions,   cols: 5  },
  ];

  layout.forEach(({keys, cols}) => {
    const row = document.createElement('div');
    row.className = 'row';
    row.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
    keys.forEach(key => {
      const isDone = key.i === '__done__';
      const isAction = key.i && key.i.startsWith('__');
      const el = document.createElement('div');
      el.className = isDone ? 'action primary' : (isAction ? 'action' : 'key');
      el.textContent = key.l;
      el.onclick = () => smartInsert(key);
      row.appendChild(el);
    });
    container.appendChild(row);
  });
}

// ---------- submit ----------
const resultEl = document.getElementById('result');
const CMD = {
  or: ' or ', then: ' ∴ ', because: ' ∵ ',
  implies: ' ⇒ ', iff: ' ⇔ ',
  sqrt: 'sqrt', cbrt: 'cbrt', pi: 'pi',
  theta: 'theta', alpha: 'alpha', beta: 'beta',
  deg: 'deg', pm: '+/-',
  le: '<=', ge: '>=', ne: '!=',
  in: ' in ', R: 'R', sum: 'sum', inf: 'inf',
  dy: 'dy/dx', dx: 'd/dx', lim: 'lim', to: '->',
  angle: 'angle ', tri: 'triangle ',
  para: 'parallel ', perp: 'perp ',
};
function submit() {
  const lines = textEl.innerText.split('\n').map(l => l.trim()).filter(l => l.length);
  const expanded = lines.map(l =>
    l.replace(/\/([A-Za-z][A-Za-z0-9_]*)/g, (m, name) => CMD[name] || m));
  const payload = {
    learner_lines: expanded,
    tab_used: state.activeTab,
    steps: expanded.length,
  };
  resultEl.classList.add('shown');
  resultEl.innerHTML =
    "<div class='label'>submitted to step_grader</div>" +
    "<pre>" + JSON.stringify(payload, null, 2) + "</pre>";
}

textEl.focus();
renderGrid();
renderBase();
</script>
</body>
</html>
"""

HTML = HTML.replace("__LAYOUT__", json.dumps(KEYBOARD, ensure_ascii=False))
HTML_PATH.write_text(HTML, encoding="utf-8")
print(f"  ✓ rebuilt {HTML_PATH.name}  ({HTML_PATH.stat().st_size:,} bytes)")

# Open in browser
webbrowser.open(HTML_PATH.as_uri())
print(f"  → opened in default browser")
print(f"  → {HTML_PATH.as_uri()}")
print()

# Layout summary
print("=" * 72)
print("WHAT YOU SHOULD SEE")
print("=" * 72)
print()
print("  Top: MethodMarker bar")
print("  Then: Working area (numbered lines, step counter)")
print("  Then: 6 tabs — Algebra · Calculus · Trig · Stats · Geometry · Greek")
print("  Then: Tab-specific symbols (changes per tab)")
print("  Then: BASE LAYER — always visible on every tab:")
b = KEYBOARD["base_layer"]
print(f"    Row 1 (variables):  {[k['l'] for k in b['variables']]}")
print(f"    Row 2 (digits):     {[k['l'] for k in b['digits']]}")
print(f"    Row 3 (operators):  {[k['l'] for k in b['operators']]}")
print(f"    Row 4 (actions):    {[k['l'] for k in b['actions']]}")
print()
print("  Geometry tab has 7 rows including 14 SA CAPS theorem shortcuts.")
````

````python
# Rebuild keyboard v3 — unicode display, ABC mode, no theorems, 
# fingerprint-derived variables

import json
import re
import webbrowser
from collections import Counter
from pathlib import Path

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
TEMPLATE_DIR = ROOT / "data" / "processed" / "tutor" / "question_templates"
LAYOUT_PATH = ROOT / "data" / "processed" / "tutor" / "keyboard_layout.json"
HTML_PATH = ROOT / "docs" / "keyboard_preview.html"


# ============================================================
# 1. Extract popular variables from the corpus
# ============================================================

# Single-letter variables used in prompts
var_pattern = re.compile(r'\b([a-z])\b')
var_hits = Counter()
exclude = set("aeiou")  # drop vowels used as articles

for p in sorted(TEMPLATE_DIR.glob("template_*.json")):
    t = json.loads(p.read_text(encoding="utf-8"))
    def scan(obj):
        if isinstance(obj, str):
            for m in var_pattern.findall(obj.lower()):
                if m not in exclude:
                    var_hits[m] += 1
        elif isinstance(obj, dict):
            for v in obj.values(): scan(v)
        elif isinstance(obj, list):
            for v in obj: scan(v)
    scan(t)

# Prefer the classic math letters, then fill from the corpus
CLASSIC = list("xyzabcnkmpqt")
popular_vars = []
for c in CLASSIC:
    popular_vars.append(c)
# Add corpus top ones not already there
for v, _ in var_hits.most_common(30):
    if v not in popular_vars and v.isalpha():
        popular_vars.append(v)
    if len(popular_vars) >= 15:
        break

print("=" * 72)
print("VARIABLES EXTRACTED FROM CORPUS")
print("=" * 72)
print("  Top 15 used single-letter variables (excluding vowels):")
for v, n in var_hits.most_common(15):
    print(f"    {v}  x{n}")
print()
print(f"  Popular set for base layer: {popular_vars[:12]}")
print()


# ============================================================
# 2. Key builder — display vs engine
# ============================================================

def K(label, display=None, engine=None, sp=False, trail=False, skip=None):
    """A key:
       label: what shows on the button
       display: what gets inserted into working area (may be unicode)
       engine: what the engine receives (if None, same as display)
    """
    d = {"l": label}
    d["d"] = display if display is not None else label
    d["e"] = engine if engine is not None else d["d"]
    if sp:    d["sp"] = True
    if trail: d["trail"] = True
    if skip:  d["skip"] = skip
    return d


# ============================================================
# 3. Base layer
# ============================================================

BASE_VARIABLES = [K(v, v, v) for v in popular_vars[:12]]

BASE_DIGITS = [
    K("1", "1", "1"), K("2", "2", "2"), K("3", "3", "3"),
    K("4", "4", "4"), K("5", "5", "5"), K("6", "6", "6"),
    K("7", "7", "7"), K("8", "8", "8"), K("9", "9", "9"),
    K("0", "0", "0"), K(".", ".", "."), K("(−)", "(-)", "(-)"),
]

BASE_OPS = [
    K("+", " + ", " + ", sp=True),
    K("−", " - ", " - ", sp=True),
    K("×", " × ", " x ", sp=True),
    K("÷", " ÷ ", " / ", sp=True),
    K("=", " = ", " = ", sp=True),
    K("(", "(", "("), K(")", ")", ")"),
    K("^", "^", "^"),
    K("√", "√", "sqrt"),
    K(";", "; ", "; ", trail=True),
]

BASE_ACTIONS = [
    K("⌫", "__back__", "__back__"),
    K("Space", " ", " "),
    K("↵", "__newline__", "__newline__"),
    K("✓ Done", "__done__", "__done__"),
    K("ABC", "__abc__", "__abc__"),
    K("▼", "__collapse__", "__collapse__"),
]

# QWERTY alphabet — used when ABC mode is on
BASE_ABC = [
    [K("q","q","q"), K("w","w","w"), K("e","e","e"), K("r","r","r"),
     K("t","t","t"), K("y","y","y"), K("u","u","u"), K("i","i","i"),
     K("o","o","o"), K("p","p","p")],
    [K("a","a","a"), K("s","s","s"), K("d","d","d"), K("f","f","f"),
     K("g","g","g"), K("h","h","h"), K("j","j","j"), K("k","k","k"),
     K("l","l","l")],
    [K("z","z","z"), K("x","x","x"), K("c","c","c"), K("v","v","v"),
     K("b","b","b"), K("n","n","n"), K("m","m","m"), K("'","'","'"),
     K("-","-","-")],
]


# ============================================================
# 4. Tab-specific symbol grids
# ============================================================

ALG = [
    [K("or", " or ", " or ", sp=True),
     K("⇒", " ⇒ ", " => ", sp=True),
     K("⇔", " ⇔ ", " <=> ", sp=True),
     K("∴", " ∴ ", " therefore ", sp=True),
     K("∵", " ∵ ", " because ", sp=True),
     K("≠", "≠", "!=")],
    [K("x²", "x²", "x^2"),
     K("xⁿ", "xⁿ", "x^"),
     K("x⁻¹", "x⁻¹", "x^-1"),
     K("∛", "∛", "cbrt"),
     K("ⁿ√", "ⁿ√", "nroot"),
     K("a/b", "a/b", "a/b")],
    [K("<", "<", "<"), K(">", ">", ">"),
     K("≤", "≤", "<="), K("≥", "≥", ">="),
     K("|x|", "|x|", "abs(x)"),
     K("Σ", "Σ", "sum")],
    [K("∈", "∈", " in "),
     K("ℝ", "ℝ", "R"),
     K("π", "π", "pi"),
     K("∞", "∞", "inf"),
     K("%", "%", "%"),
     K(",", ",", ",")],
]

CALC = [
    [K("dy/dx","dy/dx","dy/dx"), K("d/dx","d/dx","d/dx"),
     K("d²y/dx²","d²y/dx²","d2y/dx2"), K("Dₓ[ ]","Dₓ[ ]","Dx[]"),
     K("f'(x)","f'(x)","f'(x)"), K("f''(x)","f''(x)","f''(x)")],
    [K("lim","lim","lim("), K("→","→","->"),
     K("h→0","h→0","h->0"), K("x→∞","x→∞","x->inf"),
     K("Δ","Δ","Delta"), K("∞","∞","inf")],
    [K("e","e","e"), K("ln","ln","ln("), K("log","log","log("),
     K("∴"," ∴ "," therefore ", sp=True),
     K("⇒"," ⇒ "," => ", sp=True),
     K("( )","()","()", skip=1)],
]

TRIG = [
    [K("sin","sin","sin"), K("cos","cos","cos"), K("tan","tan","tan"),
     K("sin⁻¹","sin⁻¹","asin"), K("cos⁻¹","cos⁻¹","acos"),
     K("tan⁻¹","tan⁻¹","atan")],
    [K("θ","θ","theta"), K("α","α","alpha"), K("β","β","beta"),
     K("γ","γ","gamma"), K("π","π","pi"), K("°","°","°")],
    [K("2θ","2θ","2theta"), K("½θ","θ/2","theta/2"),
     K("sin²","sin²","sin^2"), K("cos²","cos²","cos^2"),
     K("tan²","tan²","tan^2"), K("rad","rad","rad")],
    [K("or"," or "," or ", sp=True),
     K("⇒"," ⇒ "," => ", sp=True),
     K("∴"," ∴ "," therefore ", sp=True),
     K("±","±","+/-"),
     K("( )","()","()", skip=1),
     K("∈","∈"," in ")],
]

STATS = [
    [K("x̄","x̄","xbar"), K("σ","σ","sigma"), K("σ²","σ²","sigma^2"),
     K("n","n","n"), K("N","N","N"), K("Σ","Σ","sum")],
    [K("Σx","Σx","sum(x)"), K("Σx²","Σx²","sum(x^2)"),
     K("xᵢ","xᵢ","x_i"), K("fᵢ","fᵢ","f_i"),
     K("Q1","Q1","Q1"), K("Q3","Q3","Q3")],
    [K("Q2","Q2","Q2"), K("min","min","min"), K("max","max","max"),
     K("P( )","P()","P()", skip=1),
     K("A|B","A|B","A|B"), K("%","%","%")],
    [K("∪","∪"," union "), K("∩","∩"," intersect "),
     K("="," = "," = ", sp=True),
     K("∴"," ∴ "," therefore ", sp=True),
     K("( )","()","()", skip=1),
     K(",",",",",")],
]

# ------------------------------------------------------------
# GEOMETRY — primitives and connectors ONLY.
# No theorems. Learner must know and write them.
# ------------------------------------------------------------
GEO = [
    [K("∠","∠","angle "),
     K("△","△","triangle "),
     K("∥","∥","parallel "),
     K("⊥","⊥","perp "),
     K("≅","≅","cong "),
     K("∼","∼","similar ")],
    [K("Â","Â","A^"), K("B̂","B̂","B^"),
     K("Ĉ","Ĉ","C^"), K("D̂","D̂","D^"),
     K("Ê","Ê","E^"), K("F̂","F̂","F^")],
    [K("∴"," ∴ "," therefore ", sp=True),
     K("∵"," ∵ "," because ", sp=True),
     K("⇒"," ⇒ "," => ", sp=True),
     K("given","given","given"),
     K("common","common","common"),
     K("vert opp","vert opp ∠s","vert opp angles")],
    [K("alt ∠s","alt ∠s","alternate angles"),
     K("co-int ∠s","co-int ∠s","co-interior angles"),
     K("corresp ∠s","corresp ∠s","corresponding angles"),
     K("arc","arc","arc "),
     K("chord","chord","chord "),
     K("rad","rad","rad ")],
]

GREEK = [
    [K("α","α","alpha"), K("β","β","beta"), K("γ","γ","gamma"),
     K("δ","δ","delta"), K("ε","ε","epsilon"), K("θ","θ","theta")],
    [K("λ","λ","lambda"), K("μ","μ","mu"), K("π","π","pi"),
     K("ρ","ρ","rho"), K("σ","σ","sigma"), K("τ","τ","tau")],
    [K("φ","φ","phi"), K("ψ","ψ","psi"), K("ω","ω","omega"),
     K("Δ","Δ","Delta"), K("Σ","Σ","Sigma"), K("Ω","Ω","Omega")],
    [K("⇒"," ⇒ "," => ", sp=True),
     K("∴"," ∴ "," therefore ", sp=True),
     K("∈","∈"," in "),
     K("∞","∞","inf"),
     K("≈","≈","approx"),
     K("≠","≠","!=")],
]


KEYBOARD = {
    "schema_version": 3,
    "tabs_order": ["Algebra", "Calculus", "Trig", "Stats", "Geometry", "Greek"],
    "base_layer": {
        "variables": BASE_VARIABLES,
        "digits":    BASE_DIGITS,
        "operators": BASE_OPS,
        "actions":   BASE_ACTIONS,
        "abc":       BASE_ABC,
    },
    "tabs": {
        "Algebra":  ALG, "Calculus": CALC, "Trig": TRIG,
        "Stats":    STATS, "Geometry": GEO, "Greek": GREEK,
    },
    "settings": {
        "line_cap": None,
        "tab_auto_switch": False,
        "keyboard_collapsible": True,
        "keyboard_default": "expanded",
    },
    "tab_auto_switch_map": {
        "calculus": "Calculus", "trig": "Trig",
        "stats": "Stats", "probability": "Stats",
        "geometry": "Geometry", "analytical": "Geometry",
        "algebra": "Algebra", "sequences": "Algebra",
        "functions": "Algebra", "finance": "Algebra",
    },
}

LAYOUT_PATH.write_text(
    json.dumps(KEYBOARD, indent=2, ensure_ascii=False), encoding="utf-8")
print(f"  ✓ keyboard_layout.json v3 written")


# ============================================================
# 5. Rebuild HTML — with unicode display, ABC mode, engine conversion
# ============================================================

HTML_TEMPLATE = r"""<!doctype html>
<html lang="en">
<head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width,initial-scale=1">
<title>MethodKeyboard — preview v3</title>
<style>
  * { box-sizing: border-box; -webkit-tap-highlight-color: transparent; }
  body { margin: 0; background: #eef2f8;
    font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif;
    display: flex; justify-content: center; padding: 12px; min-height: 100vh; }
  .phone { width: 420px; max-width: 100%; background: #fff;
    border-radius: 20px; overflow: hidden;
    box-shadow: 0 8px 32px rgba(20, 40, 90, .12);
    display: flex; flex-direction: column; }
  .topbar { display: flex; align-items: center; gap: 10px;
    padding: 12px 16px; background: #f6f8fd;
    border-bottom: 1px solid #e4e9f2; }
  .topbar .logo { width: 28px; height: 28px; border-radius: 7px;
    background: linear-gradient(135deg, #3b5bf6, #5b8cf6);
    display: flex; align-items: center; justify-content: center;
    color: #fff; font-weight: 700; font-size: 15px; }
  .topbar .brand { font-weight: 700; font-size: 14px; color: #0b1f44; }
  .topbar .tag { font-size: 10px; color: #6b7a99; }

  .working { padding: 12px 16px; background: #fff; }
  .working .head { display: flex; justify-content: space-between;
    align-items: center; font-size: 12px; color: #465a80;
    font-weight: 600; margin-bottom: 6px; }
  .working .steps { color: #6b7a99; font-weight: 500; }
  .lines { display: grid; grid-template-columns: 22px 1fr;
    column-gap: 8px; font-size: 15px; line-height: 26px;
    color: #0b1f44; max-height: 160px; overflow-y: auto; }
  .nums { color: #9aa8c2; font-size: 11px; text-align: right;
    user-select: none; padding-top: 2px; }
  .nums div { height: 26px; line-height: 26px; }
  .text { white-space: pre-wrap; word-break: break-word;
    min-height: 26px; outline: none;
    font-family: 'SF Mono', Menlo, Consolas, monospace; }

  .tabs { display: flex; overflow-x: auto; gap: 4px;
    padding: 6px 12px; background: #f6f8fd;
    border-top: 1px solid #e4e9f2;
    border-bottom: 1px solid #e4e9f2; scrollbar-width: none; }
  .tabs::-webkit-scrollbar { display: none; }
  .tab { flex: none; padding: 6px 12px; border-radius: 16px;
    font-size: 11px; font-weight: 600; color: #465a80;
    background: transparent; border: none; cursor: pointer;
    white-space: nowrap; }
  .tab.active { background: #3b5bf6; color: #fff; }

  .grid { padding: 8px 10px; background: #fff;
    display: flex; flex-direction: column; gap: 4px;
    transition: max-height .25s ease; overflow: hidden;
    max-height: 900px; }
  .grid.collapsed { max-height: 0; padding: 0; }
  .row { display: grid; gap: 4px; }
  .key { padding: 9px 3px; border-radius: 8px;
    background: #fff; border: 1px solid #dfe5ef;
    font-size: 13px; color: #0b1f44; cursor: pointer;
    text-align: center; user-select: none; font-family: inherit;
    transition: transform .05s, background .1s;
    min-height: 38px; display: flex; align-items: center;
    justify-content: center; }
  .key:active { transform: scale(.95); background: #eef2fb; }
  .key.accent { background: #eef2fb; border-color: #c5d1e9;
    color: #223a6b; font-weight: 600; }
  .key.active-mode { background: #3b5bf6; color: #fff;
    border-color: #3b5bf6; }

  .basegrid { padding: 6px 10px; background: #f6f8fd;
    border-top: 1px solid #e4e9f2;
    display: flex; flex-direction: column; gap: 4px; }
  .basegrid.collapsed { display: none; }

  .action { padding: 10px 3px; border-radius: 8px;
    background: #eef2fb; border: 1px solid #dfe5ef;
    color: #223a6b; font-size: 11px; font-weight: 600;
    cursor: pointer; font-family: inherit; text-align: center;
    min-height: 38px; display: flex; align-items: center;
    justify-content: center; }
  .action.primary { background: #3b5bf6; color: #fff;
    border-color: #3b5bf6; }
  .action:active { transform: scale(.97); }

  .result { padding: 10px 16px; background: #0b1f44; color: #fff;
    font-family: 'SF Mono', Menlo, monospace; font-size: 10px;
    line-height: 1.5; max-height: 0; overflow: hidden;
    transition: max-height .25s ease; }
  .result.shown { max-height: 240px; overflow-y: auto; }
  .result .label { color: #9ab; margin-bottom: 4px; }
  .result .duo { display: grid; grid-template-columns: 1fr 1fr;
    gap: 12px; margin-top: 6px; }
  .result pre { margin: 0; color: #cfe2ff; white-space: pre-wrap; }
</style>
</head>
<body>

<div class="phone">
  <div class="topbar">
    <div class="logo">M</div>
    <div>
      <div class="brand">MethodMarker</div>
      <div class="tag">Better Methods. Higher Marks.</div>
    </div>
  </div>

  <div class="working">
    <div class="head">
      <span>&#9998; Working</span>
      <span class="steps" id="stepCount">0 steps</span>
    </div>
    <div class="lines">
      <div class="nums" id="nums"></div>
      <div class="text" id="text" contenteditable="true" spellcheck="false"></div>
    </div>
  </div>

  <div class="tabs" id="tabs"></div>

  <div class="grid" id="grid"></div>

  <div class="basegrid" id="basegrid"></div>

  <div class="result" id="result"></div>
</div>

<script>
const LAYOUT = __LAYOUT__;
const state = { activeTab: LAYOUT.tabs_order[0], collapsed: false, abcMode: false };

const textEl = document.getElementById('text');
const numsEl = document.getElementById('nums');
const stepCountEl = document.getElementById('stepCount');

// Display char -> engine token mapping (subset; engine conversion only
// touches symbols the grader expects as ASCII)
const TO_ENGINE = {
  "θ":"theta","α":"alpha","β":"beta","γ":"gamma","δ":"delta","ε":"epsilon",
  "λ":"lambda","μ":"mu","π":"pi","ρ":"rho","σ":"sigma","τ":"tau","φ":"phi",
  "ψ":"psi","ω":"omega",
  "Σ":"sum","Δ":"Delta","Ω":"Omega",
  "√":"sqrt","∛":"cbrt",
  "≤":"<=","≥":">=","≠":"!=","±":"+/-","≈":"approx",
  "×":" x ","÷":" / ",
  "⇒":" => ","⇔":" <=> ","∴":" therefore ","∵":" because ",
  "→":"->","∞":"inf","∈":" in ","ℝ":"R",
  "∠":"angle ","△":"triangle ","∥":"parallel ","⊥":"perp ",
  "≅":"cong ","∼":"similar ",
  "Â":"A^","B̂":"B^","Ĉ":"C^","D̂":"D^","Ê":"E^","F̂":"F^",
  "x̄":"xbar","σ²":"sigma^2",
  "x²":"x^2","xⁿ":"x^","x⁻¹":"x^-1","sin²":"sin^2",
  "cos²":"cos^2","tan²":"tan^2",
  "sin⁻¹":"asin","cos⁻¹":"acos","tan⁻¹":"atan",
  "d²y/dx²":"d2y/dx2","Dₓ[ ]":"Dx[]",
};

function toEngine(text) {
  let out = text;
  // Longest symbols first to avoid conflicts
  const syms = Object.keys(TO_ENGINE).sort((a,b) => b.length - a.length);
  for (const s of syms) {
    out = out.split(s).join(TO_ENGINE[s]);
  }
  return out;
}

function refreshNumbers() {
  const lines = textEl.innerText.split("\n");
  numsEl.innerHTML = lines.map((_, i) => `<div>${i + 1}</div>`).join("");
  const n = lines.length;
  stepCountEl.textContent = (n === 1 && lines[0] === "")
    ? "0 steps" : `${n} step${n === 1 ? "" : "s"}`;
}
textEl.addEventListener("input", refreshNumbers);
refreshNumbers();

function getCaretContext() {
  const sel = window.getSelection();
  if (!sel.rangeCount) return { before: "", after: "" };
  const r = sel.getRangeAt(0);
  const pre = r.cloneRange(); pre.selectNodeContents(textEl);
  pre.setEnd(r.startContainer, r.startOffset);
  const before = pre.toString();
  const post = r.cloneRange(); post.selectNodeContents(textEl);
  post.setStart(r.endContainer, r.endOffset);
  const after = post.toString();
  return { before, after };
}

function insertAtCaret(str) {
  textEl.focus();
  const sel = window.getSelection();
  if (!sel.rangeCount) return;
  const r = sel.getRangeAt(0);
  r.deleteContents();
  const node = document.createTextNode(str);
  r.insertNode(node);
  r.setStartAfter(node);
  r.setEndAfter(node);
  sel.removeAllRanges();
  sel.addRange(r);
  refreshNumbers();
}

function smartInsert(key) {
  if (key.e === "__back__") {
    textEl.focus();
    document.execCommand("delete");
    refreshNumbers();
    return;
  }
  if (key.e === "__newline__") { insertAtCaret("\n"); return; }
  if (key.e === "__done__")    { submit(); return; }
  if (key.e === "__abc__") {
    state.abcMode = !state.abcMode;
    renderGrid();
    renderBase();
    return;
  }
  if (key.e === "__collapse__") {
    state.collapsed = !state.collapsed;
    document.getElementById("grid").classList.toggle("collapsed", state.collapsed);
    document.getElementById("basegrid").classList.toggle("collapsed", state.collapsed);
    return;
  }
  const { before, after } = getCaretContext();
  let ins = key.d;
  if (key.sp) {
    const needLead  = before.length > 0 && !/\s$/.test(before);
    const needTrail = after.length  > 0 && !/^\s/.test(after);
    ins = (needLead ? " " : "") + ins.trim() + (needTrail ? " " : "");
  }
  insertAtCaret(ins);
  if (key.skip === 1 && key.d === "()") {
    const sel = window.getSelection();
    if (sel.rangeCount) {
      const r = sel.getRangeAt(0);
      if (r.startContainer.nodeType === 3 && r.startOffset > 0) {
        const nr = r.cloneRange();
        nr.setStart(nr.startContainer, nr.startOffset - 1);
        nr.collapse(true);
        sel.removeAllRanges(); sel.addRange(nr);
      }
    }
  }
}

// ---------- tabs ----------
const tabsEl = document.getElementById("tabs");
LAYOUT.tabs_order.forEach(name => {
  const b = document.createElement("button");
  b.className = "tab" + (name === state.activeTab ? " active" : "");
  b.textContent = name;
  b.onclick = () => {
    state.activeTab = name;
    document.querySelectorAll(".tab").forEach(t =>
      t.classList.toggle("active", t.textContent === name));
    renderGrid();
  };
  tabsEl.appendChild(b);
});

// ---------- tab grid ----------
const gridEl = document.getElementById("grid");
function renderGrid() {
  if (state.abcMode) {
    // ABC mode — render QWERTY letters in the grid
    gridEl.innerHTML = "";
    const rows = LAYOUT.base_layer.abc || [];
    const cols = Math.max(...rows.map(r => r.length));
    rows.forEach(row => {
      const r = document.createElement("div");
      r.className = "row";
      r.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
      row.forEach(key => {
        const k = document.createElement("div");
        k.className = "key";
        k.textContent = key.l;
        k.onclick = () => smartInsert(key);
        r.appendChild(k);
      });
      for (let i = row.length; i < cols; i++) r.appendChild(document.createElement("div"));
      gridEl.appendChild(r);
    });
    return;
  }
  const rows = LAYOUT.tabs[state.activeTab] || [];
  gridEl.innerHTML = "";
  const cols = Math.max(...rows.map(r => r.length));
  rows.forEach(row => {
    const r = document.createElement("div");
    r.className = "row";
    r.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
    row.forEach(key => {
      const k = document.createElement("div");
      const accent = ["or","=","⇒","∴","∵"].includes(key.l) ? " accent" : "";
      k.className = "key" + accent;
      k.textContent = key.l;
      k.onclick = () => smartInsert(key);
      r.appendChild(k);
    });
    for (let i = row.length; i < cols; i++) r.appendChild(document.createElement("div"));
    gridEl.appendChild(r);
  });
}

// ---------- base layer ----------
function renderBase() {
  const b = LAYOUT.base_layer;
  const container = document.getElementById("basegrid");
  container.innerHTML = "";

  const layout = [
    { keys: b.variables, cols: 12 },
    { keys: b.digits,    cols: 12 },
    { keys: b.operators, cols: 10 },
    { keys: b.actions,   cols: 6  },
  ];

  layout.forEach(({keys, cols}) => {
    const row = document.createElement("div");
    row.className = "row";
    row.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
    keys.forEach(key => {
      const isDone = key.e === "__done__";
      const isAction = key.e && key.e.startsWith("__");
      const el = document.createElement("div");
      el.className = isDone ? "action primary"
                           : (isAction ? "action" : "key");
      if (key.e === "__abc__" && state.abcMode) el.classList.add("active-mode");
      el.textContent = key.l;
      el.onclick = () => smartInsert(key);
      row.appendChild(el);
    });
    container.appendChild(row);
  });
}

// ---------- submit ----------
const resultEl = document.getElementById("result");
const CMD = {
  or: " or ", then: " ∴ ", because: " ∵ ",
  implies: " ⇒ ", iff: " ⇔ ",
  sqrt: "sqrt", cbrt: "cbrt", pi: "pi",
  theta: "theta", alpha: "alpha", beta: "beta",
  deg: "deg", pm: "+/-",
  le: "<=", ge: ">=", ne: "!=",
  in: " in ", R: "R", sum: "sum", inf: "inf",
  dy: "dy/dx", dx: "d/dx", lim: "lim", to: "->",
  angle: "angle ", tri: "triangle ",
  para: "parallel ", perp: "perp ",
};

function submit() {
  const rawLines = textEl.innerText.split("\n").map(l => l.trim()).filter(l => l.length);
  // 1. slash-command expansion
  const slashExpanded = rawLines.map(l =>
    l.replace(/\/([A-Za-z][A-Za-z0-9_]*)/g, (m, name) => CMD[name] || m));
  // 2. symbol → engine token expansion
  const engineLines = slashExpanded.map(toEngine);

  const payload = {
    learner_lines: engineLines,
    display_lines: slashExpanded,
    tab_used: state.activeTab,
    steps: engineLines.length,
  };
  resultEl.classList.add("shown");
  resultEl.innerHTML =
    "<div class='label'>submitted to step_grader</div>" +
    "<pre>" + JSON.stringify(payload, null, 2) + "</pre>";
}

textEl.focus();
renderGrid();
renderBase();
</script>
</body>
</html>
"""

HTML = HTML_TEMPLATE.replace("__LAYOUT__", json.dumps(KEYBOARD, ensure_ascii=False))
HTML_PATH.write_text(HTML, encoding="utf-8")
print(f"  ✓ {HTML_PATH.name}  ({HTML_PATH.stat().st_size:,} bytes)")
print()

# Summary
print("=" * 72)
print("v3 CHANGES")
print("=" * 72)
print("  ✓ Unicode display — θ shows as θ, Σ as Σ, √ as √")
print("  ✓ Engine conversion on submit — grader receives 'theta', 'sum', 'sqrt'")
print("  ✓ ABC mode — tap ABC to toggle QWERTY letters")
print("  ✓ Geometry: theorems REMOVED (learner must know them)")
print("  ✓ Popular variables from corpus:")
print(f"     {popular_vars[:12]}")
print()

webbrowser.open(HTML_PATH.as_uri())
print(f"  → opened in browser: {HTML_PATH.as_uri()}")
````

````python
# Rebuild keyboard v4 — tabs match CAPS topics exactly

import json
import webbrowser
from pathlib import Path

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
LAYOUT_PATH = ROOT / "data" / "processed" / "tutor" / "keyboard_layout.json"
HTML_PATH = ROOT / "docs" / "keyboard_preview.html"


def K(label, display=None, engine=None, sp=False, trail=False, skip=None):
    d = {"l": label}
    d["d"] = display if display is not None else label
    d["e"] = engine if engine is not None else d["d"]
    if sp:    d["sp"] = True
    if trail: d["trail"] = True
    if skip:  d["skip"] = skip
    return d


# ============================================================
# BASE LAYER — same on every tab
# ============================================================

BASE_VARIABLES = [K(c) for c in "xyzabcnkmpqt"]
BASE_DIGITS = [K("1"), K("2"), K("3"), K("4"), K("5"), K("6"),
               K("7"), K("8"), K("9"), K("0"), K("."), K("(−)", "(-)")]
BASE_OPS = [
    K("+", " + ", " + ", sp=True),
    K("−", " - ", " - ", sp=True),
    K("×", " × ", " x ", sp=True),
    K("÷", " ÷ ", " / ", sp=True),
    K("=", " = ", " = ", sp=True),
    K("(", "("), K(")", ")"),
    K("^", "^"),
    K("√", "√", "sqrt"),
    K(";", "; ", "; ", trail=True),
]
BASE_ACTIONS = [
    K("⌫", "__back__", "__back__"),
    K("Space", " ", " "),
    K("↵", "__newline__", "__newline__"),
    K("✓ Done", "__done__", "__done__"),
    K("ABC", "__abc__", "__abc__"),
    K("▼", "__collapse__", "__collapse__"),
]
BASE_ABC = [
    [K("q","q","q"), K("w","w","w"), K("e","e","e"), K("r","r","r"),
     K("t","t","t"), K("y","y","y"), K("u","u","u"), K("i","i","i"),
     K("o","o","o"), K("p","p","p")],
    [K("a","a","a"), K("s","s","s"), K("d","d","d"), K("f","f","f"),
     K("g","g","g"), K("h","h","h"), K("j","j","j"), K("k","k","k"),
     K("l","l","l")],
    [K("z","z","z"), K("x","x","x"), K("c","c","c"), K("v","v","v"),
     K("b","b","b"), K("n","n","n"), K("m","m","m"), K("'","'","'"),
     K("-","-","-")],
]


# ============================================================
# TEN CAPS TABS + Communication
# ============================================================

ALGEBRA = [
    # Expressions, equations, inequalities
    [K("or", " or ", " or ", sp=True),
     K("⇒", " ⇒ ", " => ", sp=True),
     K("⇔", " ⇔ ", " <=> ", sp=True),
     K("∴", " ∴ ", " therefore ", sp=True),
     K("∵", " ∵ ", " because ", sp=True),
     K("≠", "≠", "!=")],
    [K("x²", "x²", "x^2"), K("x³", "x³", "x^3"),
     K("xⁿ", "xⁿ", "x^"), K("x⁻¹", "x⁻¹", "x^-1"),
     K("∛", "∛", "cbrt"), K("ⁿ√", "ⁿ√", "nroot")],
    [K("<", "<", "<"), K(">", ">", ">"),
     K("≤", "≤", "<="), K("≥", "≥", ">="),
     K("|x|", "|x|", "abs(x)"),
     K("a/b", "a/b", "a/b")],
    [K("∈", "∈", " in "), K("ℝ", "ℝ", "R"),
     K("π", "π", "pi"), K("∞", "∞", "inf"),
     K("%", "%", "%"), K(",", ",", ",")],
]

SEQUENCES = [
    # Sequences and series — AP, GP, quadratic, sigma
    [K("Tₙ", "Tₙ", "Tn"), K("Sₙ", "Sₙ", "Sn"),
     K("Tₙ₊₁", "Tₙ₊₁", "T_(n+1)"),
     K("Tₙ₋₁", "Tₙ₋₁", "T_(n-1)"),
     K("S∞", "S∞", "S_inf"),
     K("a", "a", "a")],
    [K("d", "d", "d"), K("r", "r", "r"), K("n", "n", "n"),
     K("Σ", "Σ", "sum"), K("∑", "∑", "sum"),
     K("...", "...", "...")],
    [K("or", " or ", " or ", sp=True),
     K("∴", " ∴ ", " therefore ", sp=True),
     K("⇒", " ⇒ ", " => ", sp=True),
     K("∈", "∈", " in "),
     K("ℕ", "ℕ", "N"),
     K("ℤ", "ℤ", "Z")],
    [K("k", "k", "k"), K("p", "p", "p"),
     K("m", "m", "m"), K("t", "t", "t"),
     K("n→∞", "n→∞", "n->inf"),
     K("≠", "≠", "!=")],
]

FUNCTIONS = [
    # Functions and graphs — parabola, hyperbola, exp, log, inverse
    [K("f(x)", "f(x)", "f(x)"),
     K("g(x)", "g(x)", "g(x)"),
     K("h(x)", "h(x)", "h(x)"),
     K("f⁻¹", "f⁻¹", "f^-1"),
     K("g⁻¹", "g⁻¹", "g^-1"),
     K("D", "D", "D")],
    [K("x", "x", "x"), K("y", "y", "y"),
     K("→", "→", "->"),
     K("∞", "∞", "inf"),
     K("a", "a", "a"),
     K("q", "q", "q")],
    [K(">", ">", ">"), K("<", "<", "<"),
     K("≤", "≤", "<="), K("≥", "≥", ">="),
     K("≠", "≠", "!="),
     K("∈", "∈", " in ")],
    [K("∴", " ∴ ", " therefore ", sp=True),
     K("or", " or ", " or ", sp=True),
     K("⇒", " ⇒ ", " => ", sp=True),
     K("ℝ", "ℝ", "R"),
     K("(", "("), K(")", ")")],
]

CALCULUS = [
    [K("dy/dx", "dy/dx", "dy/dx"),
     K("d/dx", "d/dx", "d/dx"),
     K("d²y/dx²", "d²y/dx²", "d2y/dx2"),
     K("Dₓ[ ]", "Dₓ[ ]", "Dx[]"),
     K("f'(x)", "f'(x)", "f'(x)"),
     K("f''(x)", "f''(x)", "f''(x)")],
    [K("lim", "lim", "lim("),
     K("→", "→", "->"),
     K("h→0", "h→0", "h->0"),
     K("x→∞", "x→∞", "x->inf"),
     K("Δ", "Δ", "Delta"),
     K("∞", "∞", "inf")],
    [K("e", "e", "e"), K("ln", "ln", "ln("),
     K("log", "log", "log("),
     K("∴", " ∴ ", " therefore ", sp=True),
     K("⇒", " ⇒ ", " => ", sp=True),
     K("( )", "()", "()", skip=1)],
    [K("max", "max", "max"), K("min", "min", "min"),
     K("f'", "f'", "f'"), K("f''", "f''", "f''"),
     K("=", " = ", " = ", sp=True),
     K("≠", "≠", "!=")],
]

TRIG = [
    [K("sin", "sin", "sin"), K("cos", "cos", "cos"), K("tan", "tan", "tan"),
     K("sin⁻¹", "sin⁻¹", "asin"),
     K("cos⁻¹", "cos⁻¹", "acos"),
     K("tan⁻¹", "tan⁻¹", "atan")],
    [K("θ", "θ", "theta"), K("α", "α", "alpha"),
     K("β", "β", "beta"), K("γ", "γ", "gamma"),
     K("π", "π", "pi"), K("°", "°", "°")],
    [K("2θ", "2θ", "2theta"), K("½θ", "θ/2", "theta/2"),
     K("sin²", "sin²", "sin^2"), K("cos²", "cos²", "cos^2"),
     K("tan²", "tan²", "tan^2"), K("rad", "rad", "rad")],
    [K("or", " or ", " or ", sp=True),
     K("⇒", " ⇒ ", " => ", sp=True),
     K("∴", " ∴ ", " therefore ", sp=True),
     K("±", "±", "+/-"),
     K("k∈ℤ", "k∈ℤ", "k in Z"),
     K("( )", "()", "()", skip=1)],
]

FINANCE = [
    # Finance — interest, annuities, loans, depreciation
    [K("P", "P", "P"), K("A", "A", "A"),
     K("F", "F", "F"), K("Fv", "Fv", "Fv"),
     K("Pv", "Pv", "Pv"), K("i", "i", "i")],
    [K("n", "n", "n"), K("x", "x", "x"),
     K("m", "m", "m"), K("r", "r", "r"),
     K("t", "t", "t"), K("R", "R", "R")],
    [K("%", "%", "%"), K("÷", " ÷ ", " / ", sp=True),
     K("×", " × ", " x ", sp=True),
     K("^", "^"),
     K("( )", "()", "()", skip=1),
     K(".", ".", ".")],
    [K("=", " = ", " = ", sp=True),
     K("∴", " ∴ ", " therefore ", sp=True),
     K("≈", "≈", "approx"),
     K("<", "<"), K(">", ">"),
     K(",", ",", ",")],
]

ANALYTICAL = [
    # Analytical geometry — coordinates, gradients, circles
    [K("A", "A", "A"), K("B", "B", "B"),
     K("C", "C", "C"), K("D", "D", "D"),
     K("M", "M", "M"), K("N", "N", "N")],
    [K("(x;y)", "(x;y)", "(x;y)"),
     K("(x₁;y₁)", "(x₁;y₁)", "(x1;y1)"),
     K("(x₂;y₂)", "(x₂;y₂)", "(x2;y2)"),
     K(";", "; ", "; ", trail=True),
     K(",", ",", ","),
     K("→", "→", "->")],
    [K("m", "m", "m"),
     K("tanθ", "tanθ", "tan(theta)"),
     K("θ", "θ", "theta"),
     K("°", "°", "°"),
     K("√", "√", "sqrt"),
     K("|", "|", "|")],
    [K("=", " = ", " = ", sp=True),
     K("≠", "≠", "!="),
     K("∥", "∥", "parallel "),
     K("⊥", "⊥", "perp "),
     K("( )", "()", "()", skip=1),
     K("∴", " ∴ ", " therefore ", sp=True)],
    [K("x²", "x²", "x^2"),
     K("y²", "y²", "y^2"),
     K("r", "r", "r"),
     K("a", "a", "a"),
     K("b", "b", "b"),
     K("c", "c", "c")],
]

EUCLIDEAN = [
    # Euclidean geometry — primitives only. Learner must know theorems.
    [K("∠", "∠", "angle "),
     K("△", "△", "triangle "),
     K("∥", "∥", "parallel "),
     K("⊥", "⊥", "perp "),
     K("≅", "≅", "cong "),
     K("∼", "∼", "similar ")],
    [K("Â", "Â", "A^"), K("B̂", "B̂", "B^"),
     K("Ĉ", "Ĉ", "C^"), K("D̂", "D̂", "D^"),
     K("Ê", "Ê", "E^"), K("F̂", "F̂", "F^")],
    [K("∴", " ∴ ", " therefore ", sp=True),
     K("∵", " ∵ ", " because ", sp=True),
     K("⇒", " ⇒ ", " => ", sp=True),
     K("given", "given", "given"),
     K("common", "common", "common"),
     K("vert opp", "vert opp ∠s", "vert opp angles")],
    [K("alt ∠s", "alt ∠s", "alternate angles"),
     K("co-int ∠s", "co-int ∠s", "co-interior angles"),
     K("corresp ∠s", "corresp ∠s", "corresponding angles"),
     K("arc", "arc", "arc "),
     K("chord", "chord", "chord "),
     K("rad", "rad", "rad ")],
]

STATS = [
    [K("x̄", "x̄", "xbar"), K("σ", "σ", "sigma"),
     K("σ²", "σ²", "sigma^2"),
     K("n", "n", "n"), K("N", "N", "N"),
     K("Σ", "Σ", "sum")],
    [K("Σx", "Σx", "sum(x)"),
     K("Σx²", "Σx²", "sum(x^2)"),
     K("xᵢ", "xᵢ", "x_i"), K("fᵢ", "fᵢ", "f_i"),
     K("Q1", "Q1", "Q1"), K("Q3", "Q3", "Q3")],
    [K("Q2", "Q2", "Q2"), K("min", "min", "min"),
     K("max", "max", "max"),
     K("r", "r", "r"),
     K("a", "a", "a"), K("b", "b", "b")],
    [K("ŷ", "ŷ", "yhat"),
     K("x̄", "x̄", "xbar"),
     K("ȳ", "ȳ", "ybar"),
     K("∴", " ∴ ", " therefore ", sp=True),
     K("≈", "≈", "approx"),
     K("%", "%", "%")],
    [K("ogive", "ogive", "ogive"),
     K("IQR", "IQR", "IQR"),
     K("SD", "SD", "SD"),
     K("y", "y", "y"),
     K("x", "x", "x"),
     K("(", "(")],
]

PROBABILITY = [
    [K("P(", "P(", "P("),
     K("P(A)", "P(A)", "P(A)"),
     K("P(B)", "P(B)", "P(B)"),
     K("P(A∪B)", "P(A∪B)", "P(A or B)"),
     K("P(A∩B)", "P(A∩B)", "P(A and B)"),
     K("P(A|B)", "P(A|B)", "P(A|B)")],
    [K("∪", "∪", " union "),
     K("∩", "∩", " intersect "),
     K("'", "'", "'"),
     K("A'", "A'", "A'"),
     K("|", "|", "|"),
     K("P(not", "P(not ", "P(not ")],
    [K("n(A)", "n(A)", "n(A)"),
     K("n(B)", "n(B)", "n(B)"),
     K("n(S)", "n(S)", "n(S)"),
     K("!", "!", "!"),
     K("nCr", "nCr", "nCr"),
     K("nPr", "nPr", "nPr")],
    [K("=", " = ", " = ", sp=True),
     K("×", " × ", " x ", sp=True),
     K("÷", " ÷ ", " / ", sp=True),
     K("∴", " ∴ ", " therefore ", sp=True),
     K("( )", "()", "()", skip=1),
     K("≠", "≠", "!=")],
]

GREEK = [
    [K("α","α","alpha"), K("β","β","beta"), K("γ","γ","gamma"),
     K("δ","δ","delta"), K("ε","ε","epsilon"), K("θ","θ","theta")],
    [K("λ","λ","lambda"), K("μ","μ","mu"), K("π","π","pi"),
     K("ρ","ρ","rho"), K("σ","σ","sigma"), K("τ","τ","tau")],
    [K("φ","φ","phi"), K("ψ","ψ","psi"), K("ω","ω","omega"),
     K("Δ","Δ","Delta"), K("Σ","Σ","Sigma"), K("Ω","Ω","Omega")],
    [K("⇒"," ⇒ "," => ", sp=True),
     K("∴"," ∴ "," therefore ", sp=True),
     K("∈","∈"," in "),
     K("∞","∞","inf"),
     K("≈","≈","approx"),
     K("≠","≠","!=")],
]


# ============================================================
# TEN CURRICULUM TABS
# ============================================================

KEYBOARD = {
    "schema_version": 4,
    "tabs_order": [
        "Algebra",
        "Sequences",
        "Functions",
        "Calculus",
        "Trig",
        "Finance",
        "Analytical",
        "Euclidean",
        "Stats",
        "Probability",
        "Greek",
    ],
    "base_layer": {
        "variables": BASE_VARIABLES,
        "digits":    BASE_DIGITS,
        "operators": BASE_OPS,
        "actions":   BASE_ACTIONS,
        "abc":       BASE_ABC,
    },
    "tabs": {
        "Algebra":     ALGEBRA,
        "Sequences":   SEQUENCES,
        "Functions":   FUNCTIONS,
        "Calculus":    CALCULUS,
        "Trig":        TRIG,
        "Finance":     FINANCE,
        "Analytical":  ANALYTICAL,
        "Euclidean":   EUCLIDEAN,
        "Stats":       STATS,
        "Probability": PROBABILITY,
        "Greek":       GREEK,
    },
    "settings": {
        "line_cap": None,
        "tab_auto_switch": False,
        "keyboard_collapsible": True,
        "keyboard_default": "expanded",
    },
    "tab_auto_switch_map": {
        "algebra":     "Algebra",
        "sequences":   "Sequences",
        "functions":   "Functions",
        "calculus":    "Calculus",
        "trig":        "Trig",
        "finance":     "Finance",
        "analytical":  "Analytical",
        "geometry":    "Euclidean",
        "stats":       "Stats",
        "probability": "Probability",
    },
}

LAYOUT_PATH.write_text(
    json.dumps(KEYBOARD, indent=2, ensure_ascii=False), encoding="utf-8")

print(f"  ✓ keyboard_layout.json v4")
print(f"    {len(KEYBOARD['tabs'])} tabs, matching CAPS curriculum:")
for tab in KEYBOARD["tabs_order"]:
    n = len(KEYBOARD["tabs"][tab])
    keys = sum(len(r) for r in KEYBOARD["tabs"][tab])
    print(f"      · {tab:14s}  {n} rows, {keys} keys")
print()


# ============================================================
# Rebuild HTML
# ============================================================

HTML_TEMPLATE = r"""<!doctype html>
<html lang="en">
<head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width,initial-scale=1">
<title>MethodKeyboard — preview v4</title>
<style>
  * { box-sizing: border-box; -webkit-tap-highlight-color: transparent; }
  body { margin: 0; background: #eef2f8;
    font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif;
    display: flex; justify-content: center; padding: 12px; min-height: 100vh; }
  .phone { width: 440px; max-width: 100%; background: #fff;
    border-radius: 20px; overflow: hidden;
    box-shadow: 0 8px 32px rgba(20, 40, 90, .12);
    display: flex; flex-direction: column; }
  .topbar { display: flex; align-items: center; gap: 10px;
    padding: 12px 16px; background: #f6f8fd;
    border-bottom: 1px solid #e4e9f2; }
  .topbar .logo { width: 28px; height: 28px; border-radius: 7px;
    background: linear-gradient(135deg, #3b5bf6, #5b8cf6);
    display: flex; align-items: center; justify-content: center;
    color: #fff; font-weight: 700; font-size: 15px; }
  .topbar .brand { font-weight: 700; font-size: 14px; color: #0b1f44; }
  .topbar .tag { font-size: 10px; color: #6b7a99; }

  .working { padding: 12px 16px; background: #fff; }
  .working .head { display: flex; justify-content: space-between;
    align-items: center; font-size: 12px; color: #465a80;
    font-weight: 600; margin-bottom: 6px; }
  .working .steps { color: #6b7a99; font-weight: 500; }
  .lines { display: grid; grid-template-columns: 22px 1fr;
    column-gap: 8px; font-size: 15px; line-height: 26px;
    color: #0b1f44; max-height: 160px; overflow-y: auto; }
  .nums { color: #9aa8c2; font-size: 11px; text-align: right;
    user-select: none; padding-top: 2px; }
  .nums div { height: 26px; line-height: 26px; }
  .text { white-space: pre-wrap; word-break: break-word;
    min-height: 26px; outline: none;
    font-family: 'SF Mono', Menlo, Consolas, monospace; }

  .tabs { display: flex; overflow-x: auto; gap: 4px;
    padding: 6px 12px; background: #f6f8fd;
    border-top: 1px solid #e4e9f2;
    border-bottom: 1px solid #e4e9f2; scrollbar-width: none; }
  .tabs::-webkit-scrollbar { display: none; }
  .tab { flex: none; padding: 6px 11px; border-radius: 16px;
    font-size: 11px; font-weight: 600; color: #465a80;
    background: transparent; border: none; cursor: pointer;
    white-space: nowrap; }
  .tab.active { background: #3b5bf6; color: #fff; }

  .grid { padding: 8px 10px; background: #fff;
    display: flex; flex-direction: column; gap: 4px;
    transition: max-height .25s ease; overflow: hidden;
    max-height: 900px; }
  .grid.collapsed { max-height: 0; padding: 0; }
  .row { display: grid; gap: 4px; }
  .key { padding: 9px 3px; border-radius: 8px;
    background: #fff; border: 1px solid #dfe5ef;
    font-size: 13px; color: #0b1f44; cursor: pointer;
    text-align: center; user-select: none; font-family: inherit;
    transition: transform .05s, background .1s;
    min-height: 38px; display: flex; align-items: center;
    justify-content: center; }
  .key:active { transform: scale(.95); background: #eef2fb; }
  .key.accent { background: #eef2fb; border-color: #c5d1e9;
    color: #223a6b; font-weight: 600; }
  .key.active-mode { background: #3b5bf6; color: #fff;
    border-color: #3b5bf6; }
  .key.small { font-size: 11px; }

  .basegrid { padding: 6px 10px; background: #f6f8fd;
    border-top: 1px solid #e4e9f2;
    display: flex; flex-direction: column; gap: 4px; }
  .basegrid.collapsed { display: none; }

  .action { padding: 10px 3px; border-radius: 8px;
    background: #eef2fb; border: 1px solid #dfe5ef;
    color: #223a6b; font-size: 11px; font-weight: 600;
    cursor: pointer; font-family: inherit; text-align: center;
    min-height: 38px; display: flex; align-items: center;
    justify-content: center; }
  .action.primary { background: #3b5bf6; color: #fff;
    border-color: #3b5bf6; }
  .action:active { transform: scale(.97); }

  .result { padding: 10px 16px; background: #0b1f44; color: #fff;
    font-family: 'SF Mono', Menlo, monospace; font-size: 10px;
    line-height: 1.5; max-height: 0; overflow: hidden;
    transition: max-height .25s ease; }
  .result.shown { max-height: 260px; overflow-y: auto; }
  .result .label { color: #9ab; margin-bottom: 4px; }
  .result pre { margin: 0; color: #cfe2ff; white-space: pre-wrap; }
</style>
</head>
<body>

<div class="phone">
  <div class="topbar">
    <div class="logo">M</div>
    <div>
      <div class="brand">MethodMarker</div>
      <div class="tag">Better Methods. Higher Marks.</div>
    </div>
  </div>

  <div class="working">
    <div class="head">
      <span>&#9998; Working</span>
      <span class="steps" id="stepCount">0 steps</span>
    </div>
    <div class="lines">
      <div class="nums" id="nums"></div>
      <div class="text" id="text" contenteditable="true" spellcheck="false"></div>
    </div>
  </div>

  <div class="tabs" id="tabs"></div>

  <div class="grid" id="grid"></div>

  <div class="basegrid" id="basegrid"></div>

  <div class="result" id="result"></div>
</div>

<script>
const LAYOUT = __LAYOUT__;
const state = { activeTab: LAYOUT.tabs_order[0], collapsed: false, abcMode: false };

const textEl = document.getElementById("text");
const numsEl = document.getElementById("nums");
const stepCountEl = document.getElementById("stepCount");

const TO_ENGINE = {
  "θ":"theta","α":"alpha","β":"beta","γ":"gamma","δ":"delta","ε":"epsilon",
  "λ":"lambda","μ":"mu","π":"pi","ρ":"rho","σ":"sigma","τ":"tau","φ":"phi",
  "ψ":"psi","ω":"omega",
  "Σ":"sum","∑":"sum","Δ":"Delta","Ω":"Omega",
  "√":"sqrt","∛":"cbrt","ⁿ√":"nroot",
  "≤":"<=","≥":">=","≠":"!=","±":"+/-","≈":"approx",
  "×":" x ","÷":" / ",
  "⇒":" => ","⇔":" <=> ","∴":" therefore ","∵":" because ",
  "→":"->","∞":"inf","∈":" in ","ℝ":"R","ℕ":"N","ℤ":"Z",
  "∠":"angle ","△":"triangle ","∥":"parallel ","⊥":"perp ",
  "≅":"cong ","∼":"similar ",
  "Â":"A^","B̂":"B^","Ĉ":"C^","D̂":"D^","Ê":"E^","F̂":"F^",
  "x̄":"xbar","ȳ":"ybar","ŷ":"yhat","σ²":"sigma^2",
  "x²":"x^2","x³":"x^3","xⁿ":"x^","x⁻¹":"x^-1",
  "sin²":"sin^2","cos²":"cos^2","tan²":"tan^2",
  "sin⁻¹":"asin","cos⁻¹":"acos","tan⁻¹":"atan",
  "d²y/dx²":"d2y/dx2","Dₓ[ ]":"Dx[]",
  "Tₙ":"Tn","Sₙ":"Sn","Tₙ₊₁":"T_(n+1)","Tₙ₋₁":"T_(n-1)",
  "S∞":"S_inf","n→∞":"n->inf",
  "f⁻¹":"f^-1","g⁻¹":"g^-1",
  "∪":" union ","∩":" intersect ",
  "(x;y)":"(x;y)","(x₁;y₁)":"(x1;y1)","(x₂;y₂)":"(x2;y2)",
  "tanθ":"tan(theta)",
};

function toEngine(text) {
  let out = text;
  const syms = Object.keys(TO_ENGINE).sort((a,b) => b.length - a.length);
  for (const s of syms) out = out.split(s).join(TO_ENGINE[s]);
  return out;
}

function refreshNumbers() {
  const lines = textEl.innerText.split("\n");
  numsEl.innerHTML = lines.map((_, i) => `<div>${i + 1}</div>`).join("");
  const n = lines.length;
  stepCountEl.textContent = (n === 1 && lines[0] === "")
    ? "0 steps" : `${n} step${n === 1 ? "" : "s"}`;
}
textEl.addEventListener("input", refreshNumbers);
refreshNumbers();

function getCaretContext() {
  const sel = window.getSelection();
  if (!sel.rangeCount) return { before: "", after: "" };
  const r = sel.getRangeAt(0);
  const pre = r.cloneRange(); pre.selectNodeContents(textEl);
  pre.setEnd(r.startContainer, r.startOffset);
  const before = pre.toString();
  const post = r.cloneRange(); post.selectNodeContents(textEl);
  post.setStart(r.endContainer, r.endOffset);
  const after = post.toString();
  return { before, after };
}

function insertAtCaret(str) {
  textEl.focus();
  const sel = window.getSelection();
  if (!sel.rangeCount) return;
  const r = sel.getRangeAt(0);
  r.deleteContents();
  const node = document.createTextNode(str);
  r.insertNode(node);
  r.setStartAfter(node);
  r.setEndAfter(node);
  sel.removeAllRanges();
  sel.addRange(r);
  refreshNumbers();
}

function smartInsert(key) {
  if (key.e === "__back__") {
    textEl.focus();
    document.execCommand("delete");
    refreshNumbers();
    return;
  }
  if (key.e === "__newline__") { insertAtCaret("\n"); return; }
  if (key.e === "__done__")    { submit(); return; }
  if (key.e === "__abc__") {
    state.abcMode = !state.abcMode;
    renderGrid();
    renderBase();
    return;
  }
  if (key.e === "__collapse__") {
    state.collapsed = !state.collapsed;
    document.getElementById("grid").classList.toggle("collapsed", state.collapsed);
    document.getElementById("basegrid").classList.toggle("collapsed", state.collapsed);
    return;
  }
  const { before, after } = getCaretContext();
  let ins = key.d;
  if (key.sp) {
    const needLead  = before.length > 0 && !/\s$/.test(before);
    const needTrail = after.length  > 0 && !/^\s/.test(after);
    ins = (needLead ? " " : "") + ins.trim() + (needTrail ? " " : "");
  }
  insertAtCaret(ins);
  if (key.skip === 1 && key.d === "()") {
    const sel = window.getSelection();
    if (sel.rangeCount) {
      const r = sel.getRangeAt(0);
      if (r.startContainer.nodeType === 3 && r.startOffset > 0) {
        const nr = r.cloneRange();
        nr.setStart(nr.startContainer, nr.startOffset - 1);
        nr.collapse(true);
        sel.removeAllRanges(); sel.addRange(nr);
      }
    }
  }
}

// ---------- tabs ----------
const tabsEl = document.getElementById("tabs");
LAYOUT.tabs_order.forEach(name => {
  const b = document.createElement("button");
  b.className = "tab" + (name === state.activeTab ? " active" : "");
  b.textContent = name;
  b.onclick = () => {
    state.activeTab = name;
    document.querySelectorAll(".tab").forEach(t =>
      t.classList.toggle("active", t.textContent === name));
    renderGrid();
  };
  tabsEl.appendChild(b);
});

// ---------- grid ----------
const gridEl = document.getElementById("grid");
function renderGrid() {
  if (state.abcMode) {
    gridEl.innerHTML = "";
    const rows = LAYOUT.base_layer.abc || [];
    const cols = Math.max(...rows.map(r => r.length));
    rows.forEach(row => {
      const r = document.createElement("div");
      r.className = "row";
      r.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
      row.forEach(key => {
        const k = document.createElement("div");
        k.className = "key";
        k.textContent = key.l;
        k.onclick = () => smartInsert(key);
        r.appendChild(k);
      });
      for (let i = row.length; i < cols; i++)
        r.appendChild(document.createElement("div"));
      gridEl.appendChild(r);
    });
    return;
  }
  const rows = LAYOUT.tabs[state.activeTab] || [];
  gridEl.innerHTML = "";
  const cols = Math.max(...rows.map(r => r.length));
  rows.forEach(row => {
    const r = document.createElement("div");
    r.className = "row";
    r.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
    row.forEach(key => {
      const k = document.createElement("div");
      const accent = ["or","=","⇒","∴","∵"].includes(key.l) ? " accent" : "";
      const small = key.l.length > 4 ? " small" : "";
      k.className = "key" + accent + small;
      k.textContent = key.l;
      k.onclick = () => smartInsert(key);
      r.appendChild(k);
    });
    for (let i = row.length; i < cols; i++)
      r.appendChild(document.createElement("div"));
    gridEl.appendChild(r);
  });
}

// ---------- base ----------
function renderBase() {
  const b = LAYOUT.base_layer;
  const container = document.getElementById("basegrid");
  container.innerHTML = "";
  const layout = [
    { keys: b.variables, cols: 12 },
    { keys: b.digits,    cols: 12 },
    { keys: b.operators, cols: 10 },
    { keys: b.actions,   cols: 6  },
  ];
  layout.forEach(({keys, cols}) => {
    const row = document.createElement("div");
    row.className = "row";
    row.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
    keys.forEach(key => {
      const isDone = key.e === "__done__";
      const isAction = key.e && key.e.startsWith("__");
      const el = document.createElement("div");
      el.className = isDone ? "action primary"
                           : (isAction ? "action" : "key");
      if (key.e === "__abc__" && state.abcMode) el.classList.add("active-mode");
      el.textContent = key.l;
      el.onclick = () => smartInsert(key);
      row.appendChild(el);
    });
    container.appendChild(row);
  });
}

// ---------- submit ----------
const resultEl = document.getElementById("result");
const CMD = {
  or: " or ", then: " ∴ ", because: " ∵ ",
  implies: " ⇒ ", iff: " ⇔ ",
  sqrt: "sqrt", cbrt: "cbrt", pi: "pi",
  theta: "theta", alpha: "alpha", beta: "beta",
  deg: "deg", pm: "+/-", le: "<=", ge: ">=",
  in: " in ", R: "R", sum: "sum", inf: "inf",
  dy: "dy/dx", dx: "d/dx", lim: "lim", to: "->",
};

function submit() {
  const rawLines = textEl.innerText.split("\n").map(l => l.trim()).filter(l => l.length);
  const slashExpanded = rawLines.map(l =>
    l.replace(/\/([A-Za-z][A-Za-z0-9_]*)/g, (m, name) => CMD[name] || m));
  const engineLines = slashExpanded.map(toEngine);
  const payload = {
    learner_lines: engineLines,
    display_lines: slashExpanded,
    tab_used: state.activeTab,
    steps: engineLines.length,
  };
  resultEl.classList.add("shown");
  resultEl.innerHTML =
    "<div class='label'>submitted to step_grader</div>" +
    "<pre>" + JSON.stringify(payload, null, 2) + "</pre>";
}

textEl.focus();
renderGrid();
renderBase();
</script>
</body>
</html>
"""

HTML = HTML_TEMPLATE.replace("__LAYOUT__", json.dumps(KEYBOARD, ensure_ascii=False))
HTML_PATH.write_text(HTML, encoding="utf-8")
print(f"  ✓ {HTML_PATH.name} rebuilt")
print()
print("=" * 72)
print("OPENED IN BROWSER")
print("=" * 72)
print(f"  {HTML_PATH.as_uri()}")
print()
print("  Tabs now match the CAPS curriculum:")
for tab in KEYBOARD["tabs_order"]:
    n = sum(len(r) for r in KEYBOARD["tabs"][tab])
    print(f"    · {tab:14s}  {n} keys")
print()
print("  Test:")
print("    1. Sequences tab → tap Tₙ, Sₙ, Σ, S∞, →")
print("    2. Functions tab → tap f(x), f⁻¹, →, ∞, ≤, ≥")
print("    3. Probability tab → tap P(A), ∪, ∩, P(A|B), nCr")
print("    4. Finance tab → tap P, A, F, Fv, Pv, i, n")
print("    5. Euclidean tab → tap ∠, △, ∥, ⊥, Â, ∴, given")
print("    6. ABC mode → tap ABC in bottom → QWERTY letters")

webbrowser.open(HTML_PATH.as_uri())
````

````python
# Keyboard v5 — expandable sqrt, all-variable exponents, Casio-style fraction

import json
import webbrowser
from pathlib import Path

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
LAYOUT_PATH = ROOT / "data" / "processed" / "tutor" / "keyboard_layout.json"
HTML_PATH = ROOT / "docs" / "keyboard_preview.html"


def K(label, display=None, engine=None, sp=False, trail=False, skip=None, html=None):
    d = {"l": label}
    d["d"] = display if display is not None else label
    d["e"] = engine if engine is not None else d["d"]
    if sp:    d["sp"] = True
    if trail: d["trail"] = True
    if skip is not None: d["skip"] = skip
    if html:  d["h"] = html
    return d


# ============================================================
# Load existing layout, upgrade the pieces we're changing
# ============================================================

kb = json.loads(LAYOUT_PATH.read_text(encoding="utf-8"))


# --- New base operators: fraction added ---
kb["base_layer"]["operators"] = [
    K("+", " + ", " + ", sp=True),
    K("−", " - ", " - ", sp=True),
    K("×", " × ", " x ", sp=True),
    K("÷", " ÷ ", " / ", sp=True),
    K("(", "("), K(")", ")"),
    K("^", "^"),
    K("√", "√()", "sqrt()", skip=1),
    K("a/b", "()/()", "()/()", skip=4,
      html="<span class='frac'><span>a</span><span>b</span></span>"),
    K(";", "; ", "; ", trail=True),
]

# --- New Algebra tab ---
kb["tabs"]["Algebra"] = [
    # Row 1 — relations
    [K("=", " = ", " = ", sp=True),
     K("or", " or ", " or ", sp=True),
     K("⇒", " ⇒ ", " => ", sp=True),
     K("⇔", " ⇔ ", " <=> ", sp=True),
     K("∴", " ∴ ", " therefore ", sp=True),
     K("∵", " ∵ ", " because ", sp=True)],
    # Row 2 — squares for the classic variables
    [K("x²", "x²", "x^2"),
     K("y²", "y²", "y^2"),
     K("a²", "a²", "a^2"),
     K("b²", "b²", "b^2"),
     K("c²", "c²", "c^2"),
     K("n²", "n²", "n^2")],
    # Row 3 — higher powers + sequence shorthand
    [K("x³", "x³", "x^3"),
     K("xⁿ", "xⁿ", "x^n"),
     K("aⁿ", "aⁿ", "a^n"),
     K("bⁿ", "bⁿ", "b^n"),
     K("Tₙ", "Tₙ", "Tn"),
     K("Sₙ", "Sₙ", "Sn")],
    # Row 4 — three roots + ± + abs
    [K("√", "√()", "sqrt()", skip=1),
     K("∛", "∛()", "cbrt()", skip=1),
     K("ⁿ√", "ⁿ√()", "nroot()", skip=1),
     K("±", "±", "+/-"),
     K("|x|", "|x|", "abs(x)"),
     K("x⁻¹", "x⁻¹", "x^-1")],
    # Row 5 — inequalities + sets
    [K("<", "<"), K(">", ">"),
     K("≤", "≤", "<="), K("≥", "≥", ">="),
     K("≠", "≠", "!="),
     K("∈", "∈", " in ")],
    # Row 6 — extras + quadratic formula template
    [K("ℝ", "ℝ", "R"),
     K("π", "π", "pi"),
     K("∞", "∞", "inf"),
     K("QF", "x = (-b ± √(b² - 4ac))/(2a)",
       "x = (-b ± sqrt(b^2 - 4ac))/(2a)"),
     K("%", "%", "%"),
     K("Σ", "Σ", "sum")],
]

kb["schema_version"] = 5
LAYOUT_PATH.write_text(
    json.dumps(kb, indent=2, ensure_ascii=False), encoding="utf-8")

print(f"  ✓ keyboard_layout.json v5")
print(f"    Algebra tab: {len(kb['tabs']['Algebra'])} rows, "
      f"{sum(len(r) for r in kb['tabs']['Algebra'])} keys")
print()
print("  Algebra row 2:  x²  y²  a²  b²  c²  n²       (squares for every var)")
print("  Algebra row 3:  x³  xⁿ  aⁿ  bⁿ  Tₙ  Sₙ       (higher powers)")
print("  Algebra row 4:  √   ∛   ⁿ√  ±   |x| x⁻¹       (three roots + ±)")
print("  Algebra row 6:  ℝ   π   ∞   QF  %   Σ         (QF = quadratic formula)")
print("  Base operators: + − × ÷ ( ) ^ √ a/b ;        (a/b = stacked fraction)")


# ============================================================
# Rebuild HTML with new features:
#   - HTML labels for the stacked fraction key
#   - skip: N cursor positioning (works for any insert)
#   - Tab key to jump between ()() pairs
# ============================================================

HTML_TEMPLATE = r"""<!doctype html>
<html lang="en">
<head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width,initial-scale=1">
<title>MethodKeyboard — preview v5</title>
<style>
  * { box-sizing: border-box; -webkit-tap-highlight-color: transparent; }
  body { margin: 0; background: #eef2f8;
    font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif;
    display: flex; justify-content: center; padding: 12px; min-height: 100vh; }
  .phone { width: 440px; max-width: 100%; background: #fff;
    border-radius: 20px; overflow: hidden;
    box-shadow: 0 8px 32px rgba(20, 40, 90, .12);
    display: flex; flex-direction: column; }
  .topbar { display: flex; align-items: center; gap: 10px;
    padding: 12px 16px; background: #f6f8fd;
    border-bottom: 1px solid #e4e9f2; }
  .topbar .logo { width: 28px; height: 28px; border-radius: 7px;
    background: linear-gradient(135deg, #3b5bf6, #5b8cf6);
    display: flex; align-items: center; justify-content: center;
    color: #fff; font-weight: 700; font-size: 15px; }
  .topbar .brand { font-weight: 700; font-size: 14px; color: #0b1f44; }
  .topbar .tag { font-size: 10px; color: #6b7a99; }

  .working { padding: 12px 16px; background: #fff; }
  .working .head { display: flex; justify-content: space-between;
    align-items: center; font-size: 12px; color: #465a80;
    font-weight: 600; margin-bottom: 6px; }
  .working .steps { color: #6b7a99; font-weight: 500; }
  .lines { display: grid; grid-template-columns: 22px 1fr;
    column-gap: 8px; font-size: 15px; line-height: 26px;
    color: #0b1f44; max-height: 180px; overflow-y: auto; }
  .nums { color: #9aa8c2; font-size: 11px; text-align: right;
    user-select: none; padding-top: 2px; }
  .nums div { height: 26px; line-height: 26px; }
  .text { white-space: pre-wrap; word-break: break-word;
    min-height: 26px; outline: none;
    font-family: 'SF Mono', Menlo, Consolas, monospace; }

  .tabs { display: flex; overflow-x: auto; gap: 4px;
    padding: 6px 12px; background: #f6f8fd;
    border-top: 1px solid #e4e9f2;
    border-bottom: 1px solid #e4e9f2; scrollbar-width: none; }
  .tabs::-webkit-scrollbar { display: none; }
  .tab { flex: none; padding: 6px 11px; border-radius: 16px;
    font-size: 11px; font-weight: 600; color: #465a80;
    background: transparent; border: none; cursor: pointer;
    white-space: nowrap; }
  .tab.active { background: #3b5bf6; color: #fff; }

  .grid { padding: 8px 10px; background: #fff;
    display: flex; flex-direction: column; gap: 4px;
    transition: max-height .25s ease; overflow: hidden;
    max-height: 900px; }
  .grid.collapsed { max-height: 0; padding: 0; }
  .row { display: grid; gap: 4px; }
  .key { padding: 9px 3px; border-radius: 8px;
    background: #fff; border: 1px solid #dfe5ef;
    font-size: 13px; color: #0b1f44; cursor: pointer;
    text-align: center; user-select: none; font-family: inherit;
    transition: transform .05s, background .1s;
    min-height: 38px; display: flex; align-items: center;
    justify-content: center; }
  .key:active { transform: scale(.95); background: #eef2fb; }
  .key.accent { background: #eef2fb; border-color: #c5d1e9;
    color: #223a6b; font-weight: 600; }
  .key.active-mode { background: #3b5bf6; color: #fff;
    border-color: #3b5bf6; }
  .key.small { font-size: 11px; }
  .key.tiny { font-size: 9px; }

  /* Casio-style stacked fraction inside a key */
  .key .frac {
    display: flex; flex-direction: column; align-items: center;
    line-height: 1; font-size: 11px; width: 100%;
  }
  .key .frac span { padding: 1px 0; display: block; width: 100%;
    text-align: center; }
  .key .frac span:first-child { border-bottom: 1px solid #7b8ba6; }

  .basegrid { padding: 6px 10px; background: #f6f8fd;
    border-top: 1px solid #e4e9f2;
    display: flex; flex-direction: column; gap: 4px; }
  .basegrid.collapsed { display: none; }

  .action { padding: 10px 3px; border-radius: 8px;
    background: #eef2fb; border: 1px solid #dfe5ef;
    color: #223a6b; font-size: 11px; font-weight: 600;
    cursor: pointer; font-family: inherit; text-align: center;
    min-height: 38px; display: flex; align-items: center;
    justify-content: center; }
  .action.primary { background: #3b5bf6; color: #fff;
    border-color: #3b5bf6; }
  .action:active { transform: scale(.97); }

  .result { padding: 10px 16px; background: #0b1f44; color: #fff;
    font-family: 'SF Mono', Menlo, monospace; font-size: 10px;
    line-height: 1.5; max-height: 0; overflow: hidden;
    transition: max-height .25s ease; }
  .result.shown { max-height: 260px; overflow-y: auto; }
  .result .label { color: #9ab; margin-bottom: 4px; }
  .result pre { margin: 0; color: #cfe2ff; white-space: pre-wrap; }
</style>
</head>
<body>

<div class="phone">
  <div class="topbar">
    <div class="logo">M</div>
    <div>
      <div class="brand">MethodMarker</div>
      <div class="tag">Better Methods. Higher Marks.</div>
    </div>
  </div>

  <div class="working">
    <div class="head">
      <span>&#9998; Working</span>
      <span class="steps" id="stepCount">0 steps</span>
    </div>
    <div class="lines">
      <div class="nums" id="nums"></div>
      <div class="text" id="text" contenteditable="true" spellcheck="false"></div>
    </div>
  </div>

  <div class="tabs" id="tabs"></div>

  <div class="grid" id="grid"></div>

  <div class="basegrid" id="basegrid"></div>

  <div class="result" id="result"></div>
</div>

<script>
const LAYOUT = __LAYOUT__;
const state = { activeTab: LAYOUT.tabs_order[0], collapsed: false, abcMode: false };

const textEl = document.getElementById("text");
const numsEl = document.getElementById("nums");
const stepCountEl = document.getElementById("stepCount");

const TO_ENGINE = {
  "θ":"theta","α":"alpha","β":"beta","γ":"gamma","δ":"delta","ε":"epsilon",
  "λ":"lambda","μ":"mu","π":"pi","ρ":"rho","σ":"sigma","τ":"tau","φ":"phi",
  "ψ":"psi","ω":"omega",
  "Σ":"sum","∑":"sum","Δ":"Delta","Ω":"Omega",
  "√":"sqrt","∛":"cbrt","ⁿ√":"nroot",
  "≤":"<=","≥":">=","≠":"!=","±":"+/-","≈":"approx",
  "×":" x ","÷":" / ",
  "⇒":" => ","⇔":" <=> ","∴":" therefore ","∵":" because ",
  "→":"->","∞":"inf","∈":" in ","ℝ":"R","ℕ":"N","ℤ":"Z",
  "∠":"angle ","△":"triangle ","∥":"parallel ","⊥":"perp ",
  "≅":"cong ","∼":"similar ",
  "Â":"A^","B̂":"B^","Ĉ":"C^","D̂":"D^","Ê":"E^","F̂":"F^",
  "x̄":"xbar","ȳ":"ybar","ŷ":"yhat","σ²":"sigma^2",
  "x²":"x^2","y²":"y^2","a²":"a^2","b²":"b^2","c²":"c^2","n²":"n^2",
  "x³":"x^3","xⁿ":"x^n","aⁿ":"a^n","bⁿ":"b^n","x⁻¹":"x^-1",
  "sin²":"sin^2","cos²":"cos^2","tan²":"tan^2",
  "sin⁻¹":"asin","cos⁻¹":"acos","tan⁻¹":"atan",
  "d²y/dx²":"d2y/dx2","Dₓ[ ]":"Dx[]",
  "Tₙ":"Tn","Sₙ":"Sn","Tₙ₊₁":"T_(n+1)","Tₙ₋₁":"T_(n-1)",
  "S∞":"S_inf","n→∞":"n->inf",
  "f⁻¹":"f^-1","g⁻¹":"g^-1",
  "∪":" union ","∩":" intersect ",
  "(x;y)":"(x;y)","(x₁;y₁)":"(x1;y1)","(x₂;y₂)":"(x2;y2)",
  "tanθ":"tan(theta)",
};

function toEngine(text) {
  let out = text;
  const syms = Object.keys(TO_ENGINE).sort((a,b) => b.length - a.length);
  for (const s of syms) out = out.split(s).join(TO_ENGINE[s]);
  return out;
}

function refreshNumbers() {
  const lines = textEl.innerText.split("\n");
  numsEl.innerHTML = lines.map((_, i) => `<div>${i + 1}</div>`).join("");
  const n = lines.length;
  stepCountEl.textContent = (n === 1 && lines[0] === "")
    ? "0 steps" : `${n} step${n === 1 ? "" : "s"}`;
}
textEl.addEventListener("input", refreshNumbers);
refreshNumbers();

function getCaretContext() {
  const sel = window.getSelection();
  if (!sel.rangeCount) return { before: "", after: "" };
  const r = sel.getRangeAt(0);
  const pre = r.cloneRange(); pre.selectNodeContents(textEl);
  pre.setEnd(r.startContainer, r.startOffset);
  const before = pre.toString();
  const post = r.cloneRange(); post.selectNodeContents(textEl);
  post.setStart(r.endContainer, r.endOffset);
  const after = post.toString();
  return { before, after };
}

function insertAtCaret(str) {
  textEl.focus();
  const sel = window.getSelection();
  if (!sel.rangeCount) return;
  const r = sel.getRangeAt(0);
  r.deleteContents();
  const node = document.createTextNode(str);
  r.insertNode(node);
  r.setStartAfter(node);
  r.setEndAfter(node);
  sel.removeAllRanges();
  sel.addRange(r);
  refreshNumbers();
}

function moveCursorBack(n) {
  const sel = window.getSelection();
  if (!sel.rangeCount || n <= 0) return;
  const r = sel.getRangeAt(0);
  if (r.startContainer.nodeType === 3 && r.startOffset >= n) {
    const nr = r.cloneRange();
    nr.setStart(nr.startContainer, nr.startOffset - n);
    nr.collapse(true);
    sel.removeAllRanges(); sel.addRange(nr);
  }
}

// Find next "()" pair after the cursor and jump inside
function jumpToNextParens() {
  const sel = window.getSelection();
  if (!sel.rangeCount) return;
  const r = sel.getRangeAt(0);
  const full = textEl.innerText;
  // get current character offset
  const pre = r.cloneRange();
  pre.selectNodeContents(textEl);
  pre.setEnd(r.startContainer, r.startOffset);
  const offset = pre.toString().length;
  // find next '()' after offset
  const idx = full.indexOf("()", offset + 1);
  if (idx < 0) return;
  // set cursor inside the parens (idx+1)
  const walker = document.createTreeWalker(textEl, NodeFilter.SHOW_TEXT);
  let pos = 0, targetNode = null, targetOffset = 0;
  while (walker.nextNode()) {
    const len = walker.currentNode.textContent.length;
    if (pos + len >= idx + 1) {
      targetNode = walker.currentNode;
      targetOffset = idx + 1 - pos;
      break;
    }
    pos += len;
  }
  if (targetNode) {
    const nr = document.createRange();
    nr.setStart(targetNode, targetOffset);
    nr.collapse(true);
    sel.removeAllRanges(); sel.addRange(nr);
  }
}

function smartInsert(key) {
  if (key.e === "__back__") {
    textEl.focus();
    document.execCommand("delete");
    refreshNumbers();
    return;
  }
  if (key.e === "__newline__") { insertAtCaret("\n"); return; }
  if (key.e === "__done__")    { submit(); return; }
  if (key.e === "__abc__") {
    state.abcMode = !state.abcMode;
    renderGrid();
    renderBase();
    return;
  }
  if (key.e === "__collapse__") {
    state.collapsed = !state.collapsed;
    document.getElementById("grid").classList.toggle("collapsed", state.collapsed);
    document.getElementById("basegrid").classList.toggle("collapsed", state.collapsed);
    return;
  }
  if (key.e === "__tab__") { jumpToNextParens(); return; }
  const { before, after } = getCaretContext();
  let ins = key.d;
  if (key.sp) {
    const needLead  = before.length > 0 && !/\s$/.test(before);
    const needTrail = after.length  > 0 && !/^\s/.test(after);
    ins = (needLead ? " " : "") + ins.trim() + (needTrail ? " " : "");
  }
  insertAtCaret(ins);
  // Cursor repositioning
  if (typeof key.skip === "number" && key.skip > 0) {
    moveCursorBack(key.skip);
  }
}

// ---------- tabs ----------
const tabsEl = document.getElementById("tabs");
LAYOUT.tabs_order.forEach(name => {
  const b = document.createElement("button");
  b.className = "tab" + (name === state.activeTab ? " active" : "");
  b.textContent = name;
  b.onclick = () => {
    state.activeTab = name;
    document.querySelectorAll(".tab").forEach(t =>
      t.classList.toggle("active", t.textContent === name));
    renderGrid();
  };
  tabsEl.appendChild(b);
});

// ---------- grid ----------
const gridEl = document.getElementById("grid");
function renderGrid() {
  if (state.abcMode) {
    gridEl.innerHTML = "";
    const rows = LAYOUT.base_layer.abc || [];
    const cols = Math.max(...rows.map(r => r.length));
    rows.forEach(row => {
      const r = document.createElement("div");
      r.className = "row";
      r.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
      row.forEach(key => {
        const k = document.createElement("div");
        k.className = "key";
        k.textContent = key.l;
        k.onclick = () => smartInsert(key);
        r.appendChild(k);
      });
      for (let i = row.length; i < cols; i++)
        r.appendChild(document.createElement("div"));
      gridEl.appendChild(r);
    });
    return;
  }
  const rows = LAYOUT.tabs[state.activeTab] || [];
  gridEl.innerHTML = "";
  const cols = Math.max(...rows.map(r => r.length));
  rows.forEach(row => {
    const r = document.createElement("div");
    r.className = "row";
    r.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
    row.forEach(key => {
      const k = document.createElement("div");
      const accent = ["or","=","⇒","∴","∵"].includes(key.l) ? " accent" : "";
      const small = key.l.length > 4 ? " small" : "";
      const tiny = key.l.length > 8 ? " tiny" : "";
      k.className = "key" + accent + small + tiny;
      if (key.h) {
        k.innerHTML = key.h;
      } else {
        k.textContent = key.l;
      }
      k.onclick = () => smartInsert(key);
      r.appendChild(k);
    });
    for (let i = row.length; i < cols; i++)
      r.appendChild(document.createElement("div"));
    gridEl.appendChild(r);
  });
}

// ---------- base ----------
function renderBase() {
  const b = LAYOUT.base_layer;
  const container = document.getElementById("basegrid");
  container.innerHTML = "";
  const layout = [
    { keys: b.variables, cols: 12 },
    { keys: b.digits,    cols: 12 },
    { keys: b.operators, cols: 10 },
    { keys: b.actions,   cols: 6  },
  ];
  layout.forEach(({keys, cols}) => {
    const row = document.createElement("div");
    row.className = "row";
    row.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
    keys.forEach(key => {
      const isDone = key.e === "__done__";
      const isAction = key.e && key.e.startsWith("__");
      const el = document.createElement("div");
      el.className = isDone ? "action primary"
                           : (isAction ? "action" : "key");
      if (key.e === "__abc__" && state.abcMode) el.classList.add("active-mode");
      if (key.h) el.innerHTML = key.h;
      else       el.textContent = key.l;
      el.onclick = () => smartInsert(key);
      row.appendChild(el);
    });
    container.appendChild(row);
  });
}

// ---------- submit ----------
const resultEl = document.getElementById("result");
const CMD = {
  or: " or ", then: " ∴ ", because: " ∵ ",
  implies: " ⇒ ", iff: " ⇔ ",
  sqrt: "sqrt", cbrt: "cbrt", pi: "pi",
  theta: "theta", alpha: "alpha", beta: "beta",
  deg: "deg", pm: "+/-", le: "<=", ge: ">=",
  in: " in ", R: "R", sum: "sum", inf: "inf",
  dy: "dy/dx", dx: "d/dx", lim: "lim", to: "->",
};

function submit() {
  const rawLines = textEl.innerText.split("\n").map(l => l.trim()).filter(l => l.length);
  const slashExpanded = rawLines.map(l =>
    l.replace(/\/([A-Za-z][A-Za-z0-9_]*)/g, (m, name) => CMD[name] || m));
  const engineLines = slashExpanded.map(toEngine);
  const payload = {
    learner_lines: engineLines,
    display_lines: slashExpanded,
    tab_used: state.activeTab,
    steps: engineLines.length,
  };
  resultEl.classList.add("shown");
  resultEl.innerHTML =
    "<div class='label'>submitted to step_grader</div>" +
    "<pre>" + JSON.stringify(payload, null, 2) + "</pre>";
}

textEl.focus();
renderGrid();
renderBase();
</script>
</body>
</html>
"""

HTML = HTML_TEMPLATE.replace("__LAYOUT__", json.dumps(kb, ensure_ascii=False))
HTML_PATH.write_text(HTML, encoding="utf-8")
print()
print(f"  ✓ {HTML_PATH.name} rebuilt  ({HTML_PATH.stat().st_size:,} bytes)")

webbrowser.open(HTML_PATH.as_uri())
print(f"  → opened in browser")
print()
print("=" * 72)
print("TEST IN BROWSER")
print("=" * 72)
print("  Algebra tab:")
print("    Row 2:  tap x² y² a² b² c² n²  →  all square into the working area")
print("    Row 3:  tap xⁿ aⁿ bⁿ Tₙ Sₙ     →  general exponents on any var")
print("    Row 4:  tap √  →  inserts √() with cursor inside")
print("            tap ∛  →  ∛() cursor inside")
print("            tap ⁿ√ →  ⁿ√() cursor inside")
print("            tap ±  →  inserts ±")
print("    Row 6:  tap QF  →  inserts the full quadratic formula template")
print()
print("  Base operators row:")
print("    Tap a/b (stacked) → inserts ()/() with cursor in numerator")
print()
print("  Test the quadratic formula workflow:")
print("    1. On Algebra tap QF")
print("       → x = (-b ± √(b² - 4ac))/(2a)")
print("    2. Then substitute and simplify")
````

````python
# Keyboard v6.1 — Flat Casio layout, true visual math notation
import json
import webbrowser
from pathlib import Path

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
LAYOUT_PATH = ROOT / "data" / "processed" / "tutor" / "keyboard_layout.json"
HTML_PATH = ROOT / "docs" / "keyboard_preview.html"

def K(label, display=None, engine=None, sp=False, trail=False, skip=None, html=None):
    d = {"l": label}
    d["d"] = display if display is not None else label
    d["e"] = engine if engine is not None else d["d"]
    if sp:    d["sp"] = True
    if trail: d["trail"] = True
    if skip is not None: d["skip"] = skip
    if html:  d["h"] = html
    return d

# ============================================================
# Load existing layout
# ============================================================
kb = json.loads(LAYOUT_PATH.read_text(encoding="utf-8"))

# --- Rebuild Base Operators (Flat, No SHIFT, Visual Math) ---
kb["base_layer"]["operators"] = [
    K("+", " + ", " + ", sp=True),
    K("−", " - ", " - ", sp=True),
    K("×", " × ", " x ", sp=True),
    K("÷", " ÷ ", " / ", sp=True),
    K("(", "("), 
    K(")", ")"),
    
    # True Visual Fraction (Top box, line, bottom box)
    K("▢/▢", "()/()", "()/()", skip=4,
      html="<span class='frac'><span class='box'></span><span class='line'></span><span class='box'></span></span>"),
    
    # True Visual Exponent (Base box, superscript box)
    K("▢^▢", "▢^▢", "^",
      html="<span class='exp'><span class='base'></span><span class='power'></span></span>"),
    
    # True Visual Square Root (Radical symbol + box)
    K("√", "√▢", "sqrt(",
      html="<span class='sqrt'>&#8730;<span class='sqrt-box'></span></span>"),
    
    K("x²", "x²", "x^2"),
    K("π", "π", "pi"),
    K("θ", "θ", "theta"),
]

# --- Rebuild Algebra Tab (Flat, No SHIFT, No ln) ---
kb["tabs"]["Algebra"] = [
    # Row 1 — Relations
    [K("=", " = ", " = ", sp=True),
     K("or", " or ", " or ", sp=True),
     K("⇒", " ⇒ ", " => ", sp=True),
     K("∴", " ∴ ", " therefore ", sp=True),
     K("∵", " ∵ ", " because ", sp=True)],
    
    # Row 2 — Squares
    [K("x²", "x²", "x^2"),
     K("y²", "y²", "y^2"),
     K("a²", "a²", "a^2"),
     K("b²", "b²", "b^2"),
     K("c²", "c²", "c^2")],
    
    # Row 3 — Higher powers
    [K("x³", "x³", "x^3"),
     K("xⁿ", "xⁿ", "x^n"),
     K("aⁿ", "aⁿ", "a^n"),
     K("bⁿ", "bⁿ", "b^n"),
     K("Tₙ", "Tₙ", "Tn")],
    
    # Row 4 — Roots and Reciprocals
    [K("√", "√▢", "sqrt(",
       html="<span class='sqrt'>&#8730;<span class='sqrt-box'></span></span>"),
     K("∛", "∛▢", "cbrt("),
     K("ⁿ√", "ⁿ√▢", "nroot("),
     K("±", "±", "+/-"),
     K("x⁻¹", "x⁻¹", "x^-1")],
    
    # Row 5 — Inequalities
    [K("<", "<"), K(">", ">"),
     K("≤", "≤", "<="), K("≥", "≥", ">="),
     K("≠", "≠", "!=")],
    
    # Row 6 — Extras
    [K("π", "π", "pi"),
     K("∞", "∞", "inf"),
     K("QF", "x = (-b ± √(b² - 4ac))/(2a)",
       "x = (-b ± sqrt(b^2 - 4ac))/(2a)"),
     K("Σ", "Σ", "sum")],
]

kb["schema_version"] = 6.1
LAYOUT_PATH.write_text(
    json.dumps(kb, indent=2, ensure_ascii=False), encoding="utf-8")

print(f"  ✓ keyboard_layout.json v6.1")
print(f"    Algebra tab: {len(kb['tabs']['Algebra'])} rows, "
      f"{sum(len(r) for r in kb['tabs']['Algebra'])} keys")
print("    Base operators: True visual fraction, exponent, and sqrt boxes.")

# ============================================================
# Rebuild HTML with True Visual Math CSS
# ============================================================
HTML_TEMPLATE = r"""<!doctype html>
<html lang="en">
<head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width,initial-scale=1">
<title>MethodKeyboard — v6.1</title>
<style>
  * { box-sizing: border-box; -webkit-tap-highlight-color: transparent; }
  body { margin: 0; background: #eef2f8;
    font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif;
    display: flex; justify-content: center; padding: 12px; min-height: 100vh; }
  .phone { width: 440px; max-width: 100%; background: #fff;
    border-radius: 20px; overflow: hidden;
    box-shadow: 0 8px 32px rgba(20, 40, 90, .12);
    display: flex; flex-direction: column; }
  .topbar { display: flex; align-items: center; gap: 10px;
    padding: 12px 16px; background: #f6f8fd;
    border-bottom: 1px solid #e4e9f2; }
  .topbar .logo { width: 28px; height: 28px; border-radius: 7px;
    background: linear-gradient(135deg, #3b5bf6, #5b8cf6);
    display: flex; align-items: center; justify-content: center;
    color: #fff; font-weight: 700; font-size: 15px; }
  .topbar .brand { font-weight: 700; font-size: 14px; color: #0b1f44; }
  .topbar .tag { font-size: 10px; color: #6b7a99; }

  .working { padding: 12px 16px; background: #fff; }
  .working .head { display: flex; justify-content: space-between;
    align-items: center; font-size: 12px; color: #465a80;
    font-weight: 600; margin-bottom: 6px; }
  .working .steps { color: #6b7a99; font-weight: 500; }
  .lines { display: grid; grid-template-columns: 22px 1fr;
    column-gap: 8px; font-size: 15px; line-height: 26px;
    color: #0b1f44; max-height: 180px; overflow-y: auto; }
  .nums { color: #9aa8c2; font-size: 11px; text-align: right;
    user-select: none; padding-top: 2px; }
  .nums div { height: 26px; line-height: 26px; }
  .text { white-space: pre-wrap; word-break: break-word;
    min-height: 26px; outline: none;
    font-family: 'SF Mono', Menlo, Consolas, monospace; }

  .tabs { display: flex; overflow-x: auto; gap: 4px;
    padding: 6px 12px; background: #f6f8fd;
    border-top: 1px solid #e4e9f2;
    border-bottom: 1px solid #e4e9f2; scrollbar-width: none; }
  .tabs::-webkit-scrollbar { display: none; }
  .tab { flex: none; padding: 6px 11px; border-radius: 16px;
    font-size: 11px; font-weight: 600; color: #465a80;
    background: transparent; border: none; cursor: pointer;
    white-space: nowrap; }
  .tab.active { background: #3b5bf6; color: #fff; }

  .grid { padding: 8px 10px; background: #fff;
    display: flex; flex-direction: column; gap: 4px;
    transition: max-height .25s ease; overflow: hidden;
    max-height: 900px; }
  .grid.collapsed { max-height: 0; padding: 0; }
  .row { display: grid; gap: 4px; }
  .key { padding: 9px 3px; border-radius: 8px;
    background: #fff; border: 1px solid #dfe5ef;
    font-size: 13px; color: #0b1f44; cursor: pointer;
    text-align: center; user-select: none; font-family: inherit;
    transition: transform .05s, background .1s;
    min-height: 38px; display: flex; align-items: center;
    justify-content: center; }
  .key:active { transform: scale(.95); background: #eef2fb; }
  .key.accent { background: #eef2fb; border-color: #c5d1e9;
    color: #223a6b; font-weight: 600; }
  .key.active-mode { background: #3b5bf6; color: #fff;
    border-color: #3b5bf6; }
  .key.small { font-size: 11px; }
  .key.tiny { font-size: 9px; }

  /* --- True Visual Fraction --- */
  .key .frac {
    display: flex; flex-direction: column; align-items: center;
    justify-content: center; width: 100%; height: 100%; gap: 2px;
  }
  .key .frac .box {
    width: 14px; height: 10px; border: 1px solid #7b8ba6;
    border-radius: 2px;
  }
  .key .frac .line {
    width: 20px; height: 1px; background: #7b8ba6;
  }

  /* --- True Visual Exponent --- */
  .key .exp {
    position: relative; display: flex; align-items: flex-end;
    justify-content: center; width: 100%; height: 100%;
  }
  .key .exp .base {
    width: 12px; height: 12px; border: 1px solid #7b8ba6;
    border-radius: 2px; margin-bottom: 4px;
  }
  .key .exp .power {
    width: 8px; height: 8px; border: 1px solid #7b8ba6;
    border-radius: 2px; position: absolute; top: 4px; right: 10px;
  }

  /* --- True Visual Square Root --- */
  .key .sqrt {
    display: flex; align-items: center; font-size: 18px;
    font-weight: bold; line-height: 1;
  }
  .key .sqrt .sqrt-box {
    width: 12px; height: 14px; border-top: 1px solid #7b8ba6;
    border-right: 1px solid #7b8ba6; border-bottom: 1px solid #7b8ba6;
    margin-left: -2px; margin-top: 2px;
  }

  .basegrid { padding: 6px 10px; background: #f6f8fd;
    border-top: 1px solid #e4e9f2;
    display: flex; flex-direction: column; gap: 4px; }
  .basegrid.collapsed { display: none; }

  .action { padding: 10px 3px; border-radius: 8px;
    background: #eef2fb; border: 1px solid #dfe5ef;
    color: #223a6b; font-size: 11px; font-weight: 600;
    cursor: pointer; font-family: inherit; text-align: center;
    min-height: 38px; display: flex; align-items: center;
    justify-content: center; }
  .action.primary { background: #3b5bf6; color: #fff;
    border-color: #3b5bf6; }
  .action:active { transform: scale(.97); }

  .result { padding: 10px 16px; background: #0b1f44; color: #fff;
    font-family: 'SF Mono', Menlo, monospace; font-size: 10px;
    line-height: 1.5; max-height: 0; overflow: hidden;
    transition: max-height .25s ease; }
  .result.shown { max-height: 260px; overflow-y: auto; }
  .result .label { color: #9ab; margin-bottom: 4px; }
  .result pre { margin: 0; color: #cfe2ff; white-space: pre-wrap; }
</style>
</head>
<body>

<div class="phone">
  <div class="topbar">
    <div class="logo">M</div>
    <div>
      <div class="brand">MethodMarker</div>
      <div class="tag">Better Methods. Higher Marks.</div>
    </div>
  </div>

  <div class="working">
    <div class="head">
      <span>&#9998; Working</span>
      <span class="steps" id="stepCount">0 steps</span>
    </div>
    <div class="lines">
      <div class="nums" id="nums"></div>
      <div class="text" id="text" contenteditable="true" spellcheck="false"></div>
    </div>
  </div>

  <div class="tabs" id="tabs"></div>

  <div class="grid" id="grid"></div>

  <div class="basegrid" id="basegrid"></div>

  <div class="result" id="result"></div>
</div>

<script>
const LAYOUT = __LAYOUT__;
const state = { activeTab: LAYOUT.tabs_order[0], collapsed: false, abcMode: false };

const textEl = document.getElementById("text");
const numsEl = document.getElementById("nums");
const stepCountEl = document.getElementById("stepCount");

const TO_ENGINE = {
  "θ":"theta","α":"alpha","β":"beta","γ":"gamma","δ":"delta","ε":"epsilon",
  "λ":"lambda","μ":"mu","π":"pi","ρ":"rho","σ":"sigma","τ":"tau","φ":"phi",
  "ψ":"psi","ω":"omega",
  "Σ":"sum","∑":"sum","Δ":"Delta","Ω":"Omega",
  "√":"sqrt","∛":"cbrt","ⁿ√":"nroot",
  "≤":"<=","≥":">=","≠":"!=","±":"+/-","≈":"approx",
  "×":" x ","÷":" / ",
  "⇒":" => ","⇔":" <=> ","∴":" therefore ","∵":" because ",
  "→":"->","∞":"inf","∈":" in ","ℝ":"R","ℕ":"N","ℤ":"Z",
  "x²":"x^2","y²":"y^2","a²":"a^2","b²":"b^2","c²":"c^2","n²":"n^2",
  "x³":"x^3","xⁿ":"x^n","aⁿ":"a^n","bⁿ":"b^n","x⁻¹":"x^-1",
  "sin⁻¹":"asin","cos⁻¹":"acos","tan⁻¹":"atan",
  "Tₙ":"Tn","Sₙ":"Sn",
};

function toEngine(text) {
  let out = text;
  const syms = Object.keys(TO_ENGINE).sort((a,b) => b.length - a.length);
  for (const s of syms) out = out.split(s).join(TO_ENGINE[s]);
  return out;
}

function refreshNumbers() {
  const lines = textEl.innerText.split("\n");
  numsEl.innerHTML = lines.map((_, i) => `<div>${i + 1}</div>`).join("");
  const n = lines.length;
  stepCountEl.textContent = (n === 1 && lines[0] === "")
    ? "0 steps" : `${n} step${n === 1 ? "" : "s"}`;
}
textEl.addEventListener("input", refreshNumbers);
refreshNumbers();

function getCaretContext() {
  const sel = window.getSelection();
  if (!sel.rangeCount) return { before: "", after: "" };
  const r = sel.getRangeAt(0);
  const pre = r.cloneRange(); pre.selectNodeContents(textEl);
  pre.setEnd(r.startContainer, r.startOffset);
  const before = pre.toString();
  const post = r.cloneRange(); post.selectNodeContents(textEl);
  post.setStart(r.endContainer, r.endOffset);
  const after = post.toString();
  return { before, after };
}

function insertAtCaret(str) {
  textEl.focus();
  const sel = window.getSelection();
  if (!sel.rangeCount) return;
  const r = sel.getRangeAt(0);
  r.deleteContents();
  const node = document.createTextNode(str);
  r.insertNode(node);
  r.setStartAfter(node);
  r.setEndAfter(node);
  sel.removeAllRanges();
  sel.addRange(r);
  refreshNumbers();
}

function moveCursorBack(n) {
  const sel = window.getSelection();
  if (!sel.rangeCount || n <= 0) return;
  const r = sel.getRangeAt(0);
  if (r.startContainer.nodeType === 3 && r.startOffset >= n) {
    const nr = r.cloneRange();
    nr.setStart(nr.startContainer, nr.startOffset - n);
    nr.collapse(true);
    sel.removeAllRanges(); sel.addRange(nr);
  }
}

function smartInsert(key) {
  if (key.e === "__back__") {
    textEl.focus();
    document.execCommand("delete");
    refreshNumbers();
    return;
  }
  if (key.e === "__newline__") { insertAtCaret("\n"); return; }
  if (key.e === "__done__")    { submit(); return; }
  if (key.e === "__abc__") {
    state.abcMode = !state.abcMode;
    renderGrid();
    renderBase();
    return;
  }
  if (key.e === "__collapse__") {
    state.collapsed = !state.collapsed;
    document.getElementById("grid").classList.toggle("collapsed", state.collapsed);
    document.getElementById("basegrid").classList.toggle("collapsed", state.collapsed);
    return;
  }
  
  // Handle skip logic for fractions/exponents
  const { before, after } = getCaretContext();
  let ins = key.d;
  if (key.sp) {
    const needLead  = before.length > 0 && !/\s$/.test(before);
    const needTrail = after.length  > 0 && !/^\s/.test(after);
    ins = (needLead ? " " : "") + ins.trim() + (needTrail ? " " : "");
  }
  
  if (key.skip === 4 && key.d === "()/()") {
      // Special handling for fraction to place cursor in numerator
      insertAtCaret("()/()");
      moveCursorBack(4);
  } else {
      insertAtCaret(ins);
      if (typeof key.skip === "number" && key.skip > 0) {
        moveCursorBack(key.skip);
      }
  }
}

// ---------- tabs ----------
const tabsEl = document.getElementById("tabs");
LAYOUT.tabs_order.forEach(name => {
  const b = document.createElement("button");
  b.className = "tab" + (name === state.activeTab ? " active" : "");
  b.textContent = name;
  b.onclick = () => {
    state.activeTab = name;
    document.querySelectorAll(".tab").forEach(t =>
      t.classList.toggle("active", t.textContent === name));
    renderGrid();
  };
  tabsEl.appendChild(b);
});

// ---------- grid ----------
const gridEl = document.getElementById("grid");
function renderGrid() {
  if (state.abcMode) {
    gridEl.innerHTML = "";
    const rows = LAYOUT.base_layer.abc || [];
    const cols = Math.max(...rows.map(r => r.length));
    rows.forEach(row => {
      const r = document.createElement("div");
      r.className = "row";
      r.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
      row.forEach(key => {
        const k = document.createElement("div");
        k.className = "key";
        k.textContent = key.l;
        k.onclick = () => smartInsert(key);
        r.appendChild(k);
      });
      for (let i = row.length; i < cols; i++)
        r.appendChild(document.createElement("div"));
      gridEl.appendChild(r);
    });
    return;
  }
  const rows = LAYOUT.tabs[state.activeTab] || [];
  gridEl.innerHTML = "";
  const cols = Math.max(...rows.map(r => r.length));
  rows.forEach(row => {
    const r = document.createElement("div");
    r.className = "row";
    r.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
    row.forEach(key => {
      const k = document.createElement("div");
      const accent = ["or","=","⇒","∴","∵"].includes(key.l) ? " accent" : "";
      const small = key.l.length > 4 ? " small" : "";
      const tiny = key.l.length > 8 ? " tiny" : "";
      k.className = "key" + accent + small + tiny;
      if (key.h) {
        k.innerHTML = key.h;
      } else {
        k.textContent = key.l;
      }
      k.onclick = () => smartInsert(key);
      r.appendChild(k);
    });
    for (let i = row.length; i < cols; i++)
      r.appendChild(document.createElement("div"));
    gridEl.appendChild(r);
  });
}

// ---------- base ----------
function renderBase() {
  const b = LAYOUT.base_layer;
  const container = document.getElementById("basegrid");
  container.innerHTML = "";
  const layout = [
    { keys: b.variables, cols: 12 },
    { keys: b.digits,    cols: 12 },
    { keys: b.operators, cols: 10 },
    { keys: b.actions,   cols: 6  },
  ];
  layout.forEach(({keys, cols}) => {
    const row = document.createElement("div");
    row.className = "row";
    row.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
    keys.forEach(key => {
      const isDone = key.e === "__done__";
      const isAction = key.e && key.e.startsWith("__");
      const el = document.createElement("div");
      el.className = isDone ? "action primary"
                           : (isAction ? "action" : "key");
      if (key.e === "__abc__" && state.abcMode) el.classList.add("active-mode");
      if (key.h) el.innerHTML = key.h;
      else       el.textContent = key.l;
      el.onclick = () => smartInsert(key);
      row.appendChild(el);
    });
    container.appendChild(row);
  });
}

// ---------- submit ----------
const resultEl = document.getElementById("result");
const CMD = {
  or: " or ", then: " ∴ ", because: " ∵ ",
  implies: " ⇒ ", iff: " ⇔ ",
  sqrt: "sqrt", cbrt: "cbrt", pi: "pi",
  theta: "theta", alpha: "alpha", beta: "beta",
  deg: "deg", pm: "+/-", le: "<=", ge: ">=",
  in: " in ", R: "R", sum: "sum", inf: "inf",
  dy: "dy/dx", dx: "d/dx", lim: "lim", to: "->",
};

function submit() {
  const rawLines = textEl.innerText.split("\n").map(l => l.trim()).filter(l => l.length);
  const slashExpanded = rawLines.map(l =>
    l.replace(/\/([A-Za-z][A-Za-z0-9_]*)/g, (m, name) => CMD[name] || m));
  const engineLines = slashExpanded.map(toEngine);
  const payload = {
    learner_lines: engineLines,
    display_lines: slashExpanded,
    tab_used: state.activeTab,
    steps: engineLines.length,
  };
  resultEl.classList.add("shown");
  resultEl.innerHTML =
    "<div class='label'>submitted to step_grader</div>" +
    "<pre>" + JSON.stringify(payload, null, 2) + "</pre>";
}

textEl.focus();
renderGrid();
renderBase();
</script>
</body>
</html>
"""

HTML = HTML_TEMPLATE.replace("__LAYOUT__", json.dumps(kb, ensure_ascii=False))
HTML_PATH.write_text(HTML, encoding="utf-8")
print()
print(f"  ✓ {HTML_PATH.name} rebuilt  ({HTML_PATH.stat().st_size:,} bytes)")

webbrowser.open(HTML_PATH.as_uri())
print(f"  → opened in browser")
print()
print("=" * 72)
print("TEST IN BROWSER (v6.1)")
print("=" * 72)
print("  Base operators row:")
print("    Tap ▢/▢ (stacked) → inserts ()/() with cursor in numerator")
print("    Tap ▢^▢ (exponent) → inserts visual base and power boxes")
print("    Tap √ (radical)    → inserts √() with cursor inside")
print()
print("  Algebra tab:")
print("    Row 2:  x² y² a² b² c²")
print("    Row 3:  x³ xⁿ aⁿ bⁿ Tₙ")
print("    Row 4:  √ ∛ ⁿ√ ± x⁻¹")
print("    Row 6:  π ∞ QF Σ")
````

````python
# Keyboard v6.4 — Pure CSS: Editable bounds, real lines, fixed layout
import json
import webbrowser
from pathlib import Path

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
LAYOUT_PATH = ROOT / "data" / "processed" / "tutor" / "keyboard_layout.json"
HTML_PATH = ROOT / "docs" / "keyboard_preview.html"

def K(label, display=None, engine=None, sp=False, skip=None, html=None):
    d = {"l": label}
    d["d"] = display if display is not None else label
    d["e"] = engine if engine is not None else d["d"]
    if sp:    d["sp"] = True
    if skip is not None: d["skip"] = skip
    if html:  d["h"] = html
    return d

# ============================================================
# Load existing layout
# ============================================================
kb = json.loads(LAYOUT_PATH.read_text(encoding="utf-8"))

# --- Rebuild Base Operators (Pure CSS Visuals) ---
kb["base_layer"]["operators"] = [
    K("+", " + ", " + ", sp=True),
    K("−", " - ", " - ", sp=True),
    K("×", " × ", " * ", sp=True),
    K("÷", " ÷ ", " / ", sp=True),
    K("(", "("), 
    K(")", ")"),
    
    # True CSS Fraction (Top box, line, bottom box)
    K("a/b", "()/()", "()/()", 
      html="<span class='math-frac'><span class='num' contenteditable='true'>▢</span><span class='den' contenteditable='true'>▢</span></span>"),
    
    # True CSS Exponent (Base, superscript)
    K("x^n", "▢^▢", "^", 
      html="<span class='math-exp'><span class='base' contenteditable='true'>▢</span><span class='power' contenteditable='true'>▢</span></span>"),
    
    # True CSS Square Root
    K("√", "√▢", "sqrt(", 
      html="<span class='math-sqrt'><span class='radical'>√</span><span class='radicand' contenteditable='true'>▢</span></span>"),
    
    # Base Layer Logarithms (Always visible)
    K("log", "log(", "log10("),
    K("log_a", "log_▢(", "log_(",
      html="<span class='math-log'>log<sub contenteditable='true'>▢</sub>(</span>"),
    K("ln", "ln(", "ln("),
    
    K("x²", "x²", "x^2"),
    K("π", "π", "pi"),
    K("θ", "θ", "theta"),
]

# --- Rebuild Algebra Tab (Pure CSS, Editable Bounds) ---
kb["tabs"]["Algebra"] = [
    # Row 1 — Relations
    [K("=", " = ", " = ", sp=True),
     K("or", " or ", " or ", sp=True),
     K("⇒", " ⇒ ", " => ", sp=True),
     K("∴", " ∴ ", " therefore ", sp=True),
     K("∵", " ∵ ", " because ", sp=True)],
    
    # Row 2 — Squares
    [K("x²", "x²", "x^2"),
     K("y²", "y²", "y^2"),
     K("a²", "a²", "a^2"),
     K("b²", "b²", "b^2"),
     K("c²", "c²", "c^2")],
    
    # Row 3 — Higher powers and Real Summation
    [K("x³", "x³", "x^3"),
     K("xⁿ", "xⁿ", "x^n"),
     K("aⁿ", "aⁿ", "a^n"),
     # THE REAL SUMMATION: Editable top and bottom bounds
     K("Σ", "Σ", "sum",
       html="<span class='math-sum'><span class='bound' contenteditable='true'>∞</span><span class='sigma'>Σ</span><span class='bound' contenteditable='true'>n=1</span></span>")],
    
    # Row 4 — Roots and Reciprocals
    [K("√", "√▢", "sqrt(", html="<span class='math-sqrt'><span class='radical'>√</span><span class='radicand' contenteditable='true'>▢</span></span>"),
     K("∛", "∛▢", "cbrt(", html="<span class='math-sqrt'><span class='radical'>∛</span><span class='radicand' contenteditable='true'>▢</span></span>"),
     K("ⁿ√", "ⁿ√▢", "nroot(", html="<span class='math-sqrt'><span class='radical'>ⁿ√</span><span class='radicand' contenteditable='true'>▢</span></span>"),
     K("±", " ± ", " ± "),
     K("x⁻¹", "x⁻¹", "x^-1")],
    
    # Row 5 — Inequalities
    [K("<", " < "), K(">", " > "),
     K("≤", " ≤ ", " <= "), K("≥", " ≥ ", " >= "),
     K("≠", " ≠ ", " != ")],
    
    # Row 6 — Extras
    [K("π", "π", "pi"),
     K("∞", "∞", "inf"),
     K("QF", "x = (-b ± √(b² - 4ac))/(2a)", "x = (-b ± sqrt(b^2 - 4ac))/(2a)"),
     K("Σ", "Σ", "sum",
       html="<span class='math-sum'><span class='bound' contenteditable='true'>∞</span><span class='sigma'>Σ</span><span class='bound' contenteditable='true'>n=1</span></span>")],
]

kb["schema_version"] = 6.4
LAYOUT_PATH.write_text(
    json.dumps(kb, indent=2, ensure_ascii=False), encoding="utf-8")

print(f"  ✓ keyboard_layout.json v6.4")
print("    Base operators: Expandable fraction, exponent, sqrt, and logs.")
print("    Algebra tab: Editable summation bounds (type p=1, k=0, etc.)")

# ============================================================
# Rebuild HTML with Pure CSS Math (No external libraries)
# ============================================================
HTML_TEMPLATE = r"""<!doctype html>
<html lang="en">
<head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width,initial-scale=1">
<title>MethodKeyboard — v6.4</title>
<style>
  * { box-sizing: border-box; -webkit-tap-highlight-color: transparent; }
  body { margin: 0; background: #eef2f8;
    font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif;
    display: flex; justify-content: center; padding: 12px; min-height: 100vh; }
  .phone { width: 600px; max-width: 100%; background: #fff;
    border-radius: 20px; overflow: hidden;
    box-shadow: 0 8px 32px rgba(20, 40, 90, .12);
    display: flex; flex-direction: column; }
  .topbar { display: flex; align-items: center; gap: 10px;
    padding: 12px 16px; background: #f6f8fd;
    border-bottom: 1px solid #e4e9f2; }
  .topbar .logo { width: 28px; height: 28px; border-radius: 7px;
    background: linear-gradient(135deg, #3b5bf6, #5b8cf6);
    display: flex; align-items: center; justify-content: center;
    color: #fff; font-weight: 700; font-size: 15px; }
  .topbar .brand { font-weight: 700; font-size: 14px; color: #0b1f44; }
  .topbar .tag { font-size: 10px; color: #6b7a99; }

  .working { padding: 12px 16px; background: #fff; }
  .working .head { display: flex; justify-content: space-between;
    align-items: center; font-size: 12px; color: #465a80;
    font-weight: 600; margin-bottom: 6px; }
  .working .steps { color: #6b7a99; font-weight: 500; }
  .lines { display: grid; grid-template-columns: 22px 1fr;
    column-gap: 8px; font-size: 18px; line-height: 32px;
    color: #0b1f44; max-height: 180px; overflow-y: auto; }
  .nums { color: #9aa8c2; font-size: 11px; text-align: right;
    user-select: none; padding-top: 6px; }
  .nums div { height: 32px; line-height: 32px; }
  .text { min-height: 32px; outline: none; white-space: pre-wrap; word-break: break-word; font-size: 18px; }

  .tabs { display: flex; overflow-x: auto; gap: 4px;
    padding: 6px 12px; background: #f6f8fd;
    border-top: 1px solid #e4e9f2;
    border-bottom: 1px solid #e4e9f2; scrollbar-width: none; }
  .tabs::-webkit-scrollbar { display: none; }
  .tab { flex: none; padding: 6px 11px; border-radius: 16px;
    font-size: 11px; font-weight: 600; color: #465a80;
    background: transparent; border: none; cursor: pointer;
    white-space: nowrap; }
  .tab.active { background: #3b5bf6; color: #fff; }

  .grid { padding: 8px 10px; background: #fff;
    display: flex; flex-direction: column; gap: 6px;
    transition: max-height .25s ease; overflow: hidden;
    max-height: 900px; }
  .grid.collapsed { max-height: 0; padding: 0; }
  .row { display: grid; gap: 6px; }
  .key { padding: 9px 3px; border-radius: 8px;
    background: #fff; border: 1px solid #dfe5ef;
    font-size: 14px; color: #0b1f44; cursor: pointer;
    text-align: center; user-select: none; font-family: inherit;
    transition: transform .05s, background .1s;
    min-height: 48px; display: flex; align-items: center;
    justify-content: center; }
  .key:active { transform: scale(.95); background: #eef2fb; }
  .key.accent { background: #eef2fb; border-color: #c5d1e9;
    color: #223a6b; font-weight: 600; }
  .key.active-mode { background: #3b5bf6; color: #fff;
    border-color: #3b5bf6; }
  .key.small { font-size: 12px; }
  .key.tiny { font-size: 10px; }

  /* --- Pure CSS Expandable Math Elements --- */
  .math-frac { display: inline-flex; flex-direction: column; align-items: center; vertical-align: middle; line-height: 1.1; padding: 0 2px; }
  .math-frac .num { min-width: 16px; padding: 1px 4px; text-align: center; border-bottom: 2px solid #0b1f44; }
  .math-frac .den { min-width: 16px; padding: 1px 4px; text-align: center; }

  .math-exp { display: inline-flex; align-items: flex-start; vertical-align: middle; line-height: 1; }
  .math-exp .base { min-width: 14px; }
  .math-exp .power { font-size: 11px; position: relative; top: -8px; min-width: 12px; padding-left: 2px; }

  .math-sqrt { display: inline-flex; align-items: center; vertical-align: middle; line-height: 1; }
  .math-sqrt .radical { font-size: 22px; font-weight: bold; margin-right: -2px; }
  .math-sqrt .radicand { border-top: 2px solid #0b1f44; border-right: 2px solid #0b1f44; border-bottom: 2px solid #0b1f44; min-width: 18px; min-height: 22px; padding: 0 4px; font-size: 14px; display: flex; align-items: center; }

  .math-sum { display: inline-flex; flex-direction: column; align-items: center; vertical-align: middle; line-height: 1; font-size: 14px; padding: 0 2px; }
  .math-sum .bound { font-size: 10px; min-width: 22px; text-align: center; border-bottom: 1px solid #dfe5ef; padding: 1px 0; }
  .math-sum .sigma { font-size: 20px; font-weight: bold; line-height: 1; margin: 1px 0; }

  .math-log { display: inline-flex; align-items: flex-end; vertical-align: middle; }
  .math-log sub { font-size: 10px; border-bottom: 1px solid #dfe5ef; min-width: 12px; text-align: center; }

  .basegrid { padding: 8px 10px; background: #f6f8fd;
    border-top: 1px solid #e4e9f2;
    display: flex; flex-direction: column; gap: 6px; }
  .basegrid.collapsed { display: none; }

  .action { padding: 10px 3px; border-radius: 8px;
    background: #eef2fb; border: 1px solid #dfe5ef;
    color: #223a6b; font-size: 11px; font-weight: 600;
    cursor: pointer; font-family: inherit; text-align: center;
    min-height: 48px; display: flex; align-items: center;
    justify-content: center; }
  .action.primary { background: #3b5bf6; color: #fff;
    border-color: #3b5bf6; }
  .action:active { transform: scale(.97); }

  .result { padding: 10px 16px; background: #0b1f44; color: #fff;
    font-family: 'SF Mono', Menlo, monospace; font-size: 10px;
    line-height: 1.5; max-height: 0; overflow: hidden;
    transition: max-height .25s ease; }
  .result.shown { max-height: 260px; overflow-y: auto; }
  .result .label { color: #9ab; margin-bottom: 4px; }
  .result pre { margin: 0; color: #cfe2ff; white-space: pre-wrap; }
</style>
</head>
<body>

<div class="phone">
  <div class="topbar">
    <div class="logo">M</div>
    <div>
      <div class="brand">MethodMarker</div>
      <div class="tag">Better Methods. Higher Marks.</div>
    </div>
  </div>

  <div class="working">
    <div class="head">
      <span>&#9998; Working</span>
      <span class="steps" id="stepCount">0 steps</span>
    </div>
    <div class="lines">
      <div class="nums" id="nums"></div>
      <div class="text" id="text" contenteditable="true" spellcheck="false"></div>
    </div>
  </div>

  <div class="tabs" id="tabs"></div>

  <div class="grid" id="grid"></div>

  <div class="basegrid" id="basegrid"></div>

  <div class="result" id="result"></div>
</div>

<script>
const LAYOUT = __LAYOUT__;
const state = { activeTab: LAYOUT.tabs_order[0], collapsed: false, abcMode: false };

const textEl = document.getElementById("text");
const numsEl = document.getElementById("nums");
const stepCountEl = document.getElementById("stepCount");

function refreshNumbers() {
  const lines = textEl.innerText.split("\n");
  numsEl.innerHTML = lines.map((_, i) => `<div>${i + 1}</div>`).join("");
  const n = lines.length;
  stepCountEl.textContent = (n === 1 && lines[0] === "") ? "0 steps" : `${n} step${n === 1 ? "" : "s"}`;
}
textEl.addEventListener("input", refreshNumbers);
refreshNumbers();

function insertAtCaret(htmlStr) {
  textEl.focus();
  const sel = window.getSelection();
  if (!sel.rangeCount) return;
  const r = sel.getRangeAt(0);
  r.deleteContents();
  
  const temp = document.createElement('div');
  temp.innerHTML = htmlStr;
  
  while (temp.firstChild) {
    r.insertNode(temp.firstChild);
  }
  
  // Find the first contenteditable element inside the inserted HTML
  const firstEditable = temp.querySelector('[contenteditable="true"]');
  if (firstEditable) {
    const nr = document.createRange();
    nr.selectNodeContents(firstEditable);
    nr.collapse(true);
    sel.removeAllRanges();
    sel.addRange(nr);
  } else {
    r.collapse(false);
    sel.removeAllRanges();
    sel.addRange(r);
  }
  refreshNumbers();
}

function smartInsert(key) {
  if (key.e === "__back__") {
    textEl.focus();
    document.execCommand("delete");
    refreshNumbers();
    return;
  }
  if (key.e === "__newline__") { insertAtCaret("<br>"); return; }
  if (key.e === "__done__")    { submit(); return; }
  if (key.e === "__abc__") {
    state.abcMode = !state.abcMode;
    renderBase();
    return;
  }
  if (key.e === "__collapse__") {
    state.collapsed = !state.collapsed;
    document.getElementById("grid").classList.toggle("collapsed", state.collapsed);
    document.getElementById("basegrid").classList.toggle("collapsed", state.collapsed);
    return;
  }
  
  if (key.h) {
    insertAtCaret(key.h);
  } else {
    insertAtCaret(key.d);
  }
}

// ---------- tabs ----------
const tabsEl = document.getElementById("tabs");
LAYOUT.tabs_order.forEach(name => {
  const b = document.createElement("button");
  b.className = "tab" + (name === state.activeTab ? " active" : "");
  b.textContent = name;
  b.onclick = () => {
    state.activeTab = name;
    document.querySelectorAll(".tab").forEach(t =>
      t.classList.toggle("active", t.textContent === name));
    renderGrid();
  };
  tabsEl.appendChild(b);
});

// ---------- grid ----------
const gridEl = document.getElementById("grid");
function renderGrid() {
  const rows = LAYOUT.tabs[state.activeTab] || [];
  gridEl.innerHTML = "";
  const cols = Math.max(...rows.map(r => r.length));
  rows.forEach(row => {
    const r = document.createElement("div");
    r.className = "row";
    r.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
    row.forEach(key => {
      const k = document.createElement("div");
      const accent = ["or","=","⇒","∴","∵"].includes(key.l) ? " accent" : "";
      const small = key.l.length > 4 ? " small" : "";
      const tiny = key.l.length > 8 ? " tiny" : "";
      k.className = "key" + accent + small + tiny;
      if (key.h) {
        k.innerHTML = key.h;
      } else {
        k.textContent = key.l;
      }
      k.onclick = () => smartInsert(key);
      r.appendChild(k);
    });
    for (let i = row.length; i < cols; i++)
      r.appendChild(document.createElement("div"));
    gridEl.appendChild(r);
  });
}

// ---------- base ----------
function renderBase() {
  const b = LAYOUT.base_layer;
  const container = document.getElementById("basegrid");
  container.innerHTML = "";
  
  let layout = [];
  if (state.abcMode) {
    layout = [{ keys: b.abc, cols: 10 }];
  } else {
    layout = [
      { keys: b.variables, cols: 12 },
      { keys: b.digits,    cols: 12 },
      { keys: b.operators, cols: 10 },
      { keys: b.actions,   cols: 6  },
    ];
  }
  
  layout.forEach(({keys, cols}) => {
    const row = document.createElement("div");
    row.className = "row";
    row.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
    keys.forEach(key => {
      const isDone = key.e === "__done__";
      const isAction = key.e && key.e.startsWith("__");
      const el = document.createElement("div");
      el.className = isDone ? "action primary"
                           : (isAction ? "action" : "key");
      if (key.e === "__abc__" && state.abcMode) el.classList.add("active-mode");
      if (key.h) el.innerHTML = key.h;
      else       el.textContent = key.l;
      el.onclick = () => smartInsert(key);
      row.appendChild(el);
    });
    container.appendChild(row);
  });
}

// ---------- submit ----------
const resultEl = document.getElementById("result");

function submit() {
  const rawText = textEl.innerText;
  const payload = {
    learner_input: rawText,
    tab_used: state.activeTab
  };
  resultEl.classList.add("shown");
  resultEl.innerHTML =
    "<div class='label'>submitted to step_grader</div>" +
    "<pre>" + JSON.stringify(payload, null, 2) + "</pre>";
}

textEl.focus();
renderGrid();
renderBase();
</script>
</body>
</html>
"""

HTML = HTML_TEMPLATE.replace("__LAYOUT__", json.dumps(kb, ensure_ascii=False))
HTML_PATH.write_text(HTML, encoding="utf-8")
print()
print(f"  ✓ {HTML_PATH.name} rebuilt  ({HTML_PATH.stat().st_size:,} bytes)")

webbrowser.open(HTML_PATH.as_uri())
print(f"  → opened in browser")
print()
print("=" * 72)
print("TEST IN BROWSER (v6.4)")
print("=" * 72)
print("  Base operators row:")
print("    Tap 'a/b' → inserts a real fraction. Click top box, type. Click bottom, type.")
print("    Tap '√'   → inserts a real radical. Click the box, type.")
print("    Tap 'log', 'log_a', 'ln' → permanently on the base layer.")
print()
print("  Algebra tab:")
print("    Tap 'Σ' → inserts a summation. Click the TOP bound and type 'p=1'.")
print("    Click the BOTTOM bound and type 'k=0'. It is fully editable.")
````

````python
# Keyboard v7.0 — Clean MathQuill Integration: Expandable math, editable bounds
import json
import webbrowser
from pathlib import Path

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
LAYOUT_PATH = ROOT / "data" / "processed" / "tutor" / "keyboard_layout.json"
HTML_PATH = ROOT / "docs" / "keyboard_preview.html"

def K(label, engine, html=None):
    d = {"l": label, "e": engine}
    if html: d["h"] = html
    return d

# ============================================================
# Load existing layout
# ============================================================
kb = json.loads(LAYOUT_PATH.read_text(encoding="utf-8"))

# --- Rebuild Base Operators (Clean MathQuill commands) ---
kb["base_layer"]["operators"] = [
    K("+", "+"), K("-", "-"), K("×", "\\times"), K("÷", "\\div"),
    K("(", "("), K(")", ")"),
    K("a/b", "\\frac{}{}"),
    K("x^n", "^{}"),
    K("√", "\\sqrt{}"),
    K("log", "\\log_{10}("),
    K("log_a", "\\log_{}("),
    K("ln", "\\ln("),
    K("x²", "^2"), K("π", "\\pi"), K("θ", "\\theta"),
]

# --- Tab 1: Number, Sequences & Finance (LO 1) ---
kb["tabs"]["Sequences"] = [
    [K("Σ", "\\sum_{n=1}^{\\infty}"), K("Tₙ", "T_n"), K("Sₙ", "S_n"), K("n→∞", "n \\to \\infty")],
    [K("A", "A"), K("P", "P"), K("i", "i"), K("n", "n")],
    [K("log", "\\log_{10}("), K("log_a", "\\log_{}("), K("x⁻¹", "^{-1}")],
    [K("1", "1"), K("2", "2"), K("3", "3")]
]

# --- Tab 2: Functions & Algebra (LO 2) ---
kb["tabs"]["Algebra"] = [
    [K("=", "="), K("or", "\\text{or}"), K("⇒", "\\Rightarrow"), K("∴", "\\therefore"), K("∵", "\\because")],
    [K("x²", "^2"), K("y²", "^2"), K("a²", "^2"), K("b²", "^2"), K("c²", "^2")],
    [K("x³", "^3"), K("xⁿ", "^{n}"), K("aⁿ", "^{n}"), K("f(x)", "f(x)")],
    [K("√", "\\sqrt{}"), K("∛", "\\sqrt[3]{}"), K("ⁿ√", "\\sqrt[n]{}"), K("±", "\\pm")],
    [K("<", "<"), K(">", ">"), K("≤", "\\le"), K("≥", "\\ge"), K("≠", "\\ne")],
    [K("π", "\\pi"), K("∞", "\\infty"), K("QF", "x = \\frac{-b \\pm \\sqrt{b^2 - 4ac}}{2a}")]
]

# --- Tab 3: Geometry & Trigonometry (LO 3) ---
kb["tabs"]["Trig"] = [
    [K("sin", "\\sin("), K("cos", "\\cos("), K("tan", "\\tan(")],
    [K("sin⁻¹", "\\sin^{-1}("), K("cos⁻¹", "\\cos^{-1}("), K("tan⁻¹", "\\tan^{-1}(")],
    [K("sin²θ", "\\sin^2\\theta"), K("cos²θ", "\\cos^2\\theta"), K("tan²θ", "\\tan^2\\theta")],
    [K("△", "\\triangle"), K("∥", "\\parallel"), K("⊥", "\\perp"), K("∠", "\\angle"), K("≅", "\\cong"), K("∼", "\\sim")],
    [K("Â", "\\hat{A}"), K("B̂", "\\hat{B}"), K("Ĉ", "\\hat{C}"), K("D̂", "\\hat{D}")],
    [K("(x₁;y₁)", "(x_1;y_1)"), K("(x₂;y₂)", "(x_2;y_2)"), K("d", "d = \\sqrt{(x_2-x_1)^2 + (y_2-y_1)^2}")]
]

# --- Tab 4: Data Handling & Probability (LO 4) ---
kb["tabs"]["Stats"] = [
    [K("x̄", "\\bar{x}"), K("σ", "\\sigma"), K("n", "n"), K("r", "r"), K("r²", "r^2")],
    [K("P(A)", "P(A)"), K("P(B)", "P(B)"), K("P(A∪B)", "P(A \\cup B)"), K("P(A∩B)", "P(A \\cap B)"), K("P(A')", "P(A')")],
    [K("nCr", "\\binom{n}{r}"), K("nPr", "\\frac{n!}{(n-r)!}"), K("!", "!")],
    [K("Σx", "\\sum x"), K("Σx²", "\\sum x^2"), K("E(x)", "E(x)")]
]

kb["schema_version"] = 7.0
LAYOUT_PATH.write_text(json.dumps(kb, indent=2, ensure_ascii=False), encoding="utf-8")
print("  ✓ keyboard_layout.json v7.0")

# ============================================================
# Rebuild HTML with MathQuill (Clean Integration)
# ============================================================
HTML_TEMPLATE = r"""<!doctype html>
<html lang="en">
<head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width,initial-scale=1">
<title>MethodKeyboard — v7.0</title>
<link rel="stylesheet" href="https://cdnjs.cloudflare.com/ajax/libs/mathquill/0.10.1/mathquill.min.css">
<script src="https://code.jquery.com/jquery-3.6.0.min.js"></script>
<script src="https://cdnjs.cloudflare.com/ajax/libs/mathquill/0.10.1/mathquill.min.js"></script>
<style>
  * { box-sizing: border-box; -webkit-tap-highlight-color: transparent; }
  body { margin: 0; background: #eef2f8;
    font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif;
    display: flex; justify-content: center; padding: 12px; min-height: 100vh; }
  .phone { width: 600px; max-width: 100%; background: #fff;
    border-radius: 20px; overflow: hidden;
    box-shadow: 0 8px 32px rgba(20, 40, 90, .12);
    display: flex; flex-direction: column; }
  .topbar { display: flex; align-items: center; gap: 10px;
    padding: 12px 16px; background: #f6f8fd;
    border-bottom: 1px solid #e4e9f2; }
  .topbar .logo { width: 28px; height: 28px; border-radius: 7px;
    background: linear-gradient(135deg, #3b5bf6, #5b8cf6);
    display: flex; align-items: center; justify-content: center;
    color: #fff; font-weight: 700; font-size: 15px; }
  .topbar .brand { font-weight: 700; font-size: 14px; color: #0b1f44; }
  .topbar .tag { font-size: 10px; color: #6b7a99; }

  .working { padding: 12px 16px; background: #fff; min-height: 120px; }
  .working .head { display: flex; justify-content: space-between;
    align-items: center; font-size: 12px; color: #465a80;
    font-weight: 600; margin-bottom: 6px; }
  .working .steps { color: #6b7a99; font-weight: 500; }
  #math-field { 
    min-height: 40px; font-size: 20px; outline: none; 
    background: #f9fafc; border: 1px solid #dfe5ef; 
    border-radius: 8px; padding: 8px 12px; 
  }

  .tabs { display: flex; overflow-x: auto; gap: 4px;
    padding: 6px 12px; background: #f6f8fd;
    border-top: 1px solid #e4e9f2;
    border-bottom: 1px solid #e4e9f2; scrollbar-width: none; }
  .tabs::-webkit-scrollbar { display: none; }
  .tab { flex: none; padding: 6px 11px; border-radius: 16px;
    font-size: 11px; font-weight: 600; color: #465a80;
    background: transparent; border: none; cursor: pointer;
    white-space: nowrap; }
  .tab.active { background: #3b5bf6; color: #fff; }

  .grid { padding: 8px 10px; background: #fff;
    display: flex; flex-direction: column; gap: 6px;
    overflow: hidden; }
  .row { display: grid; gap: 6px; }
  .key { padding: 9px 3px; border-radius: 8px;
    background: #fff; border: 1px solid #dfe5ef;
    font-size: 14px; color: #0b1f44; cursor: pointer;
    text-align: center; user-select: none; font-family: inherit;
    transition: transform .05s, background .1s;
    min-height: 48px; display: flex; align-items: center;
    justify-content: center; }
  .key:active { transform: scale(.95); background: #eef2fb; }
  .key.accent { background: #eef2fb; border-color: #c5d1e9;
    color: #223a6b; font-weight: 600; }
  .key.small { font-size: 12px; }
  .key.tiny { font-size: 10px; }

  .basegrid { padding: 8px 10px; background: #f6f8fd;
    border-top: 1px solid #e4e9f2;
    display: flex; flex-direction: column; gap: 6px; }

  .action { padding: 10px 3px; border-radius: 8px;
    background: #eef2fb; border: 1px solid #dfe5ef;
    color: #223a6b; font-size: 11px; font-weight: 600;
    cursor: pointer; font-family: inherit; text-align: center;
    min-height: 48px; display: flex; align-items: center;
    justify-content: center; }
  .action.primary { background: #3b5bf6; color: #fff; border-color: #3b5bf6; }
  .action:active { transform: scale(.97); }

  .result { padding: 10px 16px; background: #0b1f44; color: #fff;
    font-family: 'SF Mono', Menlo, monospace; font-size: 10px;
    line-height: 1.5; max-height: 0; overflow: hidden; transition: max-height .25s ease; }
  .result.shown { max-height: 260px; overflow-y: auto; }
  .result .label { color: #9ab; margin-bottom: 4px; }
  .result pre { margin: 0; color: #cfe2ff; white-space: pre-wrap; }
</style>
</head>
<body>

<div class="phone">
  <div class="topbar">
    <div class="logo">M</div>
    <div>
      <div class="brand">MethodMarker</div>
      <div class="tag">Better Methods. Higher Marks.</div>
    </div>
  </div>

  <div class="working">
    <div class="head">
      <span>&#9998; Working</span>
      <span class="steps" id="stepCount">0 steps</span>
    </div>
    <span id="math-field"></span>
  </div>

  <div class="tabs" id="tabs"></div>
  <div class="grid" id="grid"></div>
  <div class="basegrid" id="basegrid"></div>
  <div class="result" id="result"></div>
</div>

<script>
const LAYOUT = __LAYOUT__;
const state = { activeTab: LAYOUT.tabs_order[0] };

// Initialize MathQuill
const MQ = MathQuill.getInterface(2);
const mathField = MQ.MathField(document.getElementById('math-field'), {
  handlers: {
    edit: function() {
      const latex = mathField.latex();
      const steps = latex.split('\\\\').length;
      document.getElementById('stepCount').textContent = `${steps} step${steps === 1 ? "" : "s"}`;
    }
  }
});

function smartInsert(key) {
  if (key.e === "__done__") { submit(); return; }
  if (key.e.startsWith('\\')) {
    mathField.cmd(key.e);
  } else {
    mathField.write(key.e);
  }
}

// ---------- tabs ----------
const tabsEl = document.getElementById("tabs");
LAYOUT.tabs_order.forEach(name => {
  const b = document.createElement("button");
  b.className = "tab" + (name === state.activeTab ? " active" : "");
  b.textContent = name;
  b.onclick = () => {
    state.activeTab = name;
    document.querySelectorAll(".tab").forEach(t =>
      t.classList.toggle("active", t.textContent === name));
    renderGrid();
  };
  tabsEl.appendChild(b);
});

// ---------- grid ----------
const gridEl = document.getElementById("grid");
function renderGrid() {
  const rows = LAYOUT.tabs[state.activeTab] || [];
  gridEl.innerHTML = "";
  const cols = Math.max(...rows.map(r => r.length));
  rows.forEach(row => {
    const r = document.createElement("div");
    r.className = "row";
    r.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
    row.forEach(key => {
      const k = document.createElement("div");
      const accent = ["or","=","⇒","∴","∵"].includes(key.l) ? " accent" : "";
      const small = key.l.length > 4 ? " small" : "";
      const tiny = key.l.length > 8 ? " tiny" : "";
      k.className = "key" + accent + small + tiny;
      k.textContent = key.l;
      k.onclick = () => smartInsert(key);
      r.appendChild(k);
    });
    gridEl.appendChild(r);
  });
}

// ---------- base ----------
function renderBase() {
  const b = LAYOUT.base_layer;
  const container = document.getElementById("basegrid");
  container.innerHTML = "";
  const layout = [
    { keys: b.variables, cols: 12 },
    { keys: b.digits,    cols: 12 },
    { keys: b.operators, cols: 10 },
    { keys: b.actions,   cols: 6  },
  ];
  layout.forEach(({keys, cols}) => {
    const row = document.createElement("div");
    row.className = "row";
    row.style.gridTemplateColumns = `repeat(${cols}, 1fr)`;
    keys.forEach(key => {
      const isDone = key.e === "__done__";
      const isAction = key.e && key.e.startsWith("__");
      const el = document.createElement("div");
      el.className = isDone ? "action primary" : (isAction ? "action" : "key");
      el.textContent = key.l;
      el.onclick = () => smartInsert(key);
      row.appendChild(el);
    });
    container.appendChild(row);
  });
}

// ---------- submit ----------
const resultEl = document.getElementById("result");
function submit() {
  const latexStr = mathField.latex();
  const payload = {
    learner_latex: latexStr,
    tab_used: state.activeTab
  };
  resultEl.classList.add("shown");
  resultEl.innerHTML = "<div class='label'>submitted to step_grader</div>" +
    "<pre>" + JSON.stringify(payload, null, 2) + "</pre>";
}

mathField.focus();
renderGrid();
renderBase();
</script>
</body>
</html>
"""

HTML = HTML_TEMPLATE.replace("__LAYOUT__", json.dumps(kb, ensure_ascii=False))
HTML_PATH.write_text(HTML, encoding="utf-8")
print(f"  ✓ {HTML_PATH.name} rebuilt  ({HTML_PATH.stat().st_size:,} bytes)")

webbrowser.open(HTML_PATH.as_uri())
print("  → opened in browser")
print("=" * 72)
print("TEST IN BROWSER (v7.0)")
print("=" * 72)
print("  - Tap 'a/b' -> inserts a real expanding fraction.")
print("  - Tap '√'   -> inserts a real radical that stretches.")
print("  - Tap 'Σ'   -> inserts a summation. Click 'n=1' and type 'p=0'.")
print("  - Tap 'log', 'log_a', 'ln' -> permanently on the base layer.")
````

````python
# Keyboard v9.0 — Lines-based notebook editor, visual keys, SBC
import json, webbrowser
from pathlib import Path

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
LAYOUT_PATH = ROOT / "data" / "processed" / "tutor" / "keyboard_layout.json"
HTML_PATH   = ROOT / "docs" / "keyboard_preview.html"

kb = json.loads(LAYOUT_PATH.read_text(encoding="utf-8"))

# ────────────────────────────────────────────────────────────
# Key definitions:  k  = label shown on the button
#                   i  = HTML inserted into the working area
#                   (i is *also* shown on the button when key has no icon)
# ────────────────────────────────────────────────────────────
def key(label, ins=None, icon=None):
    d = {"k": label}
    d["i"] = ins if ins is not None else label
    if icon: d["ic"] = icon
    return d

# Visual icons for the buttons (tiny Casio-style previews)
ICON_FRAC  = "<span class='ic-frac'><span></span><span></span><span></span></span>"
ICON_EXP   = "<span class='ic-exp'><span></span><span></span></span>"
ICON_SQRT  = "<span class='ic-sqrt'>√<span></span></span>"
ICON_SUM   = "<span class='ic-sum'><span>∞</span>Σ<span>n=1</span></span>"

# The actual editable structures inserted into the working area
INS_FRAC = "<span class='mfrac'><span class='mnum cur'></span><span class='mden'></span></span>"
INS_EXP  = "<span class='mexp'><span class='mbase cur'></span><span class='mpow'></span></span>"
INS_SQRT = "<span class='msqrt'>√<span class='mrad cur'></span></span>"
INS_CBRT = "<span class='msqrt'>∛<span class='mrad cur'></span></span>"
INS_NRT  = "<span class='msqrt'>ⁿ√<span class='mrad cur'></span></span>"
INS_SUM  = "<span class='msum'><span class='mtop cur'>∞</span><span class='msig'>Σ</span><span class='mbot'>n=1</span></span>"
INS_LOG  = "<span class='mlog'>log<sub class='mlb'></sub>(</span>"
INS_LNA  = "<span class='mlog'>ln(</span>"

kb["base_layer"] = {
    "variables":[key(v) for v in "xyzabcnkmpqt"],
    "digits":   [key(d) for d in "1234567890"] + [key("."), key("(-)","-")],
    "operators":[
        key("+"), key("−","-"), key("×","*"), key("÷","/"),
        key("="), key("("), key(")"),
        key("a/b", INS_FRAC, ICON_FRAC),
        key("xⁿ",  INS_EXP,  ICON_EXP),
        key("√",   INS_SQRT, ICON_SQRT),
        key("log", INS_LOG),
        key("ln",  INS_LNA),
    ],
    "actions":[
        key("DEL","__del__"), key("AC","__clear__"),
        key("⏎","__enter__"), key("?","__ask__"),
        key("✓","__done__"),
    ],
}

kb["tabs_order"] = ["Sequences","Algebra","Functions","Calculus",
                    "Trig","Finance","Analytical","Euclidean","Stats","Probability"]

kb["tabs"] = {
 "Sequences":[
   [key("Σ", INS_SUM, ICON_SUM), key("Tₙ","T_n"), key("Sₙ","S_n"), key("n→∞","n->∞")],
   [key("A"),key("P"),key("i"),key("n")],
 ],
 "Algebra":[
   [key("="),key("or"," or "),key("⇒","=>"),key("∴","∴"),key("∵","∵")],
   [key("x²","x^2"),key("y²","y^2"),key("a²","a^2"),key("b²","b^2"),key("c²","c^2")],
   [key("x³","x^3"),key("xⁿ","x^n"),key("f(x)"),key("±","±"),key("x⁻¹","x^-1")],
   [key("√",INS_SQRT,ICON_SQRT), key("∛",INS_CBRT), key("ⁿ√",INS_NRT), key("QF","x = (-b ± √(b²-4ac))/(2a)")],
   [key("<"),key(">"),key("≤","<="),key("≥",">="),key("≠","!=")],
   [key("π","π"),key("∞","∞"),key("n"),key("R","R")],
 ],
 "Functions":[
   [key("f(x)"),key("→","->"),key("∞","∞"),key("π","π")],
   [key("dy/dx","dy/dx"),key("d/dx","d/dx"),key("∫","∫"),key("lim","lim")],
 ],
 "Calculus":[
   [key("dy/dx","dy/dx"),key("d²y/dx²","d²y/dx²"),key("∫","∫"),key("lim","lim")],
   [key("→","->"),key("∞","∞"),key("Δx","Δx")],
 ],
 "Trig":[
   [key("sin","sin("),key("cos","cos("),key("tan","tan(")],
   [key("sin⁻¹","asin("),key("cos⁻¹","acos("),key("tan⁻¹","atan(")],
   [key("sin²θ","sin²θ"),key("cos²θ","cos²θ"),key("tan²θ","tan²θ"),key("θ","θ")],
   [key("°","°"),key("π","π")],
 ],
 "Finance":[
   [key("A"),key("P"),key("i"),key("n")],
   [key("(1+i)ⁿ","(1+i)^n"),key("(1-i)ⁿ","(1-i)^n")],
 ],
 "Analytical":[
   [key("(x₁;y₁)","(x1;y1)"),key("(x₂;y₂)","(x2;y2)")],
   [key("d","d = √((x2-x1)² + (y2-y1)²)"),
    key("M","M = ((x1+x2)/2 ; (y1+y2)/2)"),
    key("m","m = (y2-y1)/(x2-x1)")],
 ],
 "Euclidean":[
   [key("△","△"),key("∥","∥"),key("⊥","⊥")],
   [key("∠","∠"),key("≅","≅"),key("∼","∼")],
   [key("Â","Â"),key("B̂","B̂"),key("Ĉ","Ĉ"),key("D̂","D̂")],
   [key("∴","∴"),key("∵","∵")],
 ],
 "Stats":[
   [key("x̄","x̄"),key("σ","σ"),key("n"),key("r"),key("r²","r²")],
   [key("Σx","Σx"),key("Σx²","Σx²"),key("E(x)","E(x)")],
 ],
 "Probability":[
   [key("P(A)"),key("P(B)"),key("P(A∪B)","P(A∪B)"),
    key("P(A∩B)","P(A∩B)"),key("P(A')","P(A')")],
   [key("nCr","nCr"),key("nPr","nPr"),key("!","!")],
 ],
}
kb["schema_version"] = 9.0
LAYOUT_PATH.write_text(json.dumps(kb,indent=2,ensure_ascii=False),encoding="utf-8")
print("  ✓ keyboard_layout.json v9.0")

# ────────────────────────────────────────────────────────────
# HTML
# ────────────────────────────────────────────────────────────
HTML = r"""<!doctype html>
<html lang="en"><head>
<meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1">
<title>MethodKeyboard v9.0</title>
<style>
*{box-sizing:border-box;-webkit-tap-highlight-color:transparent}
body{margin:0;background:#eef2f8;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,sans-serif;
 display:flex;justify-content:center;padding:12px;min-height:100vh}
.phone{width:640px;max-width:100%;background:#fff;border-radius:20px;overflow:hidden;
 box-shadow:0 8px 32px rgba(20,40,90,.12);display:flex;flex-direction:column}

/* top bar */
.topbar{display:flex;align-items:center;gap:10px;padding:12px 16px;background:#f6f8fd;border-bottom:1px solid #e4e9f2}
.logo{width:28px;height:28px;border-radius:7px;background:linear-gradient(135deg,#3b5bf6,#5b8cf6);
 display:flex;align-items:center;justify-content:center;color:#fff;font-weight:700;font-size:15px}
.brand{font-weight:700;font-size:14px;color:#0b1f44}
.tag{font-size:10px;color:#6b7a99}

/* SURFACE 1: working notebook */
.work{background:#fff;padding:12px 16px 16px;max-height:280px;overflow-y:auto}
.work-head{display:flex;justify-content:space-between;align-items:center;
 font-size:12px;color:#465a80;font-weight:600;margin-bottom:8px}
.work-head .hint{color:#9aa8c2;font-weight:500;font-size:11px}
.lines{display:flex;flex-direction:column;gap:2px}
.lrow{display:grid;grid-template-columns:28px 1fr;align-items:flex-start;column-gap:4px}
.lnum{color:#b8c2d6;font-size:11px;text-align:right;padding-top:8px;user-select:none;font-variant-numeric:tabular-nums}
.lbody{min-height:32px;padding:6px 8px;border-radius:6px;font-size:18px;line-height:1.35;
 color:#0b1f44;outline:none;font-family:'Cambria Math','Latin Modern Math',Georgia,serif;
 transition:background .1s}
.lbody:focus{background:#f5f8ff}
.lrow:last-child .lbody{background:#fbfcfe}

/* SURFACE 2: SBC */
.sbc{padding:10px 16px 12px;background:#fff;border-top:1px dashed #e4e9f2}
.sbc-head{display:flex;align-items:center;gap:6px;font-size:12px;color:#465a80;
 font-weight:600;margin-bottom:6px}
.sbc-head .dot{width:6px;height:6px;border-radius:50%;background:#f0ad4e}
#sbc{width:100%;min-height:44px;resize:vertical;font-family:inherit;font-size:13px;
 padding:8px 12px;border:1px solid #dfe5ef;border-radius:8px;background:#fffdf7;
 color:#0b1f44;outline:none}
#sbc:focus{border-color:#f0ad4e}

/* tabs */
.tabs{display:flex;overflow-x:auto;gap:4px;padding:6px 12px;background:#f6f8fd;
 border-top:1px solid #e4e9f2;border-bottom:1px solid #e4e9f2;scrollbar-width:none}
.tabs::-webkit-scrollbar{display:none}
.tab{flex:none;padding:6px 11px;border-radius:16px;font-size:11px;font-weight:600;
 color:#465a80;background:transparent;border:none;cursor:pointer;white-space:nowrap}
.tab.active{background:#3b5bf6;color:#fff}

/* keyboard */
.kb{padding:8px 10px;background:#fff;display:flex;flex-direction:column;gap:5px}
.row{display:grid;gap:5px}
.key{padding:6px 3px;border-radius:8px;background:#fff;border:1px solid #dfe5ef;
 font-size:14px;color:#0b1f44;cursor:pointer;text-align:center;user-select:none;
 font-family:inherit;min-height:46px;display:flex;align-items:center;justify-content:center}
.key:active{transform:scale(.95);background:#eef2fb}
.key.base{background:#f6f8fd}
.actions{display:grid;grid-template-columns:repeat(5,1fr);gap:5px;padding:8px 10px;
 background:#f6f8fd;border-top:1px solid #e4e9f2}
.act{padding:10px 3px;border-radius:8px;background:#eef2fb;border:1px solid #dfe5ef;
 color:#223a6b;font-size:11px;font-weight:600;cursor:pointer;text-align:center;user-select:none;
 min-height:46px;display:flex;align-items:center;justify-content:center;font-family:inherit}
.act.done{background:#3b5bf6;color:#fff;border-color:#3b5bf6}
.act:active{transform:scale(.97)}

/* ---- button icons (Casio-style previews) ---- */
.ic-frac{display:inline-flex;flex-direction:column;align-items:center;gap:1px}
.ic-frac span:nth-child(1),.ic-frac span:nth-child(3){width:12px;height:6px;border:1px solid #7b8ba6;border-radius:2px}
.ic-frac span:nth-child(2){width:16px;height:1px;background:#7b8ba6}
.ic-exp{position:relative;width:22px;height:20px;display:inline-block}
.ic-exp span:nth-child(1){position:absolute;bottom:0;left:2px;width:12px;height:12px;border:1px solid #7b8ba6;border-radius:2px}
.ic-exp span:nth-child(2){position:absolute;top:0;right:0;width:8px;height:8px;border:1px solid #7b8ba6;border-radius:2px}
.ic-sqrt{font-size:16px;font-weight:bold;display:inline-flex;align-items:center;line-height:1}
.ic-sqrt span{width:9px;height:11px;border-top:1px solid #7b8ba6;border-right:1px solid #7b8ba6;border-bottom:1px solid #7b8ba6;margin-left:-1px}
.ic-sum{display:inline-flex;flex-direction:column;align-items:center;font-size:13px;line-height:1}
.ic-sum span:first-child,.ic-sum span:last-child{font-size:8px}

/* ---- editable math structures inside the working area ---- */
.mfrac{display:inline-flex;flex-direction:column;align-items:center;vertical-align:middle;
 line-height:1.05;padding:0 2px;margin:0 2px}
.mfrac .mnum{border-bottom:1.5px solid #0b1f44;padding:0 5px;min-width:12px;text-align:center}
.mfrac .mden{padding:0 5px;min-width:12px;text-align:center}

.mexp{display:inline-flex;align-items:flex-start;vertical-align:middle;padding:0 2px}
.mexp .mbase{min-width:10px}
.mexp .mpow{font-size:.62em;position:relative;top:-.55em;padding-left:1px;min-width:10px}

.msqrt{display:inline-flex;align-items:stretch;vertical-align:middle;line-height:1;padding:0 2px}
.msqrt>span.mrad{border-top:1.5px solid #0b1f44;border-right:1.5px solid #0b1f44;
 border-bottom:1.5px solid #0b1f44;padding:3px 6px;min-width:14px;min-height:22px;
 display:inline-flex;align-items:center}

.msum{display:inline-flex;flex-direction:column;align-items:center;vertical-align:middle;
 line-height:1;padding:0 4px;margin:0 2px}
.msum .mtop{font-size:.58em;min-width:24px;text-align:center;border-bottom:1px solid #cbd5e6}
.msum .msig{font-size:1.15em;line-height:1;margin:1px 0}
.msum .mbot{font-size:.58em;min-width:24px;text-align:center}

.mlog{display:inline-flex;align-items:flex-end;vertical-align:middle;padding:0 2px}
.mlog sub{font-size:.6em;min-width:12px;text-align:center}

/* result strip */
.result{padding:10px 16px;background:#0b1f44;color:#fff;font-family:'SF Mono',Menlo,monospace;
 font-size:10px;line-height:1.5;max-height:0;overflow:hidden;transition:max-height .25s ease}
.result.shown{max-height:280px;overflow-y:auto}
.result .label{color:#9ab;margin-bottom:4px}
.result pre{margin:0;color:#cfe2ff;white-space:pre-wrap}
</style></head><body>

<div class="phone">
  <div class="topbar">
    <div class="logo">M</div>
    <div><div class="brand">MethodMarker</div><div class="tag">Better Methods. Higher Marks.</div></div>
  </div>

  <!-- surface 1: lines -->
  <div class="work">
    <div class="work-head"><span>&#9998; Working</span><span class="hint" id="hint">tap a line to write. ENTER for a new step.</span></div>
    <div class="lines" id="lines"></div>
  </div>

  <!-- surface 2: sbc -->
  <div class="sbc">
    <div class="sbc-head"><span class="dot"></span> Ask Tanaka — where did you get stuck?</div>
    <textarea id="sbc" placeholder="e.g. I did step 2 but I don't understand why the sign flips"></textarea>
  </div>

  <div class="tabs" id="tabs"></div>
  <div class="kb" id="grid"></div>
  <div class="actions" id="actions"></div>
  <div class="result" id="result"></div>
</div>

<script>
const LAYOUT = __LAYOUT__;
const state = { tab: LAYOUT.tabs_order[0] };

/* ---------- line-based editor ---------- */
const linesEl = document.getElementById('lines');

function addLine(html, focus) {
  const row = document.createElement('div');
  row.className = 'lrow';
  const num = document.createElement('div');
  num.className = 'lnum';
  const body = document.createElement('div');
  body.className = 'lbody';
  body.contentEditable = 'true';
  body.spellcheck = false;
  if (html) body.innerHTML = html;
  row.appendChild(num); row.appendChild(body);
  linesEl.appendChild(row);
  renumber();
  if (focus) {
    placeCursorIn(body, false);
  }
  return body;
}
function renumber(){
  [...linesEl.querySelectorAll('.lrow')].forEach((r,i)=>{
    r.querySelector('.lnum').textContent = i+1;
  });
}
function placeCursorIn(el, atEnd) {
  el.focus();
  const r = document.createRange();
  r.selectNodeContents(el);
  r.collapse(!atEnd);
  const sel = window.getSelection();
  sel.removeAllRanges(); sel.addRange(r);
}
function activeLine() {
  const sel = window.getSelection();
  if (!sel.rangeCount) return null;
  let n = sel.getRangeAt(0).startContainer;
  while (n && n !== linesEl) {
    if (n.classList && n.classList.contains('lbody')) return n;
    n = n.parentNode;
  }
  return null;
}

/* Backspace at start of a line → merge with previous */
/* Enter → new line below, split if cursor mid-line */
linesEl.addEventListener('keydown', e => {
  const line = e.target.closest('.lbody'); if (!line) return;

  if (e.key === 'Enter' && !e.shiftKey) {
    e.preventDefault();
    const sel = window.getSelection();
    const r = sel.getRangeAt(0);
    // split content at cursor
    const after = r.cloneRange();
    after.selectNodeContents(line);
    after.setStart(r.endContainer, r.endOffset);
    const rest = after.extractContents();
    const newBody = addLine('', true);
    // put rest content into new line
    while (rest.firstChild) newBody.insertBefore(rest.firstChild, newBody.firstChild);
    placeCursorIn(newBody, false);
    return;
  }

  if (e.key === 'Backspace') {
    const sel = window.getSelection();
    const r = sel.getRangeAt(0);
    if (!r.collapsed) return;
    // is the cursor at the very start of this line?
    const probe = r.cloneRange(); probe.selectNodeContents(line);
    probe.setEnd(r.startContainer, r.startOffset);
    if (probe.toString().length !== 0) return;
    if (line.textContent.length === 0 || probe.toString().length === 0) {
      const prev = line.closest('.lrow').previousElementSibling;
      if (prev) {
        e.preventDefault();
        const prevBody = prev.querySelector('.lbody');
        // move contents of this line into prev line
        while (line.firstChild) prevBody.appendChild(line.firstChild);
        line.closest('.lrow').remove();
        placeCursorIn(prevBody, true);
        renumber();
      } else if (line.textContent.length === 0) {
        e.preventDefault();
      }
    }
    return;
  }

  /* Up/Down at boundary → move between lines */
  if (e.key === 'ArrowUp') {
    const sel = window.getSelection();
    const r = sel.getRangeAt(0);
    const probe = r.cloneRange(); probe.selectNodeContents(line);
    probe.setEnd(r.startContainer, r.startOffset);
    if (probe.toString().length === 0) {
      const prev = line.closest('.lrow').previousElementSibling;
      if (prev) { e.preventDefault(); placeCursorIn(prev.querySelector('.lbody'), true); }
    }
  }
  if (e.key === 'ArrowDown') {
    const sel = window.getSelection();
    const r = sel.getRangeAt(0);
    const probe = r.cloneRange(); probe.selectNodeContents(line);
    probe.setStart(r.endContainer, r.endOffset);
    if (probe.toString().length === 0) {
      const next = line.closest('.lrow').nextElementSibling;
      if (next) { e.preventDefault(); placeCursorIn(next.querySelector('.lbody'), false); }
    }
  }
});

/* ---------- insertion ---------- */
function insertAtCursor(html) {
  const line = activeLine();
  if (!line) {
    // no active line → create one and insert
    const b = addLine('', true);
    b.innerHTML = html;
    placeCursorInFirstField(b);
    return;
  }
  const sel = window.getSelection();
  if (!sel.rangeCount) return;
  const r = sel.getRangeAt(0);
  r.deleteContents();

  const tmp = document.createElement('div');
  tmp.innerHTML = html;
  const last = tmp.lastChild;
  while (tmp.firstChild) r.insertNode(tmp.firstChild);

  // focus first .cur field in the inserted content
  if (last && last.querySelector) {
    const cur = last.querySelector('.cur');
    if (cur) {
      const nr = document.createRange();
      nr.selectNodeContents(cur);
      nr.collapse(true);
      sel.removeAllRanges(); sel.addRange(nr);
      return;
    }
  }
  // else just put cursor at end of the insertion
  const nr = document.createRange();
  nr.setStartAfter(last);
  nr.collapse(true);
  sel.removeAllRanges(); sel.addRange(nr);
}
function placeCursorInFirstField(el) {
  const cur = el.querySelector('.cur');
  if (cur) { const r=document.createRange(); r.selectNodeContents(cur); r.collapse(true);
    const s=window.getSelection(); s.removeAllRanges(); s.addRange(r); }
}

/* ---------- key handlers ---------- */
function handleKey(key) {
  if (key.i === '__del__')   { document.execCommand('delete'); return; }
  if (key.i === '__clear__') { linesEl.innerHTML=''; addLine('', true); return; }
  if (key.i === '__enter__') { const line=activeLine()||addLine('',false);
                               const nb=addLine('', true); return; }
  if (key.i === '__ask__')   { document.getElementById('sbc').focus(); return; }
  if (key.i === '__done__')  { submit(); return; }
  insertAtCursor(key.i);
}

/* ---------- tabs ---------- */
const tabsEl = document.getElementById('tabs');
LAYOUT.tabs_order.forEach(name => {
  const b = document.createElement('button');
  b.className = 'tab' + (name === state.tab ? ' active' : '');
  b.textContent = name;
  b.onclick = () => {
    state.tab = name;
    document.querySelectorAll('.tab').forEach(t =>
      t.classList.toggle('active', t.textContent === name));
    renderGrid();
  };
  tabsEl.appendChild(b);
});

/* ---------- key grid + base row ---------- */
const gridEl = document.getElementById('grid');
function makeKeyBtn(key) {
  const el = document.createElement('div');
  el.className = 'key';
  if (key.ic) el.innerHTML = key.ic;
  else el.textContent = key.k;
  el.onclick = () => handleKey(key);
  return el;
}
function renderGrid() {
  gridEl.innerHTML = '';
  const rows = LAYOUT.tabs[state.tab] || [];
  const cols = Math.max(...rows.map(r => r.length));
  // base layer first
  const baseLayout = [
    LAYOUT.base_layer.variables,
    LAYOUT.base_layer.digits,
    LAYOUT.base_layer.operators,
  ];
  baseLayout.forEach(keys => {
    const r = document.createElement('div');
    r.className='row';
    r.style.gridTemplateColumns = `repeat(${keys.length},1fr)`;
    keys.forEach(k => r.appendChild(makeKeyBtn(k)));
    gridEl.appendChild(r);
  });
  // then tab-specific rows
  rows.forEach(row => {
    const r = document.createElement('div');
    r.className='row';
    r.style.gridTemplateColumns = `repeat(${cols},1fr)`;
    row.forEach(k => r.appendChild(makeKeyBtn(k)));
    for (let i=row.length;i<cols;i++) r.appendChild(document.createElement('div'));
    gridEl.appendChild(r);
  });
}
const actionsEl = document.getElementById('actions');
LAYOUT.base_layer.actions.forEach(k => {
  const el = document.createElement('div');
  el.className = 'act' + (k.i === '__done__' ? ' done' : '');
  el.textContent = k.k;
  el.onclick = () => handleKey(k);
  actionsEl.appendChild(el);
});

/* ---------- submit ---------- */
const resultEl = document.getElementById('result');
function submit() {
  const steps = [...linesEl.querySelectorAll('.lbody')].map(b => b.innerText.trim());
  const payload = {
    steps: steps,
    step_count: steps.length,
    sbc_message: document.getElementById('sbc').value.trim(),
    tab_used: state.tab,
  };
  resultEl.classList.add('shown');
  resultEl.innerHTML = "<div class='label'>submitted to step_grader + intervention log</div>"+
    "<pre>"+JSON.stringify(payload,null,2)+"</pre>";
}

/* ---------- boot ---------- */
addLine('', true);
renderGrid();
</script></body></html>
"""

HTML = HTML.replace("__LAYOUT__", json.dumps(kb, ensure_ascii=False))
HTML_PATH.write_text(HTML, encoding="utf-8")
print(f"  ✓ {HTML_PATH.name}  ({HTML_PATH.stat().st_size:,} bytes)")
webbrowser.open(HTML_PATH.as_uri())
print()
print("  ✓ opened in browser (no CDN, works offline)")
print()
print("  What to test:")
print("    • Line 1 appears with cursor ready")
print("    • Type x^2 - 4 = 0 on line 1, press ENTER → line 2 appears, numbered")
print("    • Tap the fraction key → a stacked fraction appears on the current line")
print("    • Tap 'Σ' on the Sequences tab → summation with editable ∞ above and n=1 below.")
print("      Click 'n=1' and replace it with 'p=0'.")
print("    • Press BACKSPACE at the start of line 2 → it merges up into line 1")
print("    • Click any earlier line to jump the cursor back there")
print("    • Anything you type in the SBC box stays in the SBC box forever")
````

````python
# ══════════════════════════════════════════════════════════════════
# MethodMarker — final keyboard
# Cross-topic map. Auto|Manual side-by-side. Responsive. Offline.
# ══════════════════════════════════════════════════════════════════
import json, webbrowser
from pathlib import Path

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
LAYOUT_PATH = ROOT / "data" / "processed" / "tutor" / "keyboard_layout.json"
HTML_PATH   = ROOT / "docs" / "keyboard_preview.html"

# ─────────── key + slot helpers ───────────
S  = "<span class='slot'></span>"
SC = "<span class='slot cur'></span>"

def k(label, ins=None, icon=None):
    d = {"l": label, "i": ins if ins is not None else label}
    if icon: d["ic"] = icon
    return d

# ─── structural templates (empty boxes) ───
def FRAC(v1=None, v2=None):
    a = v1 if v1 is not None else ""
    b = v2 if v2 is not None else ""
    return f"<span class='mq-frac'><span class='slot cur'>{a}</span><span class='slot'>{b}</span></span>"

def EXP(base=None, power=None, cur="base"):
    b = base if base is not None else ""
    p = power if power is not None else ""
    bc = " cur" if cur == "base" else ""
    pc = " cur" if cur == "power" else ""
    return (f"<span class='mq-exp'>"
            f"<span class='slot{bc}'>{b}</span>"
            f"<span class='slot mq-pow{pc}'>{p}</span>"
            f"</span>")

def SQRT(inner=None, idx=None):
    i = inner if inner is not None else ""
    if idx is None:
        return f"<span class='mq-sqrt'>√<span class='slot cur mq-rad'>{i}</span></span>"
    return (f"<span class='mq-sqrt'><span class='slot mq-idx'>{idx}</span>"
            f"√<span class='slot cur mq-rad'>{i}</span></span>")

def SUM(top=None, bot=None):
    t = top if top is not None else ""
    b = bot if bot is not None else ""
    return (f"<span class='mq-sum'>"
            f"<span class='slot mq-top cur'>{t}</span>"
            f"<span class='mq-sig'>Σ</span>"
            f"<span class='slot mq-bot'>{b}</span>"
            f"</span>")

def LOG(base=None):
    b = base if base is not None else ""
    return f"<span class='mq-log'>log<span class='slot cur'>{b}</span>(</span>"

def PAREN(preset=None, second=None):
    """( □ ± □ ) structure. preset fills first slot, second fills second."""
    p = preset if preset is not None else ""
    s = second if second is not None else ""
    return (f"<span class='mq-paren'>(<span class='slot cur'>{p}</span>"
            f"±<span class='slot'>{s}</span>)</span>")

def BRACK(inner=None):
    i = inner if inner is not None else ""
    return f"(<span class='slot cur'>{i}</span>)"

def PAIR(x=None, y=None):
    a = x if x is not None else ""
    b = y if y is not None else ""
    return f"(<span class='slot cur'>{a}</span>;<span class='slot'>{b}</span>)"

def LOGDIV():
    return (f"<span class='mq-frac'>"
            f"<span class='slot cur'>sin(□)</span>"
            f"<span class='slot'>cos(□)</span>"
            f"</span>")

def CIRC():
    return "<span class='mq-paren'>∠</span>"

# ─── icons (tiny CSS-drawn previews on the button) ───
IC_FRAC  = "<span class='ic ic-frac'><i></i><i></i></span>"
IC_EXP   = "<span class='ic ic-exp'><i></i><i></i></span>"
IC_SQRT  = "<span class='ic ic-sqrt'>&#8730;<i></i></span>"
IC_SUM   = "<span class='ic ic-sum'><i></i>&#8721;<i></i></span>"
IC_LOG   = "<span class='ic ic-log'>log<i></i></span>"
IC_QF    = "<span class='ic ic-qf'>QF</span>"

# ══════════════════════════════════════════════════════════
# BASE LAYER — always visible
# ══════════════════════════════════════════════════════════
base = {
    "structure": [
        k("a/b",  FRAC(),  IC_FRAC),
        k("□/□",  FRAC()),
        k("x²",   EXP(power="2"),  IC_EXP),
        k("xⁿ",   EXP()),
        k("√",    SQRT(),  IC_SQRT),
        k("ⁿ√",   SQRT(idx="")),
        k("∛",    f"<span class='mq-sqrt'>∛<span class='slot cur mq-rad'></span></span>"),
        k("log",  LOG()),
        k("logₐ", LOG(base="")),
        k("π",    "π"),
        k("θ",    "θ"),
        k("∞",    "∞"),
    ],
    "operators": [
        k("+", "+"), k("−", "−"), k("×", "×"), k("÷", "÷"),
        k("=", "="), k("(", "("), k(")", ")"),
        k("<", "<"), k(">", ">"),
        k("≤", "≤"), k("≥", "≥"), k("≠", "≠"),
        k("∴", " ∴ "), k("∵", " ∵ "), k("⇒", " ⇒ "),
    ],
    "digits": [
        k("1"), k("2"), k("3"), k("4"), k("5"), k("6"),
        k("7"), k("8"), k("9"), k("0"), k("."), k("(−)", "−"),
    ],
    "variables": [
        k("x"), k("y"), k("a"), k("b"), k("c"), k("n"), k("k"), k("t"),
    ],
    "actions": [
        k("DEL", "__del__"),
        k("AC",  "__clear__"),
        k("⏎",   "__enter__"),
        k("?",   "__ask__"),
        k("✓",   "__done__"),
    ],
}

# ══════════════════════════════════════════════════════════
# TABS
# ══════════════════════════════════════════════════════════
tabs_order = ["Algebra","Sequences","Functions","Calculus","Trig",
              "Finance","Analytical","Euclidean","Stats","Probability","ABC"]

tabs = {

    # LO 2 — Algebra
    "Algebra": [
        [k("x²",EXP(power="2")), k("y²",EXP(power="2")),
         k("a²",EXP(power="2")), k("b²",EXP(power="2")),
         k("c²",EXP(power="2"))],
        [k("x³",EXP(power="3")), k("xⁿ",EXP()), k("aⁿ",EXP()),
         k("bⁿ",EXP())],
        [k("(□)²",EXP(base="()",power="2")),
         k("(□)³",EXP(base="()",power="3")),
         k("(□)ⁿ",EXP(base="()"))],
        [k("±","±"),
         k("QF","<span class='mq-qf'>x = "
              "<span class='mq-frac'><span class='slot cur'></span>"
              "<span class='slot'></span></span></span>")],
    ],

    # LO 1 — Sequences
    "Sequences": [
        [k("Σ",SUM(),IC_SUM),
         k("Tₙ",EXP(base="T",power="n")),
         k("Sₙ",EXP(base="S",power="n")),
         k("Tₙ₊₁","Tₙ₊₁"),
         k("Tₙ₋₁","Tₙ₋₁"),
         k("n→∞","n→∞")],
        [k("□²",EXP(power="2")), k("□³",EXP(power="3")), k("□ⁿ",EXP()),
         k("(□)²",EXP(base="()",power="2")),
         k("(□)³",EXP(base="()",power="3")),
         k("(□)ⁿ",EXP(base="()"))],
        [k("A"), k("P"), k("i"), k("n"), k("r")],
    ],

    # LO 2 — Functions (borrows trig functions — syllabus 10.2.2)
    "Functions": [
        [k("f(□)","f(<span class='slot cur'></span>)"),
         k("g(□)","g(<span class='slot cur'></span>)"),
         k("f⁻¹(□)","f⁻¹(<span class='slot cur'></span>)"),
         k("→","→"), k("∞","∞"), k("Δ","Δ")],
        [k("sin(□)","sin(<span class='slot cur'></span>)"),
         k("cos(□)","cos(<span class='slot cur'></span>)"),
         k("tan(□)","tan(<span class='slot cur'></span>)")],
        [k("a/b",FRAC()), k("□ⁿ",EXP())],
    ],

    # LO 2 — Calculus
    "Calculus": [
        [k("dy/dx","dy/dx"),
         k("d²y/dx²","d²y/dx²"),
         k("dⁿy/dxⁿ","dⁿy/dxⁿ"),
         k("∫","∫"), k("lim","lim"), k("→","→")],
        [k("□²",EXP(power="2")), k("□³",EXP(power="3")), k("□ⁿ",EXP()),
         k("Δx","Δx"),
         k("1/□","<span class='mq-frac'>"
                 "<span class='slot cur'>1</span>"
                 "<span class='slot'></span></span>")],
    ],

    # LO 3 — Trig
    "Trig": [
        [k("sin(","sin("), k("cos(","cos("), k("tan(","tan("),
         k("sin⁻¹(","sin⁻¹("), k("cos⁻¹(","cos⁻¹("), k("tan⁻¹(","tan⁻¹(")],
        [k("α","α"), k("β","β"), k("θ","θ"), k("2α","2α"),
         k("°","°"), k("π","π")],
        # reduction formulae — auto preset | manual empty
        [k("(□±□)",PAREN()), k("(90°±θ)",PAREN(preset="90°",second="θ")),
         k("(□±θ)",PAREN(second="θ")),
         k("(180°±θ)",PAREN(preset="180°",second="θ")),
         k("(360°±θ)",PAREN(preset="360°",second="θ")),
         k("(−θ)","(−θ)")],
        # squared identities — auto | manual
        [k("sin²θ",EXP(base="sin θ",power="2")),
         k("sinⁿ□",EXP(base="sin ")),
         k("cos²θ",EXP(base="cos θ",power="2")),
         k("cosⁿ□",EXP(base="cos "))],
        [k("tan²θ",EXP(base="tan θ",power="2")),
         k("tanⁿ□",EXP(base="tan ")),
         k("sin□/cos□",LOGDIV())],
    ],

    # LO 1 — Finance
    "Finance": [
        [k("A"), k("P"), k("i"), k("n"), k("x")],
        [k("(1+i)ⁿ",EXP(base="1+i")), k("(□)ⁿ",EXP(base="()")),
         k("(1−i)ⁿ",EXP(base="1−i"))],
        [k("□ⁿ",EXP()), k("□/□",FRAC())],
    ],

    # LO 3 — Analytical
    "Analytical": [
        [k("(□;□)",PAIR()),
         k("x₁","x₁"), k("y₁","y₁"), k("x₂","x₂"), k("y₂","y₂")],
        [k("√",SQRT()), k("□²",EXP(power="2")),
         k("tan(□)","tan(<span class='slot cur'></span>)")],
        [k("θ","θ"), k("°","°")],
        [k("(□)²+(□)²",
           f"<span class='mq-exp'><span class='slot cur'></span>"
           f"<span class='slot mq-pow'>2</span></span>"
           f"+"
           f"<span class='mq-exp'><span class='slot'></span>"
           f"<span class='slot mq-pow'>2</span></span>")],
    ],

    # LO 3 — Euclidean (borrows trig — syllabus 11.3.6)
    "Euclidean": [
        [k("△","△"), k("∥","∥"), k("⊥","⊥"),
         k("∠","∠"), k("≅","≅"), k("∼","∼")],
        [k("Â","Â"), k("B̂","B̂"), k("Ĉ","Ĉ"), k("D̂","D̂")],
        [k("sin(","sin("), k("cos(","cos("), k("tan(","tan(")],
        [k("√",SQRT()), k("∴"," ∴ "), k("∵"," ∵ ")],
    ],

    # LO 4 — Stats (borrows exponents and root)
    "Stats": [
        [k("x̄","x̄"), k("σ","σ"), k("n","n"), k("r","r"),
         k("r²",EXP(base="r",power="2"))],
        [k("Σ",SUM()), k("□²",EXP(power="2")), k("√",SQRT())],
        [k("E(□)","E(<span class='slot cur'></span>)"),
         k("P(□)","P(<span class='slot cur'></span>)")],
    ],

    # LO 4 — Probability
    "Probability": [
        [k("P(□)","P(<span class='slot cur'></span>)"),
         k("P(□∪□)","P(<span class='slot cur'></span>∪<span class='slot'></span>)"),
         k("P(□∩□)","P(<span class='slot cur'></span>∩<span class='slot'></span>)"),
         k("P(□')","P(<span class='slot cur'></span>')")],
        [k("∪","∪"), k("∩","∩"), k("'","'")],
        [k("nCr","nCr"), k("nPr","nPr"), k("!","!")],
    ],

    # ABC — the full alphabet + changeable variables
    "ABC": [
        [k(L) for L in "ABCDEFGHIJ"],
        [k(L) for L in "KLMNOPQRST"],
        [k(L) for L in "UVWXYZ"] + [k(""), k(""), k("")],
        [k("x₁","x₁"), k("y₁","y₁"), k("x₂","x₂"), k("y₂","y₂"),
         k("p","p"), k("q","q"), k("m","m"), k("r","r"),
         k("s","s"), k("u","u")],
    ],
}

# ══════════════════════════════════════════════════════════
# Write JSON
# ══════════════════════════════════════════════════════════
payload = {
    "schema_version": 10.0,
    "base": base,
    "tabs_order": tabs_order,
    "tabs": tabs,
}
LAYOUT_PATH.write_text(json.dumps(payload, indent=2, ensure_ascii=False),
                       encoding="utf-8")
print(f"  ✓ {LAYOUT_PATH.name} written")

# ══════════════════════════════════════════════════════════
# HTML
# ══════════════════════════════════════════════════════════
HTML_TPL = r"""<!doctype html>
<html lang="en"><head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width,initial-scale=1,maximum-scale=1">
<title>MethodMarker Keyboard</title>
<style>
/* ────── design tokens ────── */
:root{
  --bg:#eef2f8; --card:#fff; --line:#e4e9f2; --ink:#0b1f44;
  --ink2:#465a80; --ink3:#9aa8c2; --blue:#3b5bf6; --blue2:#5b8cf6;
  --amber:#f0ad4e; --key-h: clamp(38px, 6.4vw, 48px);
  --key-f: clamp(11px, 1.55vw, 15px);
  --pad: clamp(6px, 1.2vw, 10px);
  --radius: 12px;
}
*{box-sizing:border-box;-webkit-tap-highlight-color:transparent}
html,body{margin:0;height:100%}
body{
  background:var(--bg);color:var(--ink);
  font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,sans-serif;
  display:flex;justify-content:center;padding:8px;
  min-height:100vh;overflow-x:hidden;
}
.app{
  width:100%;max-width:720px;background:var(--card);
  border-radius:18px;overflow:hidden;
  box-shadow:0 8px 32px rgba(20,40,90,.12);
  display:flex;flex-direction:column;
  max-height:calc(100vh - 16px);
}

/* ────── topbar ────── */
.topbar{display:flex;align-items:center;gap:10px;
  padding:10px 14px;background:#f6f8fd;border-bottom:1px solid var(--line);flex:none}
.logo{width:26px;height:26px;border-radius:7px;
  background:linear-gradient(135deg,var(--blue),var(--blue2));
  display:flex;align-items:center;justify-content:center;
  color:#fff;font-weight:700;font-size:14px;flex:none}
.brand{font-weight:700;font-size:13px;color:var(--ink)}
.tag{font-size:10px;color:#6b7a99}

/* ────── working area (fixed height, internal scroll) ────── */
.work{background:#fff;padding:10px 14px 12px;flex:none;
  max-height:clamp(120px,26vh,240px);overflow-y:auto}
.work-head{display:flex;justify-content:space-between;align-items:center;
  font-size:11px;color:var(--ink2);font-weight:600;margin-bottom:6px;
  position:sticky;top:0;background:#fff;padding-bottom:4px}
.work-head .hint{color:var(--ink3);font-weight:500;font-size:10px}
.lines{display:flex;flex-direction:column;gap:1px}
.lrow{display:grid;grid-template-columns:22px 1fr;column-gap:6px;align-items:flex-start}
.lnum{color:#b8c2d6;font-size:10px;text-align:right;padding-top:9px;
  user-select:none;font-variant-numeric:tabular-nums}
.lbody{min-height:32px;padding:6px 8px;border-radius:5px;
  font-size:clamp(15px,2.1vw,19px);line-height:1.4;color:var(--ink);
  outline:none;font-family:'Cambria Math','Latin Modern Math',Georgia,serif;
  white-space:pre-wrap;word-break:break-word;
  transition:background .1s;caret-color:var(--blue)}
.lbody:focus{background:#f5f8ff}

/* ────── SBC (collapsed one line → expand on focus) ────── */
.sbc{padding:8px 14px 10px;background:#fff;border-top:1px dashed var(--line);flex:none}
.sbc-head{display:flex;align-items:center;gap:6px;font-size:11px;
  color:var(--ink2);font-weight:600;margin-bottom:5px}
.sbc-head .dot{width:6px;height:6px;border-radius:50%;background:var(--amber)}
#sbc{width:100%;height:30px;resize:none;font-family:inherit;
  font-size:12px;padding:6px 10px;border:1px solid var(--line);
  border-radius:6px;background:#fffdf7;color:var(--ink);outline:none;
  transition:height .2s ease,box-shadow .15s}
#sbc:focus{height:76px;box-shadow:0 0 0 3px rgba(240,173,78,.15)}

/* ────── tabs ────── */
.tabs{display:flex;overflow-x:auto;gap:3px;padding:5px 10px;
  background:#f6f8fd;border-top:1px solid var(--line);
  border-bottom:1px solid var(--line);scrollbar-width:none;flex:none}
.tabs::-webkit-scrollbar{display:none}
.tab{flex:none;padding:5px 10px;border-radius:14px;font-size:10.5px;
  font-weight:600;color:var(--ink2);background:transparent;
  border:none;cursor:pointer;white-space:nowrap;font-family:inherit}
.tab.active{background:var(--blue);color:#fff}

/* ────── keyboard (pinned bottom, internal scroll) ────── */
.kb-scroll{
  flex:1 1 auto;overflow-y:auto;background:#fff;
  display:flex;flex-direction:column;
}
.kb{padding:var(--pad);display:flex;flex-direction:column;gap:5px}
.row{display:grid;gap:5px}
.key{
  padding:4px 2px;border-radius:8px;background:#fff;
  border:1px solid #dfe5ef;color:var(--ink);
  font-size:var(--key-f);cursor:pointer;text-align:center;
  user-select:none;font-family:inherit;
  min-height:var(--key-h);
  display:flex;align-items:center;justify-content:center;
  transition:transform .05s,background .1s,box-shadow .1s;
  overflow:hidden;
}
.key:active{transform:scale(.95);background:#eef2fb}
.key.var{background:#f6f8fd}
.key.wide{grid-column:span 2}
.acts{display:grid;grid-template-columns:repeat(5,1fr);gap:5px;
  padding:6px var(--pad) 8px;background:#f6f8fd;
  border-top:1px solid var(--line);flex:none}
.act{
  padding:8px 2px;border-radius:8px;background:#eef2fb;
  border:1px solid #dfe5ef;color:#223a6b;
  font-size:11px;font-weight:700;cursor:pointer;text-align:center;
  user-select:none;min-height:var(--key-h);
  display:flex;align-items:center;justify-content:center;
  font-family:inherit;transition:transform .05s
}
.act.done{background:var(--blue);color:#fff;border-color:var(--blue)}
.act:active{transform:scale(.97)}

/* ────── icons on structural keys ────── */
.ic{display:inline-flex;align-items:center;justify-content:center}
.ic-frac{flex-direction:column;gap:1px}
.ic-frac i{display:block;width:14px;height:6px;
  border:1px solid #7b8ba6;border-radius:2px}
.ic-frac i:nth-child(2){width:18px;height:1px;border:none;background:#7b8ba6;border-radius:0}
.ic-exp{position:relative;width:22px;height:20px}
.ic-exp i:nth-child(1){position:absolute;bottom:0;left:0;
  width:14px;height:14px;border:1px solid #7b8ba6;border-radius:2px}
.ic-exp i:nth-child(2){position:absolute;top:0;right:0;
  width:8px;height:8px;border:1px solid #7b8ba6;border-radius:2px}
.ic-sqrt{font-size:16px;font-weight:700;line-height:1}
.ic-sqrt i{display:inline-block;width:9px;height:11px;
  border-top:1px solid #7b8ba6;border-right:1px solid #7b8ba6;
  border-bottom:1px solid #7b8ba6;margin-left:-1px}
.ic-sum{flex-direction:column;font-size:12px;line-height:1}
.ic-sum i{display:block;width:14px;height:4px;
  border:1px solid #7b8ba6;border-radius:1px;margin:1px 0}
.ic-log{font-size:11px;font-weight:700}
.ic-log i{display:inline-block;width:8px;height:6px;
  border:1px solid #7b8ba6;border-radius:2px;margin-left:1px}
.ic-qf{font-size:12px;font-weight:700;color:var(--ink2)}

/* ────── editable math structures inside the working area ────── */
.slot{
  display:inline-block;min-width:12px;min-height:18px;
  padding:0 3px;vertical-align:middle;text-align:center;
  outline:none;
}
.slot:focus{background:#e8efff;border-radius:3px;box-shadow:0 0 0 1px rgba(59,91,246,.35)}

/* fraction */
.mq-frac{display:inline-flex;flex-direction:column;align-items:center;
  vertical-align:middle;padding:0 3px;margin:0 2px;line-height:1.1}
.mq-frac .slot:first-child{border-bottom:1.7px solid var(--ink);
  min-width:18px;padding:1px 6px}
.mq-frac .slot:last-child{padding-top:2px;min-width:18px;padding-left:6px;padding-right:6px}

/* exponent */
.mq-exp{display:inline-flex;align-items:flex-start;vertical-align:middle;padding:0 2px}
.mq-exp .slot{min-width:12px}
.mq-exp .mq-pow{font-size:.62em;position:relative;top:-.55em;
  padding-left:1px;min-width:12px}

/* root */
.mq-sqrt{display:inline-flex;align-items:stretch;vertical-align:middle;
  line-height:1;padding:0 3px}
.mq-sqrt .mq-rad{border-top:1.7px solid var(--ink);
  border-right:1.7px solid var(--ink);border-bottom:1.7px solid var(--ink);
  min-width:16px;min-height:22px;padding:3px 6px;
  display:inline-flex;align-items:center}
.mq-sqrt .mq-idx{font-size:.62em;position:relative;top:-.55em;min-width:10px}

/* summation */
.mq-sum{display:inline-flex;flex-direction:column;align-items:center;
  vertical-align:middle;padding:0 6px;margin:0 3px;line-height:1}
.mq-sum .mq-top{font-size:.6em;min-width:22px;text-align:center;
  padding:1px 2px;min-height:14px}
.mq-sum .mq-sig{font-size:1.2em;line-height:1;margin:1px 0}
.mq-sum .mq-bot{font-size:.6em;min-width:22px;text-align:center;
  padding:1px 2px;min-height:14px}

/* log */
.mq-log{display:inline-flex;align-items:flex-end;vertical-align:middle;padding:0 2px}
.mq-log .slot{font-size:.62em;min-width:12px;min-height:12px;text-align:center}

/* parenthesised bracket structures */
.mq-paren{display:inline-flex;align-items:center;padding:0 2px;
  font-family:'Cambria Math',Georgia,serif}

/* quadratic formula wrapper */
.mq-qf{display:inline-flex;align-items:center;padding:0 3px;
  font-family:'Cambria Math',Georgia,serif}

/* ────── result strip ────── */
.result{padding:8px 14px;background:var(--ink);color:#fff;
  font-family:'SF Mono',Menlo,monospace;font-size:10px;line-height:1.5;
  max-height:0;overflow:hidden;transition:max-height .25s ease;flex:none}
.result.shown{max-height:240px;overflow-y:auto}
.result .label{color:#9ab;margin-bottom:4px}
.result pre{margin:0;color:#cfe2ff;white-space:pre-wrap}

/* ────── phone tweaks ────── */
@media (max-width:480px){
  .work{max-height:clamp(100px,22vh,180px)}
  .lbody{font-size:15px}
  .kb{padding:6px}
  .row{gap:4px}
  .key{min-height:38px;border-radius:6px}
  .topbar{padding:8px 12px}
}
@media (min-width:1025px){
  .work{max-height:260px}
}
</style></head><body>

<div class="app">

  <div class="topbar">
    <div class="logo">M</div>
    <div><div class="brand">MethodMarker</div>
    <div class="tag">Better Methods. Higher Marks.</div></div>
  </div>

  <div class="work">
    <div class="work-head">
      <span>&#9998; Working</span>
      <span class="hint">tap a line · ⏎ new step · ⌫ at start merges up</span>
    </div>
    <div class="lines" id="lines"></div>
  </div>

  <div class="sbc">
    <div class="sbc-head"><span class="dot"></span>
      Ask Tanaka — where did you get stuck?</div>
    <textarea id="sbc" rows="1"
      placeholder="e.g. I did step 2 but I don't understand why the sign flips"></textarea>
  </div>

  <div class="tabs" id="tabs"></div>
  <div class="kb-scroll"><div class="kb" id="kb"></div></div>
  <div class="acts" id="acts"></div>
  <div class="result" id="result"></div>
</div>

<script>
const LAYOUT = __LAYOUT__;
const state = { tab: LAYOUT.tabs_order[0] };

/* ═══════════ line editor ═══════════ */
const linesEl = document.getElementById('lines');

function addLine(html, focus){
  const row = document.createElement('div'); row.className='lrow';
  const n = document.createElement('div'); n.className='lnum';
  const b = document.createElement('div'); b.className='lbody';
  b.contentEditable='true'; b.spellcheck=false;
  if (html) b.innerHTML = html;
  row.appendChild(n); row.appendChild(b);
  linesEl.appendChild(row);
  renumber();
  if (focus) focusStart(b);
  return b;
}
function renumber(){
  [...linesEl.querySelectorAll('.lrow')].forEach((r,i)=>{
    r.querySelector('.lnum').textContent = i+1;
  });
}
function focusStart(el){
  el.focus();
  const first = el.querySelector('.slot');
  const r = document.createRange();
  if (first) r.selectNodeContents(first);
  else { r.selectNodeContents(el); r.collapse(false); }
  r.collapse(true);
  const s = window.getSelection(); s.removeAllRanges(); s.addRange(r);
}
function activeLine(){
  const sel = window.getSelection();
  if (!sel.rangeCount) return null;
  let n = sel.getRangeAt(0).startContainer;
  while (n && n !== linesEl){
    if (n.classList && n.classList.contains('lbody')) return n;
    n = n.parentNode;
  }
  return null;
}
function currentSlot(){
  const sel = window.getSelection();
  if (!sel.rangeCount) return null;
  let n = sel.getRangeAt(0).startContainer;
  if (n.nodeType === 3) n = n.parentNode;
  return n.closest ? n.closest('.slot') : null;
}

/* Enter → new line (split). Backspace at line start → merge up. Tab → next slot. */
linesEl.addEventListener('keydown', e => {
  const line = e.target.closest('.lbody'); if (!line) return;

  if (e.key === 'Enter' && !e.shiftKey){
    e.preventDefault();
    const sel = window.getSelection();
    const r = sel.getRangeAt(0);
    const after = r.cloneRange();
    after.selectNodeContents(line);
    after.setStart(r.endContainer, r.endOffset);
    const rest = after.extractContents();
    const nb = addLine('', false);
    while (rest.firstChild) nb.insertBefore(rest.firstChild, nb.firstChild);
    focusStart(nb);
    return;
  }

  if (e.key === 'Tab'){
    e.preventDefault();
    const slot = currentSlot();
    if (!slot) return;
    const struct = slot.closest('.mq-frac,.mq-exp,.mq-sqrt,.mq-sum,.mq-log,.mq-paren,.mq-qf');
    if (!struct) return;
    const slots = [...struct.querySelectorAll('.slot')];
    const i = slots.indexOf(slot);
    const next = slots[i+1];
    const sel = window.getSelection();
    if (next){
      const r = document.createRange(); r.selectNodeContents(next); r.collapse(true);
      sel.removeAllRanges(); sel.addRange(r);
    } else {
      const r = document.createRange(); r.setStartAfter(struct); r.collapse(true);
      sel.removeAllRanges(); sel.addRange(r);
    }
    return;
  }

  if (e.key === 'Backspace'){
    const sel = window.getSelection();
    const r = sel.getRangeAt(0);
    if (!r.collapsed) return;
    const probe = r.cloneRange(); probe.selectNodeContents(line);
    probe.setEnd(r.startContainer, r.startOffset);
    if (probe.toString().length !== 0) return;
    const prev = line.closest('.lrow').previousElementSibling;
    if (prev){
      e.preventDefault();
      const pb = prev.querySelector('.lbody');
      while (line.firstChild) pb.appendChild(line.firstChild);
      line.closest('.lrow').remove();
      focusStart(pb);
      renumber();
    } else if (!line.textContent.trim()){
      e.preventDefault();
    }
  }
});

/* ═══════════ insertion ═══════════ */
function insertAtCursor(html){
  let line = activeLine();
  if (!line) line = addLine('', true);
  line.focus();

  const sel = window.getSelection();
  if (!sel.rangeCount) return;
  const r = sel.getRangeAt(0);
  r.deleteContents();

  const tmp = document.createElement('span');
  tmp.innerHTML = html;
  const nodes = [...tmp.childNodes];
  if (!nodes.length) return;

  nodes.forEach(n => r.insertNode(n));
  const last = nodes[nodes.length - 1];

  /* focus first .cur slot inside what we just inserted */
  let target = null;
  if (last.nodeType === 1){
    target = last.querySelector('.slot.cur') || last.querySelector('.slot');
  }
  if (!target && last.previousSibling && last.previousSibling.nodeType === 1){
    target = last.previousSibling.querySelector &&
             last.previousSibling.querySelector('.slot');
  }
  if (target){
    const nr = document.createRange();
    nr.selectNodeContents(target); nr.collapse(true);
    sel.removeAllRanges(); sel.addRange(nr);
  } else {
    const nr = document.createRange();
    nr.setStartAfter(last); nr.collapse(true);
    sel.removeAllRanges(); sel.addRange(nr);
  }
}

/* ═══════════ key handling ═══════════ */
function handleKey(key){
  if (key.i === '__del__')   { document.execCommand('delete'); return; }
  if (key.i === '__clear__') { linesEl.innerHTML=''; addLine('', true); return; }
  if (key.i === '__enter__') { addLine('', true); return; }
  if (key.i === '__ask__')   { document.getElementById('sbc').focus(); return; }
  if (key.i === '__done__')  { submit(); return; }
  insertAtCursor(key.i);
}

/* ═══════════ tabs ═══════════ */
const tabsEl = document.getElementById('tabs');
LAYOUT.tabs_order.forEach(name => {
  const b = document.createElement('button');
  b.className = 'tab' + (name === state.tab ? ' active' : '');
  b.textContent = name;
  b.onclick = () => {
    state.tab = name;
    document.querySelectorAll('.tab').forEach(t =>
      t.classList.toggle('active', t.textContent === name));
    renderKb();
  };
  tabsEl.appendChild(b);
});

/* ═══════════ key grid ═══════════ */
const kbEl = document.getElementById('kb');
function keyBtn(key){
  const el = document.createElement('div');
  el.className = 'key';
  if (key.ic) el.innerHTML = key.ic;
  else el.textContent = key.l;
  el.onclick = () => handleKey(key);
  return el;
}
function renderKb(){
  kbEl.innerHTML = '';

  /* base layer rows (always) */
  const baseRows = [
    LAYOUT.base.structure,
    LAYOUT.base.operators,
    LAYOUT.base.digits,
    LAYOUT.base.variables,
  ];
  baseRows.forEach(keys => {
    const r = document.createElement('div'); r.className='row';
    r.style.gridTemplateColumns = `repeat(${keys.length},1fr)`;
    keys.forEach(k => r.appendChild(keyBtn(k)));
    kbEl.appendChild(r);
  });

  /* tab-specific rows */
  const rows = LAYOUT.tabs[state.tab] || [];
  rows.forEach(row => {
    const r = document.createElement('div'); r.className='row';
    r.style.gridTemplateColumns = `repeat(${row.length},1fr)`;
    row.forEach(k => r.appendChild(keyBtn(k)));
    kbEl.appendChild(r);
  });
}

/* ═══════════ actions ═══════════ */
const actsEl = document.getElementById('acts');
LAYOUT.base.actions.forEach(k => {
  const el = document.createElement('div');
  el.className = 'act' + (k.i === '__done__' ? ' done' : '');
  el.textContent = k.l;
  el.onclick = () => handleKey(k);
  actsEl.appendChild(el);
});

/* ═══════════ submit ═══════════ */
const resultEl = document.getElementById('result');
function submit(){
  const steps = [...linesEl.querySelectorAll('.lbody')]
    .map(b => b.innerText.replace(/\u200B/g,'').trim())
    .filter(s => s.length);
  const payload = {
    steps:        steps,
    step_count:   steps.length,
    sbc_message:  document.getElementById('sbc').value.trim(),
    tab_used:     state.tab,
  };
  resultEl.classList.add('shown');
  resultEl.innerHTML =
    "<div class='label'>submitted to step_grader + intervention log</div>"+
    "<pre>"+JSON.stringify(payload,null,2)+"</pre>";
}

/* ═══════════ boot ═══════════ */
addLine('', true);
renderKb();
</script>
</body></html>
"""

html = HTML_TPL.replace("__LAYOUT__", json.dumps(payload, ensure_ascii=False))
HTML_PATH.write_text(html, encoding="utf-8")
print(f"  ✓ {HTML_PATH.name} written ({HTML_PATH.stat().st_size:,} bytes)")

webbrowser.open(HTML_PATH.as_uri())
print()
print("  → opened in browser (fully offline)")
````

````python
# ══════════════════════════════════════════════════════════════════
# MethodMarker — keyboard v15
# Horizontal digits row · no scrolling · DEL direct · all keys visible
# ══════════════════════════════════════════════════════════════════
import json, webbrowser
from pathlib import Path

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
LAYOUT_PATH = ROOT / "data" / "processed" / "tutor" / "keyboard_layout.json"
HTML_PATH   = ROOT / "docs" / "keyboard_preview.html"

def k(label, ins=None, icon=None):
    d = {"l": label, "i": ins if ins is not None else label}
    if icon: d["ic"] = icon
    return d

def FRAC(v1=None, v2=None):
    a = v1 if v1 is not None else ""
    b = v2 if v2 is not None else ""
    return f"<span class='mq-frac'><span class='slot cur'>{a}</span><span class='slot'>{b}</span></span>"

def EXP(base=None, power=None):
    b = base if base is not None else ""
    p = power if power is not None else ""
    return (f"<span class='mq-exp'>"
            f"<span class='slot cur'>{b}</span>"
            f"<span class='slot mq-pow'>{p}</span>"
            f"</span>")

def SQRT(inner=None, idx=None):
    i = inner if inner is not None else ""
    if idx is None:
        return f"<span class='mq-sqrt'>√<span class='slot cur mq-rad'>{i}</span></span>"
    return (f"<span class='mq-sqrt'><span class='slot mq-idx'>{idx}</span>"
            f"√<span class='slot cur mq-rad'>{i}</span></span>")

def SUM(top=None, bot=None):
    t = top if top is not None else ""
    b = bot if bot is not None else ""
    return (f"<span class='mq-sum'>"
            f"<span class='slot mq-top cur'>{t}</span>"
            f"<span class='mq-sig'>Σ</span>"
            f"<span class='slot mq-bot'>{b}</span>"
            f"</span>")

def LOG(base=None):
    b = base if base is not None else ""
    return f"<span class='mq-log'>log<span class='slot cur'>{b}</span>(</span>"

def PAREN(preset=None, second=None):
    p = preset if preset is not None else ""
    s = second if second is not None else ""
    return (f"<span class='mq-paren'>(<span class='slot cur'>{p}</span>"
            f"±<span class='slot'>{s}</span>)</span>")

def PAIR(x=None, y=None):
    a = x if x is not None else ""
    b = y if y is not None else ""
    return f"(<span class='slot cur'>{a}</span>;<span class='slot'>{b}</span>)"

def LOGDIV():
    return ("<span class='mq-frac'>"
            "<span class='slot cur'>sin(□)</span>"
            "<span class='slot'>cos(□)</span>"
            "</span>")

IC_FRAC = "<span class='ic ic-frac'><i></i><i></i></span>"
IC_EXP  = "<span class='ic ic-exp'><i></i><i></i></span>"
IC_SQRT = "<span class='ic ic-sqrt'>&#8730;<i></i></span>"
IC_SUM  = "<span class='ic ic-sum'><i></i>&#8721;<i></i></span>"
IC_LOG  = "<span class='ic ic-log'>log<i></i></span>"

# ══════════════════════════════════════════════════════════
# BASE LAYER
# ══════════════════════════════════════════════════════════
base = {
    "structure": [
        k("a/b",  FRAC(),  IC_FRAC),
        k("□/□",  FRAC()),
        k("x²",   EXP(power="2"),  IC_EXP),
        k("xⁿ",   EXP()),
        k("√",    SQRT(),  IC_SQRT),
        k("ⁿ√",   SQRT(idx="")),
        k("∛",    "<span class='mq-sqrt'>∛<span class='slot cur mq-rad'></span></span>"),
        k("log",  LOG()),
        k("logₐ", LOG(base="")),
        k("π",    "π"),
        k("θ",    "θ"),
        k("∞",    "∞"),
    ],
    "operators": [
        k("+", "+"), k("−", "−"), k("×", "×"), k("÷", "÷"),
        k("=", "="), k("(", "("), k(")", ")"),
        k("<", "<"), k(">", ">"),
        k("≤", "≤"), k("≥", "≥"), k("≠", "≠"),
        k("∴", " ∴ "), k("∵", " ∵ "), k("⇒", " ⇒ "),
    ],
    # ONE horizontal row of digits
    "digits": [
        k("1"), k("2"), k("3"), k("4"), k("5"), k("6"),
        k("7"), k("8"), k("9"), k("0"), k("."), k("(−)", "−"),
    ],
    "variables": [
        k("x"), k("y"), k("a"), k("b"), k("c"), k("n"), k("k"), k("t"),
    ],
    "actions": [
        k("DEL", "__del__"),
        k("AC",  "__clear__"),
        k("ABC", "__abc__"),
        k("⏎",   "__enter__"),
        k("?",   "__ask__"),
        k("✓",   "__done__"),
    ],
    "abc": [
        [k(L) for L in "ABCDEFGHIJ"],
        [k(L) for L in "KLMNOPQRST"],
        [k(L) for L in "UVWXYZ"] + [k(""), k(""), k("")],
        [k("x₁","x₁"), k("y₁","y₁"), k("x₂","x₂"), k("y₂","y₂"),
         k("p","p"), k("q","q"), k("m","m"), k("r","r"), k("s","s"), k("u","u")],
    ],
}

# ══════════════════════════════════════════════════════════
# TABS
# ══════════════════════════════════════════════════════════
tabs_order = ["Algebra","Sequences","Functions","Calculus","Trig",
              "Finance","Analytical","Euclidean","Stats","Probability"]

tabs = {
    "Algebra": [
        [k("x²",EXP(power="2")), k("y²",EXP(power="2")),
         k("a²",EXP(power="2")), k("b²",EXP(power="2")),
         k("c²",EXP(power="2"))],
        [k("x³",EXP(power="3")), k("xⁿ",EXP()), k("aⁿ",EXP()), k("bⁿ",EXP())],
        [k("(□)²",EXP(base="()",power="2")),
         k("(□)³",EXP(base="()",power="3")),
         k("(□)ⁿ",EXP(base="()"))],
        [k("±","±"),
         k("QF","<span class='mq-qf'>x = "
              "<span class='mq-frac'><span class='slot cur'></span>"
              "<span class='slot'></span></span></span>")],
    ],
    "Sequences": [
        [k("Σ",SUM(),IC_SUM),
         k("Tₙ",EXP(base="T",power="n")),
         k("Sₙ",EXP(base="S",power="n")),
         k("Tₙ₊₁","Tₙ₊₁"), k("Tₙ₋₁","Tₙ₋₁"), k("n→∞","n→∞")],
        [k("□²",EXP(power="2")), k("□³",EXP(power="3")), k("□ⁿ",EXP()),
         k("(□)²",EXP(base="()",power="2")),
         k("(□)³",EXP(base="()",power="3")),
         k("(□)ⁿ",EXP(base="()"))],
        [k("A"), k("P"), k("i"), k("n"), k("r")],
    ],
    "Functions": [
        [k("f(□)","f(<span class='slot cur'></span>)"),
         k("g(□)","g(<span class='slot cur'></span>)"),
         k("f⁻¹(□)","f⁻¹(<span class='slot cur'></span>)"),
         k("→","→"), k("∞","∞"), k("Δ","Δ")],
        [k("sin(□)","sin(<span class='slot cur'></span>)"),
         k("cos(□)","cos(<span class='slot cur'></span>)"),
         k("tan(□)","tan(<span class='slot cur'></span>)")],
        [k("a/b",FRAC()), k("□ⁿ",EXP())],
    ],
    "Calculus": [
        [k("dy/dx","dy/dx"), k("d²y/dx²","d²y/dx²"),
         k("dⁿy/dxⁿ","dⁿy/dxⁿ"), k("∫","∫"), k("lim","lim"), k("→","→")],
        [k("□²",EXP(power="2")), k("□³",EXP(power="3")), k("□ⁿ",EXP()),
         k("Δx","Δx"),
         k("1/□","<span class='mq-frac'><span class='slot cur'>1</span>"
                 "<span class='slot'></span></span>")],
    ],
    "Trig": [
        [k("sin(","sin("), k("cos(","cos("), k("tan(","tan("),
         k("sin⁻¹(","sin⁻¹("), k("cos⁻¹(","cos⁻¹("), k("tan⁻¹(","tan⁻¹(")],
        [k("α","α"), k("β","β"), k("θ","θ"), k("2α","2α"), k("°","°"), k("π","π")],
        [k("(□±□)",PAREN()), k("(90°±θ)",PAREN(preset="90°",second="θ")),
         k("(□±θ)",PAREN(second="θ")),
         k("(180°±θ)",PAREN(preset="180°",second="θ")),
         k("(360°±θ)",PAREN(preset="360°",second="θ")),
         k("(−θ)","(−θ)")],
        [k("sin²θ",EXP(base="sin θ",power="2")),
         k("sinⁿ□",EXP(base="sin ")),
         k("cos²θ",EXP(base="cos θ",power="2")),
         k("cosⁿ□",EXP(base="cos "))],
        [k("tan²θ",EXP(base="tan θ",power="2")),
         k("tanⁿ□",EXP(base="tan ")),
         k("sin□/cos□",LOGDIV())],
    ],
    "Finance": [
        [k("A"), k("P"), k("i"), k("n"), k("x")],
        [k("(1+i)ⁿ",EXP(base="1+i")), k("(□)ⁿ",EXP(base="()")),
         k("(1−i)ⁿ",EXP(base="1−i"))],
        [k("□ⁿ",EXP()), k("□/□",FRAC())],
    ],
    "Analytical": [
        [k("(□;□)",PAIR()), k("x₁","x₁"), k("y₁","y₁"), k("x₂","x₂"), k("y₂","y₂")],
        [k("√",SQRT()), k("□²",EXP(power="2")),
         k("tan(□)","tan(<span class='slot cur'></span>)")],
        [k("θ","θ"), k("°","°")],
        [k("(□)²+(□)²",
           "<span class='mq-exp'><span class='slot cur'></span>"
           "<span class='slot mq-pow'>2</span></span>+"
           "<span class='mq-exp'><span class='slot'></span>"
           "<span class='slot mq-pow'>2</span></span>")],
    ],
    "Euclidean": [
        [k("△","△"), k("∥","∥"), k("⊥","⊥"),
         k("∠","∠"), k("≅","≅"), k("∼","∼")],
        [k("Â","Â"), k("B̂","B̂"), k("Ĉ","Ĉ"), k("D̂","D̂")],
        [k("sin(","sin("), k("cos(","cos("), k("tan(","tan(")],
        [k("√",SQRT()), k("∴"," ∴ "), k("∵"," ∵ ")],
    ],
    "Stats": [
        [k("x̄","x̄"), k("σ","σ"), k("n","n"), k("r","r"),
         k("r²",EXP(base="r",power="2"))],
        [k("Σ",SUM()), k("□²",EXP(power="2")), k("√",SQRT())],
        [k("E(□)","E(<span class='slot cur'></span>)"),
         k("P(□)","P(<span class='slot cur'></span>)")],
    ],
    "Probability": [
        [k("P(□)","P(<span class='slot cur'></span>)"),
         k("P(□∪□)","P(<span class='slot cur'></span>∪<span class='slot'></span>)"),
         k("P(□∩□)","P(<span class='slot cur'></span>∩<span class='slot'></span>)"),
         k("P(□')","P(<span class='slot cur'></span>')")],
        [k("∪","∪"), k("∩","∩"), k("'","'")],
        [k("nCr","nCr"), k("nPr","nPr"), k("!","!")],
    ],
}

payload = {"schema_version": 15.0, "base": base,
           "tabs_order": tabs_order, "tabs": tabs}
LAYOUT_PATH.write_text(json.dumps(payload, indent=2, ensure_ascii=False),
                       encoding="utf-8")
print(f"  ✓ {LAYOUT_PATH.name} written")

# ══════════════════════════════════════════════════════════
# HTML
# ══════════════════════════════════════════════════════════
HTML_TPL = r"""<!doctype html>
<html lang="en"><head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width,initial-scale=1,maximum-scale=1">
<title>MethodMarker Keyboard</title>
<style>
:root{
  --bg:#eef2f8; --card:#fff; --line:#e4e9f2; --ink:#0b1f44;
  --ink2:#465a80; --ink3:#9aa8c2; --blue:#3b5bf6; --blue2:#5b8cf6;
  --amber:#f0ad4e;
  --key-h: clamp(30px, 5.2vw, 40px);
  --key-f: clamp(10px, 1.4vw, 13px);
  --pad: clamp(4px, 0.9vw, 8px);
}
*{box-sizing:border-box;-webkit-tap-highlight-color:transparent}
html,body{margin:0;height:100%}
body{
  background:var(--bg);color:var(--ink);
  font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,sans-serif;
  display:flex;justify-content:center;padding:6px;
  min-height:100vh;overflow:hidden;
}
.app{
  width:100%;max-width:720px;background:var(--card);
  border-radius:16px;overflow:hidden;
  box-shadow:0 8px 32px rgba(20,40,90,.12);
  display:flex;flex-direction:column;
  height:calc(100vh - 12px);
}
.topbar{display:flex;align-items:center;gap:8px;
  padding:6px 12px;background:#f6f8fd;border-bottom:1px solid var(--line);flex:none}
.logo{width:22px;height:22px;border-radius:6px;
  background:linear-gradient(135deg,var(--blue),var(--blue2));
  display:flex;align-items:center;justify-content:center;
  color:#fff;font-weight:700;font-size:12px;flex:none}
.brand{font-weight:700;font-size:12px;color:var(--ink)}
.tag{font-size:9px;color:#6b7a99}

.work{background:#fff;padding:6px 12px 8px;flex:none;
  max-height:clamp(80px,16vh,140px);overflow-y:auto}
.work-head{display:flex;justify-content:space-between;align-items:center;
  font-size:10px;color:var(--ink2);font-weight:600;margin-bottom:4px;
  position:sticky;top:0;background:#fff;padding-bottom:2px}
.work-head .hint{color:var(--ink3);font-weight:500;font-size:9px}
.lines{display:flex;flex-direction:column;gap:1px}
.lrow{display:grid;grid-template-columns:20px 1fr;column-gap:5px;align-items:flex-start}
.lnum{color:#b8c2d6;font-size:9px;text-align:right;padding-top:7px;
  user-select:none;font-variant-numeric:tabular-nums}
.lbody{min-height:26px;padding:4px 6px;border-radius:5px;
  font-size:clamp(14px,1.9vw,17px);line-height:1.35;color:var(--ink);
  outline:none;font-family:'Cambria Math','Latin Modern Math',Georgia,serif;
  white-space:pre-wrap;word-break:break-word;
  transition:background .1s;caret-color:var(--blue)}
.lbody:focus{background:#f5f8ff}

.sbc{padding:5px 12px 7px;background:#fff;border-top:1px dashed var(--line);flex:none}
.sbc-head{display:flex;align-items:center;gap:5px;font-size:10px;
  color:var(--ink2);font-weight:600;margin-bottom:4px}
.sbc-head .dot{width:5px;height:5px;border-radius:50%;background:var(--amber)}
#sbc{width:100%;height:26px;resize:none;font-family:inherit;
  font-size:11px;padding:5px 8px;border:1px solid var(--line);
  border-radius:6px;background:#fffdf7;color:var(--ink);outline:none;
  transition:height .2s ease,box-shadow .15s}
#sbc:focus{height:60px;box-shadow:0 0 0 3px rgba(240,173,78,.15)}

.tabs{display:flex;overflow-x:auto;gap:3px;padding:4px 8px;
  background:#f6f8fd;border-top:1px solid var(--line);
  border-bottom:1px solid var(--line);scrollbar-width:none;flex:none}
.tabs::-webkit-scrollbar{display:none}
.tab{flex:none;padding:4px 9px;border-radius:12px;font-size:10px;
  font-weight:600;color:var(--ink2);background:transparent;
  border:none;cursor:pointer;white-space:nowrap;font-family:inherit}
.tab.active{background:var(--blue);color:#fff}

.kb-scroll{flex:1 1 auto;overflow-y:auto;background:#fff;display:flex;flex-direction:column}
.kb{padding:var(--pad);display:flex;flex-direction:column;gap:4px}
.row{display:grid;gap:4px}
.key{padding:2px;border-radius:7px;background:#fff;
  border:1px solid #dfe5ef;color:var(--ink);
  font-size:var(--key-f);cursor:pointer;text-align:center;
  user-select:none;font-family:inherit;min-height:var(--key-h);
  display:flex;align-items:center;justify-content:center;
  transition:transform .05s,background .1s;overflow:hidden}
.key:active{transform:scale(.95);background:#eef2fb}
.acts{display:grid;grid-template-columns:repeat(6,1fr);gap:4px;
  padding:5px var(--pad) 6px;background:#f6f8fd;
  border-top:1px solid var(--line);flex:none}
.act{padding:6px 2px;border-radius:7px;background:#eef2fb;
  border:1px solid #dfe5ef;color:#223a6b;
  font-size:10px;font-weight:700;cursor:pointer;text-align:center;
  user-select:none;min-height:var(--key-h);
  display:flex;align-items:center;justify-content:center;
  font-family:inherit;transition:transform .05s}
.act.done{background:var(--blue);color:#fff;border-color:var(--blue)}
.act.active-mode{background:var(--blue);color:#fff;border-color:var(--blue)}
.act:active{transform:scale(.97)}

.ic{display:inline-flex;align-items:center;justify-content:center}
.ic-frac{flex-direction:column;gap:1px}
.ic-frac i{display:block;width:12px;height:5px;
  border:1px solid #7b8ba6;border-radius:2px}
.ic-frac i:nth-child(2){width:16px;height:1px;border:none;background:#7b8ba6;border-radius:0}
.ic-exp{position:relative;width:20px;height:18px}
.ic-exp i:nth-child(1){position:absolute;bottom:0;left:0;
  width:12px;height:12px;border:1px solid #7b8ba6;border-radius:2px}
.ic-exp i:nth-child(2){position:absolute;top:0;right:0;
  width:7px;height:7px;border:1px solid #7b8ba6;border-radius:2px}
.ic-sqrt{font-size:14px;font-weight:700;line-height:1}
.ic-sqrt i{display:inline-block;width:8px;height:10px;
  border-top:1px solid #7b8ba6;border-right:1px solid #7b8ba6;
  border-bottom:1px solid #7b8ba6;margin-left:-1px}
.ic-sum{flex-direction:column;font-size:11px;line-height:1}
.ic-sum i{display:block;width:12px;height:4px;
  border:1px solid #7b8ba6;border-radius:1px;margin:1px 0}
.ic-log{font-size:10px;font-weight:700}
.ic-log i{display:inline-block;width:7px;height:5px;
  border:1px solid #7b8ba6;border-radius:2px;margin-left:1px}

.slot{display:inline-block;min-width:10px;min-height:16px;
  padding:0 2px;vertical-align:middle;text-align:center;outline:none}
.slot:focus{background:#e8efff;border-radius:3px;
  box-shadow:0 0 0 1px rgba(59,91,246,.35)}

.mq-frac{display:inline-flex;flex-direction:column;align-items:center;
  vertical-align:middle;padding:0 3px;margin:0 2px;line-height:1.1}
.mq-frac .slot:first-child{border-bottom:1.6px solid var(--ink);
  min-width:16px;padding:1px 5px}
.mq-frac .slot:last-child{padding-top:2px;min-width:16px;
  padding-left:5px;padding-right:5px}

.mq-exp{display:inline-flex;align-items:flex-start;vertical-align:middle;padding:0 2px}
.mq-exp .slot{min-width:10px}
.mq-exp .mq-pow{font-size:.62em;position:relative;top:-.55em;
  padding-left:1px;min-width:10px}

.mq-sqrt{display:inline-flex;align-items:stretch;vertical-align:middle;
  line-height:1;padding:0 3px}
.mq-sqrt .mq-rad{border-top:1.6px solid var(--ink);
  border-right:1.6px solid var(--ink);border-bottom:1.6px solid var(--ink);
  min-width:14px;min-height:20px;padding:2px 5px;
  display:inline-flex;align-items:center}
.mq-sqrt .mq-idx{font-size:.62em;position:relative;top:-.55em;min-width:9px}

.mq-sum{display:inline-flex;flex-direction:column;align-items:center;
  vertical-align:middle;padding:0 5px;margin:0 2px;line-height:1}
.mq-sum .mq-top{font-size:.6em;min-width:20px;text-align:center;
  padding:1px 2px;min-height:12px}
.mq-sum .mq-sig{font-size:1.15em;line-height:1;margin:1px 0}
.mq-sum .mq-bot{font-size:.6em;min-width:20px;text-align:center;
  padding:1px 2px;min-height:12px}

.mq-log{display:inline-flex;align-items:flex-end;vertical-align:middle;padding:0 2px}
.mq-log .slot{font-size:.62em;min-width:10px;min-height:11px;text-align:center}

.mq-paren{display:inline-flex;align-items:center;padding:0 2px;
  font-family:'Cambria Math',Georgia,serif}
.mq-qf{display:inline-flex;align-items:center;padding:0 3px;
  font-family:'Cambria Math',Georgia,serif}

.result{padding:6px 12px;background:var(--ink);color:#fff;
  font-family:'SF Mono',Menlo,monospace;font-size:9px;line-height:1.4;
  max-height:0;overflow:hidden;transition:max-height .25s ease;flex:none}
.result.shown{max-height:180px;overflow-y:auto}
.result .label{color:#9ab;margin-bottom:3px}
.result pre{margin:0;color:#cfe2ff;white-space:pre-wrap}

@media (max-width:480px){
  .work{max-height:clamp(70px,14vh,110px)}
  .lbody{font-size:14px}
  .kb{padding:4px}
  .row{gap:3px}
  .key{min-height:32px;border-radius:6px}
  .act{font-size:9px;padding:4px 1px;min-height:32px}
  .topbar{padding:5px 10px}
  .tab{font-size:9px;padding:3px 8px}
}
</style></head><body>

<div class="app">

  <div class="topbar">
    <div class="logo">M</div>
    <div><div class="brand">MethodMarker</div>
    <div class="tag">Better Methods. Higher Marks.</div></div>
  </div>

  <div class="work">
    <div class="work-head">
      <span>&#9998; Working</span>
      <span class="hint">tap a line · ⏎ new step · ⌫ at start merges up</span>
    </div>
    <div class="lines" id="lines"></div>
  </div>

  <div class="sbc">
    <div class="sbc-head"><span class="dot"></span>
      Ask Tanaka — where did you get stuck?</div>
    <textarea id="sbc" rows="1"
      placeholder="e.g. I did step 2 but I don't understand why the sign flips"></textarea>
  </div>

  <div class="tabs" id="tabs"></div>
  <div class="kb-scroll"><div class="kb" id="kb"></div></div>
  <div class="acts" id="acts"></div>
  <div class="result" id="result"></div>
</div>

<script>
const LAYOUT = __LAYOUT__;
const state = { tab: LAYOUT.tabs_order[0], abcMode: false };

/* ═══════════ line editor ═══════════ */
const linesEl = document.getElementById('lines');

function addLine(html, focus){
  const row = document.createElement('div'); row.className='lrow';
  const n = document.createElement('div'); n.className='lnum';
  const b = document.createElement('div'); b.className='lbody';
  b.contentEditable='true'; b.spellcheck=false;
  if (html) b.innerHTML = html;
  row.appendChild(n); row.appendChild(b);
  linesEl.appendChild(row);
  renumber();
  if (focus) focusStart(b);
  return b;
}
function renumber(){
  [...linesEl.querySelectorAll('.lrow')].forEach((r,i)=>{
    r.querySelector('.lnum').textContent = i+1;
  });
}
function focusStart(el){
  el.focus();
  const first = el.querySelector('.slot');
  const r = document.createRange();
  if (first) r.selectNodeContents(first);
  else { r.selectNodeContents(el); r.collapse(false); }
  r.collapse(true);
  const s = window.getSelection(); s.removeAllRanges(); s.addRange(r);
}
function activeLine(){
  const sel = window.getSelection();
  if (!sel.rangeCount) return null;
  let n = sel.getRangeAt(0).startContainer;
  while (n && n !== linesEl){
    if (n.classList && n.classList.contains('lbody')) return n;
    n = n.parentNode;
  }
  return null;
}
function currentSlot(){
  const sel = window.getSelection();
  if (!sel.rangeCount) return null;
  let n = sel.getRangeAt(0).startContainer;
  if (n.nodeType === 3) n = n.parentNode;
  return n.closest ? n.closest('.slot') : null;
}

/* ═══════════ backspace — direct, callable from key or button ═══════════ */
function doBackspace(){
  let line = activeLine();
  if (!line){
    const all = linesEl.querySelectorAll('.lbody');
    if (all.length){
      line = all[all.length-1];
      line.focus();
      const r = document.createRange();
      r.selectNodeContents(line); r.collapse(false);
      const s = window.getSelection();
      s.removeAllRanges(); s.addRange(r);
    } else {
      return;
    }
  }

  const sel = window.getSelection();
  if (!sel.rangeCount) return;
  const r = sel.getRangeAt(0);

  if (!r.collapsed){
    r.deleteContents();
    return;
  }

  const slot = currentSlot();
  if (slot){
    const slotText = slot.textContent.replace(/\u200B/g,'');
    const probe = r.cloneRange(); probe.selectNodeContents(slot);
    probe.setEnd(r.startContainer, r.startOffset);
    const before = probe.toString().replace(/\u200B/g,'');
    if (slotText.length === 0 || before.length === 0){
      const struct = slot.closest(
        '.mq-frac,.mq-exp,.mq-sqrt,.mq-sum,.mq-log,.mq-paren,.mq-qf'
      ) || slot;
      const after = document.createRange();
      after.setStartAfter(struct); after.collapse(true);
      struct.remove();
      sel.removeAllRanges(); sel.addRange(after);
      return;
    }
    deleteCharBefore(r);
    return;
  }

  const probe = r.cloneRange(); probe.selectNodeContents(line);
  probe.setEnd(r.startContainer, r.startOffset);
  if (probe.toString().length === 0){
    const prev = line.closest('.lrow').previousElementSibling;
    if (prev){
      const pb = prev.querySelector('.lbody');
      while (line.firstChild) pb.appendChild(line.firstChild);
      line.closest('.lrow').remove();
      focusStart(pb);
      renumber();
      return;
    }
  }
  deleteCharBefore(r);
}

function deleteCharBefore(r){
  const node = r.startContainer;
  const offset = r.startOffset;
  const sel = window.getSelection();

  if (node.nodeType === 3 && offset > 0){
    node.deleteData(offset - 1, 1);
    const nr = document.createRange();
    nr.setStart(node, offset - 1); nr.collapse(true);
    sel.removeAllRanges(); sel.addRange(nr);
    return;
  }

  if (node.nodeType === 3 && offset === 0){
    let prev = node.previousSibling;
    while (prev && prev.nodeType !== 3 && !prev.textContent){
      prev = prev.previousSibling;
    }
    if (prev){
      const after = document.createRange();
      after.setStartAfter(prev); after.collapse(true);
      prev.remove();
      sel.removeAllRanges(); sel.addRange(after);
    }
    return;
  }

  if (node.nodeType === 1){
    let leaf = node;
    while (leaf.lastChild && leaf.nodeType === 1) leaf = leaf.lastChild;
    if (leaf.nodeType === 3){
      const nr = document.createRange();
      nr.setStart(leaf, Math.max(0, leaf.textContent.length - 1));
      nr.collapse(true);
      sel.removeAllRanges(); sel.addRange(nr);
      doBackspace();
    } else if (leaf.previousSibling){
      const after = document.createRange();
      after.setStartAfter(leaf.previousSibling); after.collapse(true);
      leaf.previousSibling.remove();
      sel.removeAllRanges(); sel.addRange(after);
    }
  }
}

/* ═══════════ line keydown ═══════════ */
linesEl.addEventListener('keydown', e => {
  const line = e.target.closest('.lbody'); if (!line) return;

  if (e.key === 'Enter' && !e.shiftKey){
    e.preventDefault();
    const sel = window.getSelection();
    const r = sel.getRangeAt(0);
    const after = r.cloneRange();
    after.selectNodeContents(line);
    after.setStart(r.endContainer, r.endOffset);
    const rest = after.extractContents();
    const nb = addLine('', false);
    while (rest.firstChild) nb.insertBefore(rest.firstChild, nb.firstChild);
    focusStart(nb);
    return;
  }

  if (e.key === 'Tab'){
    e.preventDefault();
    const slot = currentSlot();
    if (!slot){
      const r = document.createRange();
      r.selectNodeContents(line); r.collapse(false);
      const sel = window.getSelection();
      sel.removeAllRanges(); sel.addRange(r);
      return;
    }
    const struct = slot.closest('.mq-frac,.mq-exp,.mq-sqrt,.mq-sum,.mq-log,.mq-paren,.mq-qf');
    if (!struct) return;
    const slots = [...struct.querySelectorAll('.slot')];
    const i = slots.indexOf(slot);
    const next = slots[i+1];
    const sel = window.getSelection();
    if (next){
      const r = document.createRange(); r.selectNodeContents(next); r.collapse(true);
      sel.removeAllRanges(); sel.addRange(r);
    } else {
      const r = document.createRange(); r.setStartAfter(struct); r.collapse(true);
      sel.removeAllRanges(); sel.addRange(r);
    }
    return;
  }

  if (e.key === 'ArrowRight'){
    const slot = currentSlot();
    if (!slot) return;
    const struct = slot.closest('.mq-frac,.mq-exp,.mq-sqrt,.mq-sum,.mq-log,.mq-paren,.mq-qf');
    if (!struct) return;
    const slots = [...struct.querySelectorAll('.slot')];
    if (slots.indexOf(slot) !== slots.length - 1) return;
    e.preventDefault();
    const r = document.createRange();
    r.setStartAfter(struct); r.collapse(true);
    const sel = window.getSelection();
    sel.removeAllRanges(); sel.addRange(r);
    return;
  }

  if (e.key === 'Backspace'){
    e.preventDefault();
    doBackspace();
  }
});

/* ═══════════ insertion ═══════════ */
function insertAtCursor(html){
  let line = activeLine();
  if (!line){
    const all = linesEl.querySelectorAll('.lbody');
    line = all.length ? all[all.length-1] : addLine('', true);
    line.focus();
  }
  line.focus();

  const sel = window.getSelection();
  if (!sel.rangeCount) return;
  const r = sel.getRangeAt(0);
  r.deleteContents();

  const tmp = document.createElement('span');
  tmp.innerHTML = html;
  const nodes = [...tmp.childNodes];
  if (!nodes.length) return;

  const inside = currentSlot();
  if (inside){
    while (tmp.firstChild) inside.appendChild(tmp.firstChild);
    const nested = inside.querySelector('.slot.cur') || inside.querySelector('.slot');
    if (nested){
      const nr = document.createRange(); nr.selectNodeContents(nested); nr.collapse(true);
      sel.removeAllRanges(); sel.addRange(nr);
    }
    return;
  }

  nodes.forEach(n => r.insertNode(n));
  const last = nodes[nodes.length - 1];

  let target = null;
  if (last.nodeType === 1){
    target = last.querySelector('.slot.cur') || last.querySelector('.slot');
  }
  if (target){
    const nr = document.createRange();
    nr.selectNodeContents(target); nr.collapse(true);
    sel.removeAllRanges(); sel.addRange(nr);
  } else {
    const nr = document.createRange();
    nr.setStartAfter(last); nr.collapse(true);
    sel.removeAllRanges(); sel.addRange(nr);
  }
}

/* ═══════════ key handling ═══════════ */
function handleKey(key){
  if (key.i === '__abc__'){
    state.abcMode = !state.abcMode;
    document.querySelectorAll('.act').forEach(a => {
      if (a.textContent === 'ABC') a.classList.toggle('active-mode', state.abcMode);
    });
    renderKb();
    return;
  }
  if (key.i === '__del__'){ doBackspace(); return; }
  if (key.i === '__clear__') { linesEl.innerHTML=''; addLine('', true); return; }
  if (key.i === '__enter__') { addLine('', true); return; }
  if (key.i === '__ask__')   { document.getElementById('sbc').focus(); return; }
  if (key.i === '__done__')  { submit(); return; }
  insertAtCursor(key.i);
}

/* ═══════════ tabs ═══════════ */
const tabsEl = document.getElementById('tabs');
LAYOUT.tabs_order.forEach(name => {
  const b = document.createElement('button');
  b.className = 'tab' + (name === state.tab ? ' active' : '');
  b.textContent = name;
  b.onclick = () => {
    state.tab = name;
    state.abcMode = false;
    document.querySelectorAll('.tab').forEach(t =>
      t.classList.toggle('active', t.textContent === name));
    document.querySelectorAll('.act').forEach(a => {
      if (a.textContent === 'ABC') a.classList.remove('active-mode');
    });
    renderKb();
  };
  tabsEl.appendChild(b);
});

/* ═══════════ key grid ═══════════ */
const kbEl = document.getElementById('kb');
function keyBtn(key){
  const el = document.createElement('div');
  el.className = 'key';
  if (key.ic) el.innerHTML = key.ic;
  else el.textContent = key.l;
  el.onclick = () => handleKey(key);
  return el;
}
function renderRow(keys){
  const r = document.createElement('div'); r.className='row';
  r.style.gridTemplateColumns = `repeat(${keys.length},1fr)`;
  keys.forEach(k => r.appendChild(keyBtn(k)));
  kbEl.appendChild(r);
}
function renderKb(){
  kbEl.innerHTML = '';
  if (state.abcMode){
    (LAYOUT.base.abc || []).forEach(row => renderRow(row));
    return;
  }
  renderRow(LAYOUT.base.structure);
  renderRow(LAYOUT.base.operators);
  /* digits is now a flat list — render as one horizontal row */
  renderRow(LAYOUT.base.digits);
  renderRow(LAYOUT.base.variables);
  const rows = LAYOUT.tabs[state.tab] || [];
  rows.forEach(row => renderRow(row));
}

/* ═══════════ actions ═══════════ */
const actsEl = document.getElementById('acts');
LAYOUT.base.actions.forEach(k => {
  const el = document.createElement('div');
  el.className = 'act' + (k.i === '__done__' ? ' done' : '');
  el.textContent = k.l;
  el.onclick = () => handleKey(k);
  actsEl.appendChild(el);
});

/* ═══════════ submit ═══════════ */
const resultEl = document.getElementById('result');
function submit(){
  const steps = [...linesEl.querySelectorAll('.lbody')]
    .map(b => b.innerText.replace(/\u200B/g,'').trim())
    .filter(s => s.length);
  const payload = {
    steps:        steps,
    step_count:   steps.length,
    sbc_message:  document.getElementById('sbc').value.trim(),
    tab_used:     state.tab,
  };
  resultEl.classList.add('shown');
  resultEl.innerHTML =
    "<div class='label'>submitted to step_grader + intervention log</div>"+
    "<pre>"+JSON.stringify(payload,null,2)+"</pre>";
}

/* ═══════════ boot ═══════════ */
addLine('', true);
renderKb();
</script>
</body></html>
"""

html = HTML_TPL.replace("__LAYOUT__", json.dumps(payload, ensure_ascii=False))
HTML_PATH.write_text(html, encoding="utf-8")
print(f"  ✓ {HTML_PATH.name} written ({HTML_PATH.stat().st_size:,} bytes)")

webbrowser.open(HTML_PATH.as_uri())
print("  → opened in browser (fully offline)")
print()
print("  v15:")
print("    · Digits in ONE horizontal row: 1 2 3 4 5 6 7 8 9 0 . (−)")
print("    · Saves 3 keyboard rows → no scrolling needed on typical screens")
print("    · DEL calls doBackspace() directly · works even outside a line")
````

````python
# ══════════════════════════════════════════════════════════════════
# MethodMarker — keyboard v15
# Horizontal digits row · no scrolling · DEL direct · all keys visible
# ══════════════════════════════════════════════════════════════════
import json, webbrowser
from pathlib import Path

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
LAYOUT_PATH = ROOT / "data" / "processed" / "tutor" / "keyboard_layout.json"
HTML_PATH   = ROOT / "docs" / "keyboard_preview.html"

def k(label, ins=None, icon=None):
    d = {"l": label, "i": ins if ins is not None else label}
    if icon: d["ic"] = icon
    return d

def FRAC(v1=None, v2=None):
    a = v1 if v1 is not None else ""
    b = v2 if v2 is not None else ""
    return f"<span class='mq-frac'><span class='slot cur'>{a}</span><span class='slot'>{b}</span></span>"

def EXP(base=None, power=None):
    b = base if base is not None else ""
    p = power if power is not None else ""
    return (f"<span class='mq-exp'>"
            f"<span class='slot cur'>{b}</span>"
            f"<span class='slot mq-pow'>{p}</span>"
            f"</span>")

def SQRT(inner=None, idx=None):
    i = inner if inner is not None else ""
    if idx is None:
        return f"<span class='mq-sqrt'>√<span class='slot cur mq-rad'>{i}</span></span>"
    return (f"<span class='mq-sqrt'><span class='slot mq-idx'>{idx}</span>"
            f"√<span class='slot cur mq-rad'>{i}</span></span>")

def SUM(top=None, bot=None):
    t = top if top is not None else ""
    b = bot if bot is not None else ""
    return (f"<span class='mq-sum'>"
            f"<span class='slot mq-top cur'>{t}</span>"
            f"<span class='mq-sig'>Σ</span>"
            f"<span class='slot mq-bot'>{b}</span>"
            f"</span>")

def LOG(base=None):
    b = base if base is not None else ""
    return f"<span class='mq-log'>log<span class='slot cur'>{b}</span>(</span>"

def PAREN(preset=None, second=None):
    p = preset if preset is not None else ""
    s = second if second is not None else ""
    return (f"<span class='mq-paren'>(<span class='slot cur'>{p}</span>"
            f"±<span class='slot'>{s}</span>)</span>")

def PAIR(x=None, y=None):
    a = x if x is not None else ""
    b = y if y is not None else ""
    return f"(<span class='slot cur'>{a}</span>;<span class='slot'>{b}</span>)"

def LOGDIV():
    return ("<span class='mq-frac'>"
            "<span class='slot cur'>sin(□)</span>"
            "<span class='slot'>cos(□)</span>"
            "</span>")

IC_FRAC = "<span class='ic ic-frac'><i></i><i></i></span>"
IC_EXP  = "<span class='ic ic-exp'><i></i><i></i></span>"
IC_SQRT = "<span class='ic ic-sqrt'>&#8730;<i></i></span>"
IC_SUM  = "<span class='ic ic-sum'><i></i>&#8721;<i></i></span>"
IC_LOG  = "<span class='ic ic-log'>log<i></i></span>"

# ══════════════════════════════════════════════════════════
# BASE LAYER
# ══════════════════════════════════════════════════════════
base = {
    "structure": [
        k("a/b",  FRAC(),  IC_FRAC),
        k("□/□",  FRAC()),
        k("x²",   EXP(power="2"),  IC_EXP),
        k("xⁿ",   EXP()),
        k("√",    SQRT(),  IC_SQRT),
        k("ⁿ√",   SQRT(idx="")),
        k("∛",    "<span class='mq-sqrt'>∛<span class='slot cur mq-rad'></span></span>"),
        k("log",  LOG()),
        k("logₐ", LOG(base="")),
        k("π",    "π"),
        k("θ",    "θ"),
        k("∞",    "∞"),
    ],
    "operators": [
        k("+", "+"), k("−", "−"), k("×", "×"), k("÷", "÷"),
        k("=", "="), k("(", "("), k(")", ")"),
        k("<", "<"), k(">", ">"),
        k("≤", "≤"), k("≥", "≥"), k("≠", "≠"),
        k("∴", " ∴ "), k("∵", " ∵ "), k("⇒", " ⇒ "),
    ],
    # ONE horizontal row of digits
    "digits": [
        k("1"), k("2"), k("3"), k("4"), k("5"), k("6"),
        k("7"), k("8"), k("9"), k("0"), k("."), k("(−)", "−"),
    ],
    "variables": [
        k("x"), k("y"), k("a"), k("b"), k("c"), k("n"), k("k"), k("t"),
    ],
    "actions": [
        k("DEL", "__del__"),
        k("AC",  "__clear__"),
        k("ABC", "__abc__"),
        k("⏎",   "__enter__"),
        k("?",   "__ask__"),
        k("✓",   "__done__"),
    ],
    "abc": [
        [k(L) for L in "ABCDEFGHIJ"],
        [k(L) for L in "KLMNOPQRST"],
        [k(L) for L in "UVWXYZ"] + [k(""), k(""), k("")],
        [k("x₁","x₁"), k("y₁","y₁"), k("x₂","x₂"), k("y₂","y₂"),
         k("p","p"), k("q","q"), k("m","m"), k("r","r"), k("s","s"), k("u","u")],
    ],
}

# ══════════════════════════════════════════════════════════
# TABS
# ══════════════════════════════════════════════════════════
tabs_order = ["Algebra","Sequences","Functions","Calculus","Trig",
              "Finance","Analytical","Euclidean","Stats","Probability"]

tabs = {
    "Algebra": [
        [k("x²",EXP(power="2")), k("y²",EXP(power="2")),
         k("a²",EXP(power="2")), k("b²",EXP(power="2")),
         k("c²",EXP(power="2"))],
        [k("x³",EXP(power="3")), k("xⁿ",EXP()), k("aⁿ",EXP()), k("bⁿ",EXP())],
        [k("(□)²",EXP(base="()",power="2")),
         k("(□)³",EXP(base="()",power="3")),
         k("(□)ⁿ",EXP(base="()"))],
        [k("±","±"),
         k("QF","<span class='mq-qf'>x = "
              "<span class='mq-frac'><span class='slot cur'></span>"
              "<span class='slot'></span></span></span>")],
    ],
    "Sequences": [
        [k("Σ",SUM(),IC_SUM),
         k("Tₙ",EXP(base="T",power="n")),
         k("Sₙ",EXP(base="S",power="n")),
         k("Tₙ₊₁","Tₙ₊₁"), k("Tₙ₋₁","Tₙ₋₁"), k("n→∞","n→∞")],
        [k("□²",EXP(power="2")), k("□³",EXP(power="3")), k("□ⁿ",EXP()),
         k("(□)²",EXP(base="()",power="2")),
         k("(□)³",EXP(base="()",power="3")),
         k("(□)ⁿ",EXP(base="()"))],
        [k("A"), k("P"), k("i"), k("n"), k("r")],
    ],
    "Functions": [
        [k("f(□)","f(<span class='slot cur'></span>)"),
         k("g(□)","g(<span class='slot cur'></span>)"),
         k("f⁻¹(□)","f⁻¹(<span class='slot cur'></span>)"),
         k("→","→"), k("∞","∞"), k("Δ","Δ")],
        [k("sin(□)","sin(<span class='slot cur'></span>)"),
         k("cos(□)","cos(<span class='slot cur'></span>)"),
         k("tan(□)","tan(<span class='slot cur'></span>)")],
        [k("a/b",FRAC()), k("□ⁿ",EXP())],
    ],
    "Calculus": [
        [k("dy/dx","dy/dx"), k("d²y/dx²","d²y/dx²"),
         k("dⁿy/dxⁿ","dⁿy/dxⁿ"), k("∫","∫"), k("lim","lim"), k("→","→")],
        [k("□²",EXP(power="2")), k("□³",EXP(power="3")), k("□ⁿ",EXP()),
         k("Δx","Δx"),
         k("1/□","<span class='mq-frac'><span class='slot cur'>1</span>"
                 "<span class='slot'></span></span>")],
    ],
    "Trig": [
        [k("sin(","sin("), k("cos(","cos("), k("tan(","tan("),
         k("sin⁻¹(","sin⁻¹("), k("cos⁻¹(","cos⁻¹("), k("tan⁻¹(","tan⁻¹(")],
        [k("α","α"), k("β","β"), k("θ","θ"), k("2α","2α"), k("°","°"), k("π","π")],
        [k("(□±□)",PAREN()), k("(90°±θ)",PAREN(preset="90°",second="θ")),
         k("(□±θ)",PAREN(second="θ")),
         k("(180°±θ)",PAREN(preset="180°",second="θ")),
         k("(360°±θ)",PAREN(preset="360°",second="θ")),
         k("(−θ)","(−θ)")],
        [k("sin²θ",EXP(base="sin θ",power="2")),
         k("sinⁿ□",EXP(base="sin ")),
         k("cos²θ",EXP(base="cos θ",power="2")),
         k("cosⁿ□",EXP(base="cos "))],
        [k("tan²θ",EXP(base="tan θ",power="2")),
         k("tanⁿ□",EXP(base="tan ")),
         k("sin□/cos□",LOGDIV())],
    ],
    "Finance": [
        [k("A"), k("P"), k("i"), k("n"), k("x")],
        [k("(1+i)ⁿ",EXP(base="1+i")), k("(□)ⁿ",EXP(base="()")),
         k("(1−i)ⁿ",EXP(base="1−i"))],
        [k("□ⁿ",EXP()), k("□/□",FRAC())],
    ],
    "Analytical": [
        [k("(□;□)",PAIR()), k("x₁","x₁"), k("y₁","y₁"), k("x₂","x₂"), k("y₂","y₂")],
        [k("√",SQRT()), k("□²",EXP(power="2")),
         k("tan(□)","tan(<span class='slot cur'></span>)")],
        [k("θ","θ"), k("°","°")],
        [k("(□)²+(□)²",
           "<span class='mq-exp'><span class='slot cur'></span>"
           "<span class='slot mq-pow'>2</span></span>+"
           "<span class='mq-exp'><span class='slot'></span>"
           "<span class='slot mq-pow'>2</span></span>")],
    ],
    "Euclidean": [
        [k("△","△"), k("∥","∥"), k("⊥","⊥"),
         k("∠","∠"), k("≅","≅"), k("∼","∼")],
        [k("Â","Â"), k("B̂","B̂"), k("Ĉ","Ĉ"), k("D̂","D̂")],
        [k("sin(","sin("), k("cos(","cos("), k("tan(","tan(")],
        [k("√",SQRT()), k("∴"," ∴ "), k("∵"," ∵ ")],
    ],
    "Stats": [
        [k("x̄","x̄"), k("σ","σ"), k("n","n"), k("r","r"),
         k("r²",EXP(base="r",power="2"))],
        [k("Σ",SUM()), k("□²",EXP(power="2")), k("√",SQRT())],
        [k("E(□)","E(<span class='slot cur'></span>)"),
         k("P(□)","P(<span class='slot cur'></span>)")],
    ],
    "Probability": [
        [k("P(□)","P(<span class='slot cur'></span>)"),
         k("P(□∪□)","P(<span class='slot cur'></span>∪<span class='slot'></span>)"),
         k("P(□∩□)","P(<span class='slot cur'></span>∩<span class='slot'></span>)"),
         k("P(□')","P(<span class='slot cur'></span>')")],
        [k("∪","∪"), k("∩","∩"), k("'","'")],
        [k("nCr","nCr"), k("nPr","nPr"), k("!","!")],
    ],
}

payload = {"schema_version": 15.0, "base": base,
           "tabs_order": tabs_order, "tabs": tabs}
LAYOUT_PATH.write_text(json.dumps(payload, indent=2, ensure_ascii=False),
                       encoding="utf-8")
print(f"  ✓ {LAYOUT_PATH.name} written")

# ══════════════════════════════════════════════════════════
# HTML
# ══════════════════════════════════════════════════════════
HTML_TPL = r"""<!doctype html>
<html lang="en"><head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width,initial-scale=1,maximum-scale=1">
<title>MethodMarker Keyboard</title>
<style>
:root{
  --bg:#eef2f8; --card:#fff; --line:#e4e9f2; --ink:#0b1f44;
  --ink2:#465a80; --ink3:#9aa8c2; --blue:#3b5bf6; --blue2:#5b8cf6;
  --amber:#f0ad4e;
  --key-h: clamp(30px, 5.2vw, 40px);
  --key-f: clamp(10px, 1.4vw, 13px);
  --pad: clamp(4px, 0.9vw, 8px);
}
*{box-sizing:border-box;-webkit-tap-highlight-color:transparent}
html,body{margin:0;height:100%}
body{
  background:var(--bg);color:var(--ink);
  font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,sans-serif;
  display:flex;justify-content:center;padding:6px;
  min-height:100vh;overflow:hidden;
}
.app{
  width:100%;max-width:720px;background:var(--card);
  border-radius:16px;overflow:hidden;
  box-shadow:0 8px 32px rgba(20,40,90,.12);
  display:flex;flex-direction:column;
  height:calc(100vh - 12px);
}
.topbar{display:flex;align-items:center;gap:8px;
  padding:6px 12px;background:#f6f8fd;border-bottom:1px solid var(--line);flex:none}
.logo{width:22px;height:22px;border-radius:6px;
  background:linear-gradient(135deg,var(--blue),var(--blue2));
  display:flex;align-items:center;justify-content:center;
  color:#fff;font-weight:700;font-size:12px;flex:none}
.brand{font-weight:700;font-size:12px;color:var(--ink)}
.tag{font-size:9px;color:#6b7a99}

.work{background:#fff;padding:6px 12px 8px;flex:none;
  max-height:clamp(80px,16vh,140px);overflow-y:auto}
.work-head{display:flex;justify-content:space-between;align-items:center;
  font-size:10px;color:var(--ink2);font-weight:600;margin-bottom:4px;
  position:sticky;top:0;background:#fff;padding-bottom:2px}
.work-head .hint{color:var(--ink3);font-weight:500;font-size:9px}
.lines{display:flex;flex-direction:column;gap:1px}
.lrow{display:grid;grid-template-columns:20px 1fr;column-gap:5px;align-items:flex-start}
.lnum{color:#b8c2d6;font-size:9px;text-align:right;padding-top:7px;
  user-select:none;font-variant-numeric:tabular-nums}
.lbody{min-height:26px;padding:4px 6px;border-radius:5px;
  font-size:clamp(14px,1.9vw,17px);line-height:1.35;color:var(--ink);
  outline:none;font-family:'Cambria Math','Latin Modern Math',Georgia,serif;
  white-space:pre-wrap;word-break:break-word;
  transition:background .1s;caret-color:var(--blue)}
.lbody:focus{background:#f5f8ff}

.sbc{padding:5px 12px 7px;background:#fff;border-top:1px dashed var(--line);flex:none}
.sbc-head{display:flex;align-items:center;gap:5px;font-size:10px;
  color:var(--ink2);font-weight:600;margin-bottom:4px}
.sbc-head .dot{width:5px;height:5px;border-radius:50%;background:var(--amber)}
#sbc{width:100%;height:26px;resize:none;font-family:inherit;
  font-size:11px;padding:5px 8px;border:1px solid var(--line);
  border-radius:6px;background:#fffdf7;color:var(--ink);outline:none;
  transition:height .2s ease,box-shadow .15s}
#sbc:focus{height:60px;box-shadow:0 0 0 3px rgba(240,173,78,.15)}

.tabs{display:flex;overflow-x:auto;gap:3px;padding:4px 8px;
  background:#f6f8fd;border-top:1px solid var(--line);
  border-bottom:1px solid var(--line);scrollbar-width:none;flex:none}
.tabs::-webkit-scrollbar{display:none}
.tab{flex:none;padding:4px 9px;border-radius:12px;font-size:10px;
  font-weight:600;color:var(--ink2);background:transparent;
  border:none;cursor:pointer;white-space:nowrap;font-family:inherit}
.tab.active{background:var(--blue);color:#fff}

.kb-scroll{flex:1 1 auto;overflow-y:auto;background:#fff;display:flex;flex-direction:column}
.kb{padding:var(--pad);display:flex;flex-direction:column;gap:4px}
.row{display:grid;gap:4px}
.key{padding:2px;border-radius:7px;background:#fff;
  border:1px solid #dfe5ef;color:var(--ink);
  font-size:var(--key-f);cursor:pointer;text-align:center;
  user-select:none;font-family:inherit;min-height:var(--key-h);
  display:flex;align-items:center;justify-content:center;
  transition:transform .05s,background .1s;overflow:hidden}
.key:active{transform:scale(.95);background:#eef2fb}
.acts{display:grid;grid-template-columns:repeat(6,1fr);gap:4px;
  padding:5px var(--pad) 6px;background:#f6f8fd;
  border-top:1px solid var(--line);flex:none}
.act{padding:6px 2px;border-radius:7px;background:#eef2fb;
  border:1px solid #dfe5ef;color:#223a6b;
  font-size:10px;font-weight:700;cursor:pointer;text-align:center;
  user-select:none;min-height:var(--key-h);
  display:flex;align-items:center;justify-content:center;
  font-family:inherit;transition:transform .05s}
.act.done{background:var(--blue);color:#fff;border-color:var(--blue)}
.act.active-mode{background:var(--blue);color:#fff;border-color:var(--blue)}
.act:active{transform:scale(.97)}

.ic{display:inline-flex;align-items:center;justify-content:center}
.ic-frac{flex-direction:column;gap:1px}
.ic-frac i{display:block;width:12px;height:5px;
  border:1px solid #7b8ba6;border-radius:2px}
.ic-frac i:nth-child(2){width:16px;height:1px;border:none;background:#7b8ba6;border-radius:0}
.ic-exp{position:relative;width:20px;height:18px}
.ic-exp i:nth-child(1){position:absolute;bottom:0;left:0;
  width:12px;height:12px;border:1px solid #7b8ba6;border-radius:2px}
.ic-exp i:nth-child(2){position:absolute;top:0;right:0;
  width:7px;height:7px;border:1px solid #7b8ba6;border-radius:2px}
.ic-sqrt{font-size:14px;font-weight:700;line-height:1}
.ic-sqrt i{display:inline-block;width:8px;height:10px;
  border-top:1px solid #7b8ba6;border-right:1px solid #7b8ba6;
  border-bottom:1px solid #7b8ba6;margin-left:-1px}
.ic-sum{flex-direction:column;font-size:11px;line-height:1}
.ic-sum i{display:block;width:12px;height:4px;
  border:1px solid #7b8ba6;border-radius:1px;margin:1px 0}
.ic-log{font-size:10px;font-weight:700}
.ic-log i{display:inline-block;width:7px;height:5px;
  border:1px solid #7b8ba6;border-radius:2px;margin-left:1px}

.slot{display:inline-block;min-width:10px;min-height:16px;
  padding:0 2px;vertical-align:middle;text-align:center;outline:none}
.slot:focus{background:#e8efff;border-radius:3px;
  box-shadow:0 0 0 1px rgba(59,91,246,.35)}

.mq-frac{display:inline-flex;flex-direction:column;align-items:center;
  vertical-align:middle;padding:0 3px;margin:0 2px;line-height:1.1}
.mq-frac .slot:first-child{border-bottom:1.6px solid var(--ink);
  min-width:16px;padding:1px 5px}
.mq-frac .slot:last-child{padding-top:2px;min-width:16px;
  padding-left:5px;padding-right:5px}

.mq-exp{display:inline-flex;align-items:flex-start;vertical-align:middle;padding:0 2px}
.mq-exp .slot{min-width:10px}
.mq-exp .mq-pow{font-size:.62em;position:relative;top:-.55em;
  padding-left:1px;min-width:10px}

.mq-sqrt{display:inline-flex;align-items:stretch;vertical-align:middle;
  line-height:1;padding:0 3px}
.mq-sqrt .mq-rad{border-top:1.6px solid var(--ink);
  border-right:1.6px solid var(--ink);border-bottom:1.6px solid var(--ink);
  min-width:14px;min-height:20px;padding:2px 5px;
  display:inline-flex;align-items:center}
.mq-sqrt .mq-idx{font-size:.62em;position:relative;top:-.55em;min-width:9px}

.mq-sum{display:inline-flex;flex-direction:column;align-items:center;
  vertical-align:middle;padding:0 5px;margin:0 2px;line-height:1}
.mq-sum .mq-top{font-size:.6em;min-width:20px;text-align:center;
  padding:1px 2px;min-height:12px}
.mq-sum .mq-sig{font-size:1.15em;line-height:1;margin:1px 0}
.mq-sum .mq-bot{font-size:.6em;min-width:20px;text-align:center;
  padding:1px 2px;min-height:12px}

.mq-log{display:inline-flex;align-items:flex-end;vertical-align:middle;padding:0 2px}
.mq-log .slot{font-size:.62em;min-width:10px;min-height:11px;text-align:center}

.mq-paren{display:inline-flex;align-items:center;padding:0 2px;
  font-family:'Cambria Math',Georgia,serif}
.mq-qf{display:inline-flex;align-items:center;padding:0 3px;
  font-family:'Cambria Math',Georgia,serif}

.result{padding:6px 12px;background:var(--ink);color:#fff;
  font-family:'SF Mono',Menlo,monospace;font-size:9px;line-height:1.4;
  max-height:0;overflow:hidden;transition:max-height .25s ease;flex:none}
.result.shown{max-height:180px;overflow-y:auto}
.result .label{color:#9ab;margin-bottom:3px}
.result pre{margin:0;color:#cfe2ff;white-space:pre-wrap}

@media (max-width:480px){
  .work{max-height:clamp(70px,14vh,110px)}
  .lbody{font-size:14px}
  .kb{padding:4px}
  .row{gap:3px}
  .key{min-height:32px;border-radius:6px}
  .act{font-size:9px;padding:4px 1px;min-height:32px}
  .topbar{padding:5px 10px}
  .tab{font-size:9px;padding:3px 8px}
}
</style></head><body>

<div class="app">

  <div class="topbar">
    <div class="logo">M</div>
    <div><div class="brand">MethodMarker</div>
    <div class="tag">Better Methods. Higher Marks.</div></div>
  </div>

  <div class="work">
    <div class="work-head">
      <span>&#9998; Working</span>
      <span class="hint">tap a line · ⏎ new step · ⌫ at start merges up</span>
    </div>
    <div class="lines" id="lines"></div>
  </div>

  <div class="sbc">
    <div class="sbc-head"><span class="dot"></span>
      Ask Tanaka — where did you get stuck?</div>
    <textarea id="sbc" rows="1"
      placeholder="e.g. I did step 2 but I don't understand why the sign flips"></textarea>
  </div>

  <div class="tabs" id="tabs"></div>
  <div class="kb-scroll"><div class="kb" id="kb"></div></div>
  <div class="acts" id="acts"></div>
  <div class="result" id="result"></div>
</div>

<script>
const LAYOUT = __LAYOUT__;
const state = { tab: LAYOUT.tabs_order[0], abcMode: false };

/* ═══════════ line editor ═══════════ */
const linesEl = document.getElementById('lines');

function addLine(html, focus){
  const row = document.createElement('div'); row.className='lrow';
  const n = document.createElement('div'); n.className='lnum';
  const b = document.createElement('div'); b.className='lbody';
  b.contentEditable='true'; b.spellcheck=false;
  if (html) b.innerHTML = html;
  row.appendChild(n); row.appendChild(b);
  linesEl.appendChild(row);
  renumber();
  if (focus) focusStart(b);
  return b;
}
function renumber(){
  [...linesEl.querySelectorAll('.lrow')].forEach((r,i)=>{
    r.querySelector('.lnum').textContent = i+1;
  });
}
function focusStart(el){
  el.focus();
  const first = el.querySelector('.slot');
  const r = document.createRange();
  if (first) r.selectNodeContents(first);
  else { r.selectNodeContents(el); r.collapse(false); }
  r.collapse(true);
  const s = window.getSelection(); s.removeAllRanges(); s.addRange(r);
}
function activeLine(){
  const sel = window.getSelection();
  if (!sel.rangeCount) return null;
  let n = sel.getRangeAt(0).startContainer;
  while (n && n !== linesEl){
    if (n.classList && n.classList.contains('lbody')) return n;
    n = n.parentNode;
  }
  return null;
}
function currentSlot(){
  const sel = window.getSelection();
  if (!sel.rangeCount) return null;
  let n = sel.getRangeAt(0).startContainer;
  if (n.nodeType === 3) n = n.parentNode;
  return n.closest ? n.closest('.slot') : null;
}

/* ═══════════ backspace — direct, callable from key or button ═══════════ */
function doBackspace(){
  let line = activeLine();
  if (!line){
    const all = linesEl.querySelectorAll('.lbody');
    if (all.length){
      line = all[all.length-1];
      line.focus();
      const r = document.createRange();
      r.selectNodeContents(line); r.collapse(false);
      const s = window.getSelection();
      s.removeAllRanges(); s.addRange(r);
    } else {
      return;
    }
  }

  const sel = window.getSelection();
  if (!sel.rangeCount) return;
  const r = sel.getRangeAt(0);

  if (!r.collapsed){
    r.deleteContents();
    return;
  }

  const slot = currentSlot();
  if (slot){
    const slotText = slot.textContent.replace(/\u200B/g,'');
    const probe = r.cloneRange(); probe.selectNodeContents(slot);
    probe.setEnd(r.startContainer, r.startOffset);
    const before = probe.toString().replace(/\u200B/g,'');
    if (slotText.length === 0 || before.length === 0){
      const struct = slot.closest(
        '.mq-frac,.mq-exp,.mq-sqrt,.mq-sum,.mq-log,.mq-paren,.mq-qf'
      ) || slot;
      const after = document.createRange();
      after.setStartAfter(struct); after.collapse(true);
      struct.remove();
      sel.removeAllRanges(); sel.addRange(after);
      return;
    }
    deleteCharBefore(r);
    return;
  }

  const probe = r.cloneRange(); probe.selectNodeContents(line);
  probe.setEnd(r.startContainer, r.startOffset);
  if (probe.toString().length === 0){
    const prev = line.closest('.lrow').previousElementSibling;
    if (prev){
      const pb = prev.querySelector('.lbody');
      while (line.firstChild) pb.appendChild(line.firstChild);
      line.closest('.lrow').remove();
      focusStart(pb);
      renumber();
      return;
    }
  }
  deleteCharBefore(r);
}

function deleteCharBefore(r){
  const node = r.startContainer;
  const offset = r.startOffset;
  const sel = window.getSelection();

  if (node.nodeType === 3 && offset > 0){
    node.deleteData(offset - 1, 1);
    const nr = document.createRange();
    nr.setStart(node, offset - 1); nr.collapse(true);
    sel.removeAllRanges(); sel.addRange(nr);
    return;
  }

  if (node.nodeType === 3 && offset === 0){
    let prev = node.previousSibling;
    while (prev && prev.nodeType !== 3 && !prev.textContent){
      prev = prev.previousSibling;
    }
    if (prev){
      const after = document.createRange();
      after.setStartAfter(prev); after.collapse(true);
      prev.remove();
      sel.removeAllRanges(); sel.addRange(after);
    }
    return;
  }

  if (node.nodeType === 1){
    let leaf = node;
    while (leaf.lastChild && leaf.nodeType === 1) leaf = leaf.lastChild;
    if (leaf.nodeType === 3){
      const nr = document.createRange();
      nr.setStart(leaf, Math.max(0, leaf.textContent.length - 1));
      nr.collapse(true);
      sel.removeAllRanges(); sel.addRange(nr);
      doBackspace();
    } else if (leaf.previousSibling){
      const after = document.createRange();
      after.setStartAfter(leaf.previousSibling); after.collapse(true);
      leaf.previousSibling.remove();
      sel.removeAllRanges(); sel.addRange(after);
    }
  }
}

/* ═══════════ line keydown ═══════════ */
linesEl.addEventListener('keydown', e => {
  const line = e.target.closest('.lbody'); if (!line) return;

  if (e.key === 'Enter' && !e.shiftKey){
    e.preventDefault();
    const sel = window.getSelection();
    const r = sel.getRangeAt(0);
    const after = r.cloneRange();
    after.selectNodeContents(line);
    after.setStart(r.endContainer, r.endOffset);
    const rest = after.extractContents();
    const nb = addLine('', false);
    while (rest.firstChild) nb.insertBefore(rest.firstChild, nb.firstChild);
    focusStart(nb);
    return;
  }

  if (e.key === 'Tab'){
    e.preventDefault();
    const slot = currentSlot();
    if (!slot){
      const r = document.createRange();
      r.selectNodeContents(line); r.collapse(false);
      const sel = window.getSelection();
      sel.removeAllRanges(); sel.addRange(r);
      return;
    }
    const struct = slot.closest('.mq-frac,.mq-exp,.mq-sqrt,.mq-sum,.mq-log,.mq-paren,.mq-qf');
    if (!struct) return;
    const slots = [...struct.querySelectorAll('.slot')];
    const i = slots.indexOf(slot);
    const next = slots[i+1];
    const sel = window.getSelection();
    if (next){
      const r = document.createRange(); r.selectNodeContents(next); r.collapse(true);
      sel.removeAllRanges(); sel.addRange(r);
    } else {
      const r = document.createRange(); r.setStartAfter(struct); r.collapse(true);
      sel.removeAllRanges(); sel.addRange(r);
    }
    return;
  }

  if (e.key === 'ArrowRight'){
    const slot = currentSlot();
    if (!slot) return;
    const struct = slot.closest('.mq-frac,.mq-exp,.mq-sqrt,.mq-sum,.mq-log,.mq-paren,.mq-qf');
    if (!struct) return;
    const slots = [...struct.querySelectorAll('.slot')];
    if (slots.indexOf(slot) !== slots.length - 1) return;
    e.preventDefault();
    const r = document.createRange();
    r.setStartAfter(struct); r.collapse(true);
    const sel = window.getSelection();
    sel.removeAllRanges(); sel.addRange(r);
    return;
  }

  if (e.key === 'Backspace'){
    e.preventDefault();
    doBackspace();
  }
});

/* ═══════════ insertion ═══════════ */
function insertAtCursor(html){
  let line = activeLine();
  if (!line){
    const all = linesEl.querySelectorAll('.lbody');
    line = all.length ? all[all.length-1] : addLine('', true);
    line.focus();
  }
  line.focus();

  const sel = window.getSelection();
  if (!sel.rangeCount) return;
  const r = sel.getRangeAt(0);
  r.deleteContents();

  const tmp = document.createElement('span');
  tmp.innerHTML = html;
  const nodes = [...tmp.childNodes];
  if (!nodes.length) return;

  const inside = currentSlot();
  if (inside){
    while (tmp.firstChild) inside.appendChild(tmp.firstChild);
    const nested = inside.querySelector('.slot.cur') || inside.querySelector('.slot');
    if (nested){
      const nr = document.createRange(); nr.selectNodeContents(nested); nr.collapse(true);
      sel.removeAllRanges(); sel.addRange(nr);
    }
    return;
  }

  nodes.forEach(n => r.insertNode(n));
  const last = nodes[nodes.length - 1];

  let target = null;
  if (last.nodeType === 1){
    target = last.querySelector('.slot.cur') || last.querySelector('.slot');
  }
  if (target){
    const nr = document.createRange();
    nr.selectNodeContents(target); nr.collapse(true);
    sel.removeAllRanges(); sel.addRange(nr);
  } else {
    const nr = document.createRange();
    nr.setStartAfter(last); nr.collapse(true);
    sel.removeAllRanges(); sel.addRange(nr);
  }
}

/* ═══════════ key handling ═══════════ */
function handleKey(key){
  if (key.i === '__abc__'){
    state.abcMode = !state.abcMode;
    document.querySelectorAll('.act').forEach(a => {
      if (a.textContent === 'ABC') a.classList.toggle('active-mode', state.abcMode);
    });
    renderKb();
    return;
  }
  if (key.i === '__del__'){ doBackspace(); return; }
  if (key.i === '__clear__') { linesEl.innerHTML=''; addLine('', true); return; }
  if (key.i === '__enter__') { addLine('', true); return; }
  if (key.i === '__ask__')   { document.getElementById('sbc').focus(); return; }
  if (key.i === '__done__')  { submit(); return; }
  insertAtCursor(key.i);
}

/* ═══════════ tabs ═══════════ */
const tabsEl = document.getElementById('tabs');
LAYOUT.tabs_order.forEach(name => {
  const b = document.createElement('button');
  b.className = 'tab' + (name === state.tab ? ' active' : '');
  b.textContent = name;
  b.onclick = () => {
    state.tab = name;
    state.abcMode = false;
    document.querySelectorAll('.tab').forEach(t =>
      t.classList.toggle('active', t.textContent === name));
    document.querySelectorAll('.act').forEach(a => {
      if (a.textContent === 'ABC') a.classList.remove('active-mode');
    });
    renderKb();
  };
  tabsEl.appendChild(b);
});

/* ═══════════ key grid ═══════════ */
const kbEl = document.getElementById('kb');
function keyBtn(key){
  const el = document.createElement('div');
  el.className = 'key';
  if (key.ic) el.innerHTML = key.ic;
  else el.textContent = key.l;
  el.onclick = () => handleKey(key);
  return el;
}
function renderRow(keys){
  const r = document.createElement('div'); r.className='row';
  r.style.gridTemplateColumns = `repeat(${keys.length},1fr)`;
  keys.forEach(k => r.appendChild(keyBtn(k)));
  kbEl.appendChild(r);
}
function renderKb(){
  kbEl.innerHTML = '';
  if (state.abcMode){
    (LAYOUT.base.abc || []).forEach(row => renderRow(row));
    return;
  }
  renderRow(LAYOUT.base.structure);
  renderRow(LAYOUT.base.operators);
  /* digits is now a flat list — render as one horizontal row */
  renderRow(LAYOUT.base.digits);
  renderRow(LAYOUT.base.variables);
  const rows = LAYOUT.tabs[state.tab] || [];
  rows.forEach(row => renderRow(row));
}

/* ═══════════ actions ═══════════ */
const actsEl = document.getElementById('acts');
LAYOUT.base.actions.forEach(k => {
  const el = document.createElement('div');
  el.className = 'act' + (k.i === '__done__' ? ' done' : '');
  el.textContent = k.l;
  el.onclick = () => handleKey(k);
  actsEl.appendChild(el);
});

/* ═══════════ submit ═══════════ */
const resultEl = document.getElementById('result');
function submit(){
  const steps = [...linesEl.querySelectorAll('.lbody')]
    .map(b => b.innerText.replace(/\u200B/g,'').trim())
    .filter(s => s.length);
  const payload = {
    steps:        steps,
    step_count:   steps.length,
    sbc_message:  document.getElementById('sbc').value.trim(),
    tab_used:     state.tab,
  };
  resultEl.classList.add('shown');
  resultEl.innerHTML =
    "<div class='label'>submitted to step_grader + intervention log</div>"+
    "<pre>"+JSON.stringify(payload,null,2)+"</pre>";
}

/* ═══════════ boot ═══════════ */
addLine('', true);
renderKb();
</script>
</body></html>
"""

html = HTML_TPL.replace("__LAYOUT__", json.dumps(payload, ensure_ascii=False))
HTML_PATH.write_text(html, encoding="utf-8")
print(f"  ✓ {HTML_PATH.name} written ({HTML_PATH.stat().st_size:,} bytes)")

webbrowser.open(HTML_PATH.as_uri())
print("  → opened in browser (fully offline)")
print()
print("  v15:")
print("    · Digits in ONE horizontal row: 1 2 3 4 5 6 7 8 9 0 . (−)")
print("    · Saves 3 keyboard rows → no scrolling needed on typical screens")
print("    · DEL calls doBackspace() directly · works even outside a line")
````

````python
# ══════════════════════════════════════════════════════════════════
# MethodMarker — Keyboard v18  (working, no experimental radical)
# ══════════════════════════════════════════════════════════════════
import json, webbrowser
from pathlib import Path

ROOT = Path(r"C:\Users\Administrator\Desktop\Matric-Maths-Exam-Intelligence")
LAYOUT_PATH = ROOT / "data" / "processed" / "tutor" / "keyboard_layout.json"
HTML_PATH   = ROOT / "docs" / "keyboard_preview.html"

def k(label, ins=None, icon=None):
    d = {"l": label, "i": ins if ins is not None else label}
    if icon: d["ic"] = icon
    return d

def FRAC(v1=None, v2=None):
    a = v1 if v1 is not None else ""
    b = v2 if v2 is not None else ""
    return f"<span class='mq-frac'><span class='slot cur'>{a}</span><span class='slot'>{b}</span></span>"

def EXP(base=None, power=None):
    b = base if base is not None else ""
    p = power if power is not None else ""
    return (f"<span class='mq-exp'>"
            f"<span class='slot cur'>{b}</span>"
            f"<span class='slot mq-pow'>{p}</span>"
            f"</span>")

def SQRT(inner=None, idx=None):
    i = inner if inner is not None else ""
    if idx is None:
        return f"<span class='mq-sqrt'>√<span class='slot cur mq-rad'>{i}</span></span>"
    return (f"<span class='mq-sqrt'><span class='slot mq-idx'>{idx}</span>"
            f"√<span class='slot cur mq-rad'>{i}</span></span>")

def SUM(top=None, bot=None):
    t = top if top is not None else ""
    b = bot if bot is not None else ""
    return (f"<span class='mq-sum'>"
            f"<span class='slot mq-top cur'>{t}</span>"
            f"<span class='mq-sig'>Σ</span>"
            f"<span class='slot mq-bot'>{b}</span>"
            f"</span>")

def LOG(base=None):
    b = base if base is not None else ""
    return f"<span class='mq-log'>log<span class='slot cur'>{b}</span>(</span>"

def PAREN(preset=None, second=None):
    p = preset if preset is not None else ""
    s = second if second is not None else ""
    return (f"<span class='mq-paren'>(<span class='slot cur'>{p}</span>"
            f"±<span class='slot'>{s}</span>)</span>")

def PAIR(x=None, y=None):
    a = x if x is not None else ""
    b = y if y is not None else ""
    return f"(<span class='slot cur'>{a}</span>;<span class='slot'>{b}</span>)"

def LOGDIV():
    return ("<span class='mq-frac'>"
            "<span class='slot cur'>sin(□)</span>"
            "<span class='slot'>cos(□)</span>"
            "</span>")

IC_FRAC = "<span class='ic ic-frac'><i></i><i></i></span>"
IC_EXP  = "<span class='ic ic-exp'><i></i><i></i></span>"
IC_SQRT = "<span class='ic ic-sqrt'>&#8730;<i></i></span>"
IC_SUM  = "<span class='ic ic-sum'><i></i>&#8721;<i></i></span>"
IC_LOG  = "<span class='ic ic-log'>log<i></i></span>"

base = {
    "structure": [
        k("a/b",  FRAC(),  IC_FRAC),
        k("□/□",  FRAC()),
        k("x²",   EXP(power="2"),  IC_EXP),
        k("xⁿ",   EXP()),
        k("√",    SQRT(),  IC_SQRT),
        k("ⁿ√",   SQRT(idx="")),
        k("∛",    "<span class='mq-sqrt'><span class='slot mq-idx'>3</span>√<span class='slot cur mq-rad'></span></span>"),
        k("log",  LOG()),
        k("logₐ", LOG(base="")),
        k("π",    "π"),
        k("θ",    "θ"),
        k("∞",    "∞"),
    ],
    "operators": [
        k("+", "+"), k("−", "−"), k("×", "×"), k("÷", "÷"),
        k("=", "="), k("(", "("), k(")", ")"),
        k("<", "<"), k(">", ">"),
        k("≤", "≤"), k("≥", "≥"), k("≠", "≠"),
        k("∴", " ∴ "), k("∵", " ∵ "), k("⇒", " ⇒ "),
    ],
    "digits": [
        k("1"), k("2"), k("3"), k("4"), k("5"), k("6"),
        k("7"), k("8"), k("9"), k("0"), k("."), k("(−)", "−"),
    ],
    "variables": [
        k("x"), k("y"), k("a"), k("b"), k("c"), k("n"), k("k"), k("t"),
    ],
    "actions": [
        k("SPACE", "__space__"),
        k("DEL",   "__del__"),
        k("AC",    "__clear__"),
        k("ABC",   "__abc__"),
        k("⏎",     "__enter__"),
        k("?",     "__ask__"),
        k("✓",     "__done__"),
    ],
    "abc": [
        [k(L) for L in "ABCDEFGHIJ"],
        [k(L) for L in "KLMNOPQRST"],
        [k(L) for L in "UVWXYZ"] + [k(""), k(""), k("")],
        [k("x₁","x₁"), k("y₁","y₁"), k("x₂","x₂"), k("y₂","y₂"),
         k("p","p"), k("q","q"), k("m","m"), k("r","r"), k("s","s"), k("u","u")],
    ],
}

tabs_order = ["Algebra","Sequences","Functions","Calculus","Trig",
              "Finance","Analytical","Euclidean","Stats","Probability"]

trig_default = [
    k("INV", "__inv__"),
    k("sin(","sin("), k("cos(","cos("), k("tan(","tan("),
    k("sin²θ",EXP(base="sin θ",power="2")),
    k("cos²θ",EXP(base="cos θ",power="2")),
]
trig_inverse = [
    k("INV", "__inv__"),
    k("sin⁻¹(","sin⁻¹("), k("cos⁻¹(","cos⁻¹("), k("tan⁻¹(","tan⁻¹("),
    k("sinⁿ□",EXP(base="sin ")),
    k("cosⁿ□",EXP(base="cos ")),
]

tabs = {
    "Algebra": [
        [k("x²",EXP(power="2")), k("y²",EXP(power="2")),
         k("a²",EXP(power="2")), k("b²",EXP(power="2")),
         k("c²",EXP(power="2"))],
        [k("x³",EXP(power="3")), k("xⁿ",EXP()), k("aⁿ",EXP()), k("bⁿ",EXP())],
        [k("(□)²",EXP(base="()",power="2")),
         k("(□)³",EXP(base="()",power="3")),
         k("(□)ⁿ",EXP(base="()"))],
        [k("±","±"),
         k("QF","<span class='mq-qf'>x = "
              "<span class='mq-frac'><span class='slot cur'></span>"
              "<span class='slot'></span></span></span>")],
    ],
    "Sequences": [
        [k("Σ",SUM(),IC_SUM),
         k("Tₙ",EXP(base="T",power="n")),
         k("Sₙ",EXP(base="S",power="n")),
         k("Tₙ₊₁","Tₙ₊₁"), k("Tₙ₋₁","Tₙ₋₁"), k("n→∞","n→∞")],
        [k("□²",EXP(power="2")), k("□³",EXP(power="3")), k("□ⁿ",EXP()),
         k("(□)²",EXP(base="()",power="2")),
         k("(□)³",EXP(base="()",power="3")),
         k("(□)ⁿ",EXP(base="()"))],
        [k("A"), k("P"), k("i"), k("n"), k("r")],
    ],
    "Functions": [
        [k("f(□)","f(<span class='slot cur'></span>)"),
         k("g(□)","g(<span class='slot cur'></span>)"),
         k("f⁻¹(□)","f⁻¹(<span class='slot cur'></span>)"),
         k("→","→"), k("∞","∞"), k("Δ","Δ")],
        [k("sin(□)","sin(<span class='slot cur'></span>)"),
         k("cos(□)","cos(<span class='slot cur'></span>)"),
         k("tan(□)","tan(<span class='slot cur'></span>)")],
        [k("a/b",FRAC()), k("□ⁿ",EXP())],
    ],
    "Calculus": [
        [k("dy/dx","dy/dx"), k("d²y/dx²","d²y/dx²"),
         k("dⁿy/dxⁿ","dⁿy/dxⁿ"), k("∫","∫"), k("lim","lim"), k("→","→")],
        [k("□²",EXP(power="2")), k("□³",EXP(power="3")), k("□ⁿ",EXP()),
         k("Δx","Δx"),
         k("1/□","<span class='mq-frac'><span class='slot cur'>1</span>"
                 "<span class='slot'></span></span>")],
    ],
    "Trig": [
        trig_default,
        [k("α","α"), k("β","β"), k("θ","θ"), k("2α","2α"), k("°","°"), k("π","π")],
        [k("(□±□)",PAREN()), k("(90°±θ)",PAREN(preset="90°",second="θ")),
         k("(□±θ)",PAREN(second="θ")),
         k("(180°±θ)",PAREN(preset="180°",second="θ")),
         k("(360°±θ)",PAREN(preset="360°",second="θ")),
         k("(−θ)","(−θ)")],
        [k("sin²θ",EXP(base="sin θ",power="2")),
         k("sinⁿ□",EXP(base="sin ")),
         k("cos²θ",EXP(base="cos θ",power="2")),
         k("cosⁿ□",EXP(base="cos "))],
        [k("tan²θ",EXP(base="tan θ",power="2")),
         k("tanⁿ□",EXP(base="tan ")),
         k("sin□/cos□",LOGDIV())],
    ],
    "Finance": [
        [k("A"), k("P"), k("i"), k("n"), k("x")],
        [k("(1+i)ⁿ",EXP(base="1+i")), k("(□)ⁿ",EXP(base="()")),
         k("(1−i)ⁿ",EXP(base="1−i"))],
        [k("□ⁿ",EXP()), k("□/□",FRAC())],
    ],
    "Analytical": [
        [k("(□;□)",PAIR()), k("x₁","x₁"), k("y₁","y₁"), k("x₂","x₂"), k("y₂","y₂")],
        [k("√",SQRT()), k("□²",EXP(power="2")),
         k("tan(□)","tan(<span class='slot cur'></span>)")],
        [k("θ","θ"), k("°","°")],
        [k("(□)²+(□)²",
           "<span class='mq-exp'><span class='slot cur'></span>"
           "<span class='slot mq-pow'>2</span></span>+"
           "<span class='mq-exp'><span class='slot'></span>"
           "<span class='slot mq-pow'>2</span></span>")],
    ],
    "Euclidean": [
        [k("△","△"), k("∥","∥"), k("⊥","⊥"),
         k("∠","∠"), k("≅","≅"), k("∼","∼")],
        [k("Â","Â"), k("B̂","B̂"), k("Ĉ","Ĉ"), k("D̂","D̂")],
        [k("sin(","sin("), k("cos(","cos("), k("tan(","tan(")],
        [k("√",SQRT()), k("∴"," ∴ "), k("∵"," ∵ ")],
    ],
    "Stats": [
        [k("x̄","x̄"), k("σ","σ"), k("n","n"), k("r","r"),
         k("r²",EXP(base="r",power="2"))],
        [k("Σ",SUM()), k("□²",EXP(power="2")), k("√",SQRT())],
        [k("E(□)","E(<span class='slot cur'></span>)"),
         k("P(□)","P(<span class='slot cur'></span>)")],
    ],
    "Probability": [
        [k("P(□)","P(<span class='slot cur'></span>)"),
         k("P(□∪□)","P(<span class='slot cur'></span>∪<span class='slot'></span>)"),
         k("P(□∩□)","P(<span class='slot cur'></span>∩<span class='slot'></span>)"),
         k("P(□')","P(<span class='slot cur'></span>')")],
        [k("∪","∪"), k("∩","∩"), k("'","'")],
        [k("nCr","nCr"), k("nPr","nPr"), k("!","!")],
    ],
}

payload = {
    "schema_version": 18.0,
    "base": base,
    "tabs_order": tabs_order,
    "tabs": tabs,
    "trig_default": trig_default,
    "trig_inverse": trig_inverse,
}
LAYOUT_PATH.write_text(json.dumps(payload, indent=2, ensure_ascii=False),
                       encoding="utf-8")
print(f"  ✓ {LAYOUT_PATH.name} written")

HTML_TPL = r"""<!doctype html>
<html lang="en"><head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width,initial-scale=1,maximum-scale=1">
<title>MethodMarker Keyboard</title>
<style>
:root{
  --bg:#eef2f8; --card:#fff; --line:#e4e9f2; --ink:#0b1f44;
  --ink2:#465a80; --ink3:#9aa8c2; --blue:#3b5bf6; --blue2:#5b8cf6;
  --amber:#f0ad4e;
  --key-h: clamp(30px, 5.2vw, 40px);
  --key-f: clamp(10px, 1.4vw, 13px);
  --pad: clamp(4px, 0.9vw, 8px);
}
*{box-sizing:border-box;-webkit-tap-highlight-color:transparent}
html,body{margin:0;height:100%}
body{
  background:var(--bg);color:var(--ink);
  font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,sans-serif;
  display:flex;justify-content:center;padding:6px;
  min-height:100vh;overflow:hidden;
}
.app{
  width:100%;max-width:720px;background:var(--card);
  border-radius:16px;overflow:hidden;
  box-shadow:0 8px 32px rgba(20,40,90,.12);
  display:flex;flex-direction:column;
  height:calc(100vh - 12px);
}
.topbar{display:flex;align-items:center;gap:8px;
  padding:6px 12px;background:#f6f8fd;border-bottom:1px solid var(--line);flex:none}
.logo{width:22px;height:22px;border-radius:6px;
  background:linear-gradient(135deg,var(--blue),var(--blue2));
  display:flex;align-items:center;justify-content:center;
  color:#fff;font-weight:700;font-size:12px;flex:none}
.brand{font-weight:700;font-size:12px;color:var(--ink)}
.tag{font-size:9px;color:#6b7a99}

.work{background:#fff;padding:6px 12px 8px;flex:none;
  max-height:clamp(80px,16vh,140px);overflow-y:auto}
.work-head{display:flex;justify-content:space-between;align-items:center;
  font-size:10px;color:var(--ink2);font-weight:600;margin-bottom:4px;
  position:sticky;top:0;background:#fff;padding-bottom:2px}
.work-head .hint{color:var(--ink3);font-weight:500;font-size:9px}
.lines{display:flex;flex-direction:column;gap:1px}
.lrow{display:grid;grid-template-columns:20px 1fr;column-gap:5px;align-items:flex-start}
.lnum{color:#b8c2d6;font-size:9px;text-align:right;padding-top:7px;
  user-select:none;font-variant-numeric:tabular-nums}
.lbody{min-height:26px;padding:4px 6px;border-radius:5px;
  font-size:clamp(14px,1.9vw,17px);line-height:1.35;color:var(--ink);
  outline:none;font-family:'Cambria Math','Latin Modern Math',Georgia,serif;
  white-space:pre-wrap;word-break:break-word;
  transition:background .1s;caret-color:var(--blue)}
.lbody:focus{background:#f5f8ff}

.sbc{padding:5px 12px 7px;background:#fff;border-top:1px dashed var(--line);flex:none}
.sbc-head{display:flex;align-items:center;gap:5px;font-size:10px;
  color:var(--ink2);font-weight:600;margin-bottom:4px}
.sbc-head .dot{width:5px;height:5px;border-radius:50%;background:var(--amber)}
#sbc{width:100%;height:26px;resize:none;font-family:inherit;
  font-size:11px;padding:5px 8px;border:1px solid var(--line);
  border-radius:6px;background:#fffdf7;color:var(--ink);outline:none;
  transition:height .2s ease,box-shadow .15s}
#sbc:focus{height:60px;box-shadow:0 0 0 3px rgba(240,173,78,.15)}

.tabs{display:flex;overflow-x:auto;gap:3px;padding:4px 8px;
  background:#f6f8fd;border-top:1px solid var(--line);
  border-bottom:1px solid var(--line);scrollbar-width:none;flex:none}
.tabs::-webkit-scrollbar{display:none}
.tab{flex:none;padding:4px 9px;border-radius:12px;font-size:10px;
  font-weight:600;color:var(--ink2);background:transparent;
  border:none;cursor:pointer;white-space:nowrap;font-family:inherit}
.tab.active{background:var(--blue);color:#fff}

.kb-scroll{flex:1 1 auto;overflow-y:auto;background:#fff;display:flex;flex-direction:column}
.kb{padding:var(--pad);display:flex;flex-direction:column;gap:4px}
.row{display:grid;gap:4px}
.key{padding:2px;border-radius:7px;background:#fff;
  border:1px solid #dfe5ef;color:var(--ink);
  font-size:var(--key-f);cursor:pointer;text-align:center;
  user-select:none;font-family:inherit;min-height:var(--key-h);
  display:flex;align-items:center;justify-content:center;
  transition:transform .05s,background .1s;overflow:hidden}
.key:active{transform:scale(.95);background:#eef2fb}
.key.inv-active{background:var(--amber);color:#111;border-color:var(--amber)}
.acts{display:grid;grid-template-columns:repeat(7,1fr);gap:4px;
  padding:5px var(--pad) 6px;background:#f6f8fd;
  border-top:1px solid var(--line);flex:none}
.act{padding:6px 2px;border-radius:7px;background:#eef2fb;
  border:1px solid #dfe5ef;color:#223a6b;
  font-size:10px;font-weight:700;cursor:pointer;text-align:center;
  user-select:none;min-height:var(--key-h);
  display:flex;align-items:center;justify-content:center;
  font-family:inherit;transition:transform .05s}
.act.done{background:var(--blue);color:#fff;border-color:var(--blue)}
.act.active-mode{background:var(--blue);color:#fff;border-color:var(--blue)}
.act:active{transform:scale(.97)}

.ic{display:inline-flex;align-items:center;justify-content:center}
.ic-frac{flex-direction:column;gap:1px}
.ic-frac i{display:block;width:12px;height:5px;
  border:1px solid #7b8ba6;border-radius:2px}
.ic-frac i:nth-child(2){width:16px;height:1px;border:none;background:#7b8ba6;border-radius:0}
.ic-exp{position:relative;width:20px;height:18px}
.ic-exp i:nth-child(1){position:absolute;bottom:0;left:0;
  width:12px;height:12px;border:1px solid #7b8ba6;border-radius:2px}
.ic-exp i:nth-child(2){position:absolute;top:0;right:0;
  width:7px;height:7px;border:1px solid #7b8ba6;border-radius:2px}
.ic-sqrt{font-size:14px;font-weight:700;line-height:1}
.ic-sqrt i{display:inline-block;width:8px;height:10px;
  border-top:1px solid #7b8ba6;border-right:1px solid #7b8ba6;
  border-bottom:1px solid #7b8ba6;margin-left:-1px}
.ic-sum{flex-direction:column;font-size:11px;line-height:1}
.ic-sum i{display:block;width:12px;height:4px;
  border:1px solid #7b8ba6;border-radius:1px;margin:1px 0}
.ic-log{font-size:10px;font-weight:700}
.ic-log i{display:inline-block;width:7px;height:5px;
  border:1px solid #7b8ba6;border-radius:2px;margin-left:1px}

.slot{display:inline-block;min-width:10px;min-height:16px;
  padding:0 2px;vertical-align:middle;text-align:center;outline:none}
.slot:focus{background:#e8efff;border-radius:3px;
  box-shadow:0 0 0 1px rgba(59,91,246,.35)}

/* fraction */
.mq-frac{display:inline-flex;flex-direction:column;align-items:center;
  vertical-align:middle;padding:0 3px;margin:0 2px;line-height:1.1}
.mq-frac .slot:first-child{border-bottom:1.6px solid var(--ink);
  min-width:16px;padding:1px 5px}
.mq-frac .slot:last-child{padding-top:2px;min-width:16px;
  padding-left:5px;padding-right:5px}

/* exponent */
.mq-exp{display:inline-flex;align-items:flex-start;vertical-align:middle;padding:0 2px}
.mq-exp .slot{min-width:10px}
.mq-exp .mq-pow{font-size:.62em;position:relative;top:-.55em;
  padding-left:1px;min-width:10px}

/* square root — simple shape, no growth */
.mq-sqrt{display:inline-flex;align-items:center;vertical-align:middle;
  line-height:1;padding:0 3px}
.mq-sqrt .mq-rad{border-top:1.6px solid var(--ink);
  border-right:1.6px solid var(--ink);border-bottom:1.6px solid var(--ink);
  min-width:14px;min-height:20px;padding:2px 5px;
  display:inline-flex;align-items:center}
.mq-sqrt .mq-idx{font-size:.62em;position:relative;top:-.55em;min-width:9px}

.mq-sum{display:inline-flex;flex-direction:column;align-items:center;
  vertical-align:middle;padding:0 5px;margin:0 2px;line-height:1}
.mq-sum .mq-top{font-size:.6em;min-width:20px;text-align:center;
  padding:1px 2px;min-height:12px}
.mq-sum .mq-sig{font-size:1.15em;line-height:1;margin:1px 0}
.mq-sum .mq-bot{font-size:.6em;min-width:20px;text-align:center;
  padding:1px 2px;min-height:12px}

.mq-log{display:inline-flex;align-items:flex-end;vertical-align:middle;padding:0 2px}
.mq-log .slot{font-size:.62em;min-width:10px;min-height:11px;text-align:center}

.mq-paren{display:inline-flex;align-items:center;padding:0 2px;
  font-family:'Cambria Math',Georgia,serif}
.mq-qf{display:inline-flex;align-items:center;padding:0 3px;
  font-family:'Cambria Math',Georgia,serif}

.result{padding:6px 12px;background:var(--ink);color:#fff;
  font-family:'SF Mono',Menlo,monospace;font-size:9px;line-height:1.4;
  max-height:0;overflow:hidden;transition:max-height .25s ease;flex:none}
.result.shown{max-height:180px;overflow-y:auto}
.result .label{color:#9ab;margin-bottom:3px}
.result pre{margin:0;color:#cfe2ff;white-space:pre-wrap}

@media (max-width:480px){
  .work{max-height:clamp(70px,14vh,110px)}
  .lbody{font-size:14px}
  .kb{padding:4px}
  .row{gap:3px}
  .key{min-height:32px;border-radius:6px}
  .act{font-size:9px;padding:4px 1px;min-height:32px}
  .topbar{padding:5px 10px}
  .tab{font-size:9px;padding:3px 8px}
}
</style></head><body>

<div class="app">
  <div class="topbar">
    <div class="logo">M</div>
    <div><div class="brand">MethodMarker</div>
    <div class="tag">Better Methods. Higher Marks.</div></div>
  </div>

  <div class="work">
    <div class="work-head">
      <span>&#9998; Working</span>
      <span class="hint">tap a line · ⏎ new step · ⌫ at start merges up</span>
    </div>
    <div class="lines" id="lines"></div>
  </div>

  <div class="sbc">
    <div class="sbc-head"><span class="dot"></span>
      Ask Tanaka — where did you get stuck?</div>
    <textarea id="sbc" rows="1"
      placeholder="e.g. I did step 2 but I don't understand why the sign flips"></textarea>
  </div>

  <div class="tabs" id="tabs"></div>
  <div class="kb-scroll"><div class="kb" id="kb"></div></div>
  <div class="acts" id="acts"></div>
  <div class="result" id="result"></div>
</div>

<script>
const LAYOUT = __LAYOUT__;
const state = { tab: LAYOUT.tabs_order[0], abcMode: false, invMode: false };

const linesEl = document.getElementById('lines');
function addLine(html, focus){
  const row = document.createElement('div'); row.className='lrow';
  const n = document.createElement('div'); n.className='lnum';
  const b = document.createElement('div'); b.className='lbody';
  b.contentEditable='true'; b.spellcheck=false;
  if (html) b.innerHTML = html;
  row.appendChild(n); row.appendChild(b);
  linesEl.appendChild(row);
  renumber();
  if (focus) focusStart(b);
  return b;
}
function renumber(){
  [...linesEl.querySelectorAll('.lrow')].forEach((r,i)=>{
    r.querySelector('.lnum').textContent = i+1;
  });
}
function focusStart(el){
  el.focus();
  const first = el.querySelector('.slot');
  const r = document.createRange();
  if (first) r.selectNodeContents(first);
  else { r.selectNodeContents(el); r.collapse(false); }
  r.collapse(true);
  const s = window.getSelection(); s.removeAllRanges(); s.addRange(r);
}
function activeLine(){
  const sel = window.getSelection();
  if (!sel.rangeCount) return null;
  let n = sel.getRangeAt(0).startContainer;
  while (n && n !== linesEl){
    if (n.classList && n.classList.contains('lbody')) return n;
    n = n.parentNode;
  }
  return null;
}
function currentSlot(){
  const sel = window.getSelection();
  if (!sel.rangeCount) return null;
  let n = sel.getRangeAt(0).startContainer;
  if (n.nodeType === 3) n = n.parentNode;
  return n.closest ? n.closest('.slot') : null;
}

function doBackspace(){
  let line = activeLine();
  if (!line){
    const all = linesEl.querySelectorAll('.lbody');
    if (all.length){
      line = all[all.length-1];
      line.focus();
      const r = document.createRange();
      r.selectNodeContents(line); r.collapse(false);
      const s = window.getSelection();
      s.removeAllRanges(); s.addRange(r);
    } else return;
  }
  const sel = window.getSelection();
  if (!sel.rangeCount) return;
  const r = sel.getRangeAt(0);
  if (!r.collapsed){ r.deleteContents(); return; }

  const slot = currentSlot();
  if (slot){
    const slotText = slot.textContent.replace(/\u200B/g,'');
    const probe = r.cloneRange(); probe.selectNodeContents(slot);
    probe.setEnd(r.startContainer, r.startOffset);
    const before = probe.toString().replace(/\u200B/g,'');
    if (slotText.length === 0 || before.length === 0){
      const struct = slot.closest(
        '.mq-frac,.mq-exp,.mq-sqrt,.mq-sum,.mq-log,.mq-paren,.mq-qf'
      ) || slot;
      const after = document.createRange();
      after.setStartAfter(struct); after.collapse(true);
      struct.remove();
      sel.removeAllRanges(); sel.addRange(after);
      return;
    }
    deleteCharBefore(r); return;
  }

  const probe = r.cloneRange(); probe.selectNodeContents(line);
  probe.setEnd(r.startContainer, r.startOffset);
  if (probe.toString().length === 0){
    const prev = line.closest('.lrow').previousElementSibling;
    if (prev){
      const pb = prev.querySelector('.lbody');
      while (line.firstChild) pb.appendChild(line.firstChild);
      line.closest('.lrow').remove();
      focusStart(pb);
      renumber();
      return;
    }
  }
  deleteCharBefore(r);
}
function deleteCharBefore(r){
  const node = r.startContainer;
  const offset = r.startOffset;
  const sel = window.getSelection();
  if (node.nodeType === 3 && offset > 0){
    node.deleteData(offset - 1, 1);
    const nr = document.createRange();
    nr.setStart(node, offset - 1); nr.collapse(true);
    sel.removeAllRanges(); sel.addRange(nr);
    return;
  }
  if (node.nodeType === 3 && offset === 0){
    let prev = node.previousSibling;
    while (prev && prev.nodeType !== 3 && !prev.textContent){
      prev = prev.previousSibling;
    }
    if (prev){
      const after = document.createRange();
      after.setStartAfter(prev); after.collapse(true);
      prev.remove();
      sel.removeAllRanges(); sel.addRange(after);
    }
    return;
  }
  if (node.nodeType === 1){
    let leaf = node;
    while (leaf.lastChild && leaf.nodeType === 1) leaf = leaf.lastChild;
    if (leaf.nodeType === 3){
      const nr = document.createRange();
      nr.setStart(leaf, Math.max(0, leaf.textContent.length - 1));
      nr.collapse(true);
      sel.removeAllRanges(); sel.addRange(nr);
      doBackspace();
    } else if (leaf.previousSibling){
      const after = document.createRange();
      after.setStartAfter(leaf.previousSibling); after.collapse(true);
      leaf.previousSibling.remove();
      sel.removeAllRanges(); sel.addRange(after);
    }
  }
}

linesEl.addEventListener('keydown', e => {
  const line = e.target.closest('.lbody'); if (!line) return;
  if (e.key === 'Enter' && !e.shiftKey){
    e.preventDefault();
    const sel = window.getSelection();
    const r = sel.getRangeAt(0);
    const after = r.cloneRange();
    after.selectNodeContents(line);
    after.setStart(r.endContainer, r.endOffset);
    const rest = after.extractContents();
    const nb = addLine('', false);
    while (rest.firstChild) nb.insertBefore(rest.firstChild, nb.firstChild);
    focusStart(nb);
    return;
  }
  if (e.key === 'Tab'){
    e.preventDefault();
    const slot = currentSlot();
    if (!slot){
      const r = document.createRange();
      r.selectNodeContents(line); r.collapse(false);
      const sel = window.getSelection();
      sel.removeAllRanges(); sel.addRange(r);
      return;
    }
    const struct = slot.closest('.mq-frac,.mq-exp,.mq-sqrt,.mq-sum,.mq-log,.mq-paren,.mq-qf');
    if (!struct) return;
    const slots = [...struct.querySelectorAll('.slot')];
    const i = slots.indexOf(slot);
    const next = slots[i+1];
    const sel = window.getSelection();
    if (next){
      const r = document.createRange(); r.selectNodeContents(next); r.collapse(true);
      sel.removeAllRanges(); sel.addRange(r);
    } else {
      const r = document.createRange(); r.setStartAfter(struct); r.collapse(true);
      sel.removeAllRanges(); sel.addRange(r);
    }
    return;
  }
  if (e.key === 'ArrowRight'){
    const slot = currentSlot();
    if (!slot) return;
    const struct = slot.closest('.mq-frac,.mq-exp,.mq-sqrt,.mq-sum,.mq-log,.mq-paren,.mq-qf');
    if (!struct) return;
    const slots = [...struct.querySelectorAll('.slot')];
    if (slots.indexOf(slot) !== slots.length - 1) return;
    e.preventDefault();
    const r = document.createRange();
    r.setStartAfter(struct); r.collapse(true);
    const sel = window.getSelection();
    sel.removeAllRanges(); sel.addRange(r);
    return;
  }
  if (e.key === 'Backspace'){ e.preventDefault(); doBackspace(); }
});

function insertAtCursor(html){
  let line = activeLine();
  if (!line){
    const all = linesEl.querySelectorAll('.lbody');
    line = all.length ? all[all.length-1] : addLine('', true);
    line.focus();
  }
  line.focus();
  const sel = window.getSelection();
  if (!sel.rangeCount) return;
  const r = sel.getRangeAt(0);
  r.deleteContents();

  const tmp = document.createElement('span');
  tmp.innerHTML = html;
  const nodes = [...tmp.childNodes];
  if (!nodes.length) return;

  const inside = currentSlot();
  if (inside){
    while (tmp.firstChild) inside.appendChild(tmp.firstChild);
    const nested = inside.querySelector('.slot.cur') || inside.querySelector('.slot');
    if (nested){
      const nr = document.createRange(); nr.selectNodeContents(nested); nr.collapse(true);
      sel.removeAllRanges(); sel.addRange(nr);
    }
    return;
  }
  nodes.forEach(n => r.insertNode(n));
  const last = nodes[nodes.length - 1];
  let target = null;
  if (last.nodeType === 1){
    target = last.querySelector('.slot.cur') || last.querySelector('.slot');
  }
  if (target){
    const nr = document.createRange();
    nr.selectNodeContents(target); nr.collapse(true);
    sel.removeAllRanges(); sel.addRange(nr);
  } else {
    const nr = document.createRange();
    nr.setStartAfter(last); nr.collapse(true);
    sel.removeAllRanges(); sel.addRange(nr);
  }
}

function handleKey(key){
  if (key.i === '__abc__'){
    state.abcMode = !state.abcMode;
    document.querySelectorAll('.act').forEach(a => {
      if (a.textContent === 'ABC') a.classList.toggle('active-mode', state.abcMode);
    });
    renderKb();
    return;
  }
  if (key.i === '__inv__'){
    state.invMode = !state.invMode;
    renderKb();
    return;
  }
  if (key.i === '__space__'){ insertAtCursor(" "); return; }
  if (key.i === '__del__'){ doBackspace(); return; }
  if (key.i === '__clear__') { linesEl.innerHTML=''; addLine('', true); return; }
  if (key.i === '__enter__') { addLine('', true); return; }
  if (key.i === '__ask__')   { document.getElementById('sbc').focus(); return; }
  if (key.i === '__done__')  { submit(); return; }
  insertAtCursor(key.i);
}

const tabsEl = document.getElementById('tabs');
LAYOUT.tabs_order.forEach(name => {
  const b = document.createElement('button');
  b.className = 'tab' + (name === state.tab ? ' active' : '');
  b.textContent = name;
  b.onclick = () => {
    state.tab = name;
    state.abcMode = false;
    state.invMode = false;
    document.querySelectorAll('.tab').forEach(t =>
      t.classList.toggle('active', t.textContent === name));
    document.querySelectorAll('.act').forEach(a => {
      if (a.textContent === 'ABC') a.classList.remove('active-mode');
    });
    renderKb();
  };
  tabsEl.appendChild(b);
});

const kbEl = document.getElementById('kb');
function keyBtn(key){
  const el = document.createElement('div');
  el.className = 'key';
  if (key.i === '__inv__' && state.invMode) el.classList.add('inv-active');
  if (key.ic) el.innerHTML = key.ic;
  else el.textContent = key.l;
  el.onclick = () => handleKey(key);
  return el;
}
function renderRow(keys){
  const r = document.createElement('div'); r.className='row';
  r.style.gridTemplateColumns = `repeat(${keys.length},1fr)`;
  keys.forEach(k => r.appendChild(keyBtn(k)));
  kbEl.appendChild(r);
}
function renderKb(){
  kbEl.innerHTML = '';
  if (state.abcMode){
    (LAYOUT.base.abc || []).forEach(row => renderRow(row));
    return;
  }
  renderRow(LAYOUT.base.structure);
  renderRow(LAYOUT.base.operators);
  renderRow(LAYOUT.base.digits);
  renderRow(LAYOUT.base.variables);
  const rows = LAYOUT.tabs[state.tab] || [];
  rows.forEach(row => {
    if (state.tab === 'Trig' && row === LAYOUT.trig_default){
      renderRow(state.invMode ? LAYOUT.trig_inverse : LAYOUT.trig_default);
    } else {
      renderRow(row);
    }
  });
}

const actsEl = document.getElementById('acts');
LAYOUT.base.actions.forEach(k => {
  const el = document.createElement('div');
  el.className = 'act' + (k.i === '__done__' ? ' done' : '');
  el.textContent = k.l;
  el.onclick = () => handleKey(k);
  actsEl.appendChild(el);
});

const resultEl = document.getElementById('result');
function submit(){
  const steps = [...linesEl.querySelectorAll('.lbody')]
    .map(b => b.innerText.replace(/\u200B/g,'').trim())
    .filter(s => s.length);
  const payload = {
    steps:        steps,
    step_count:   steps.length,
    sbc_message:  document.getElementById('sbc').value.trim(),
    tab_used:     state.tab,
  };
  resultEl.classList.add('shown');
  resultEl.innerHTML =
    "<div class='label'>submitted to step_grader + intervention log</div>"+
    "<pre>"+JSON.stringify(payload,null,2)+"</pre>";
}

addLine('', true);
renderKb();
</script>
</body></html>
"""

html = HTML_TPL.replace("__LAYOUT__", json.dumps(payload, ensure_ascii=False))
HTML_PATH.write_text(html, encoding="utf-8")
print(f"  ✓ {HTML_PATH.name} written ({HTML_PATH.stat().st_size:,} bytes)")
webbrowser.open(HTML_PATH.as_uri())
print("  → opened in browser")
print()
print("  v18: keyboard restored to working state")
````
